<div class="part-header"><h2>1. Theory</h2></div><h3 class="sub">🔹 1.1 &nbsp;What the GIL is</h3><div class="theory-box">The <strong>GIL</strong> (Global Interpreter Lock) is a single mutex in CPython that allows <strong>only one thread to execute Python bytecode at a time</strong> — even on a multi-core machine. It exists because CPython's memory management (reference counting) isn't thread-safe; one global lock makes the interpreter simple and fast for single-threaded code, at the cost of true multi-threaded parallelism.</div><div class="info-box">💡 The GIL is a <strong>CPython implementation detail</strong>, not part of the Python language — Jython/IronPython don't have it, and CPython 3.13+ ships an experimental "free-threaded" (no-GIL) build. But for the CPython you almost certainly use, the GIL is real and shapes every threading decision.</div>

<h3 class="sub">🔹 1.2 &nbsp;Concurrency vs parallelism</h3><div class="theory-box">These are different ideas. <strong>Concurrency</strong> is <em>managing</em> many tasks by interleaving them (making progress on several, not necessarily at the same instant). <strong>Parallelism</strong> is <em>actually running</em> multiple tasks at the same instant on multiple cores. The GIL blocks parallelism of Python <strong>bytecode</strong> within one process — you can be concurrent, but not truly parallel with threads.</div><div class="mut-grid"><div class="mut-card mc-imm"><div class="mc-title">🔀 Concurrency</div><div class="mc-body">interleave many tasks<br>one at a time, switching<br>threads / async<br>great for waiting (I/O)</div></div><div class="mut-card mc-mut"><div class="mc-title">⚡ Parallelism</div><div class="mc-body">run simultaneously<br>multiple CPU cores<br>processes<br>great for computing (CPU)</div></div></div><div class="info-box">💡 <strong>SQL / Power BI anchor:</strong> concurrency is like one analyst juggling several queries, switching whenever one is waiting on the database; parallelism is several analysts each on their own machine running queries at the same time.</div>

<h3 class="sub">🔹 1.4 &nbsp;The crucial nuance — I/O-bound threading <em>does</em> speed up</h3><div class="theory-box">The GIL is <strong>released</strong> during blocking I/O (network, disk, <code>time.sleep</code>) and inside many C extensions (NumPy heavy math). Rule of thumb: <strong>I/O-bound → threads / async; CPU-bound → processes.</strong> The next chunks build each tool on this split.</div><hr class="divider"><div style="background:#1e1e2e; border-left:4px solid #cba6f7; padding:16px 20px; border-radius:8px; font-family:monospace; color:#cdd6f4;"><h4 style="color:#cba6f7; margin:0 0 12px 0;">🔑 Chunk A — Key Takeaways</h4><ul style="margin:0; padding-left:20px; line-height:2.1"><li>The GIL lets only one thread run Python bytecode at a time (a CPython detail, for safe refcounting)</li><li>Concurrency = interleaving tasks; parallelism = running at once. The GIL blocks bytecode parallelism</li><li>CPU-bound threading = no speedup (measured 3252 → 3486 ms) — use processes</li><li>I/O-bound threading = real speedup (604 → 313 ms) — the GIL releases while waiting</li></ul></div>

<div class="chunk-badge">Part 1 · Chunk B — Threading for I/O-bound Work</div><h3 class="sub">🔹 1.5 &nbsp;<code>threading</code> and <code>ThreadPoolExecutor</code></h3><div class="theory-box">For I/O-bound work, threads overlap the waiting. The low-level API is <code>threading.Thread(target, args)</code> + <code>start()</code>/<code>join()</code>; the high-level, preferred one is <code>concurrent.futures.ThreadPoolExecutor</code> — a worker pool with <code>submit()</code>/<code>map()</code> that returns results and handles lifecycle for you.</div><div class="code-block"><span class="ck">import</span> time<br><span class="ck">from</span> concurrent.futures <span class="ck">import</span> ThreadPoolExecutor<br><br><span class="ck">def</span> <span class="cm">fetch</span>(x):<br>&nbsp;&nbsp;&nbsp;&nbsp;time.<span class="cm">sleep</span>(<span class="cn">0.2</span>)   <span class="cc"># simulates a network/disk wait (releases the GIL)</span><br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">return</span> x * <span class="cn">2</span><br><br><span class="ck">with</span> ThreadPoolExecutor(max_workers=<span class="cn">5</span>) <span class="ck">as</span> ex:<br>&nbsp;&nbsp;&nbsp;&nbsp;results = <span class="cm">list</span>(ex.<span class="cm">map</span>(fetch, <span class="cm">range</span>(<span class="cn">5</span>)))   <span class="cc"># [0, 2, 4, 6, 8]</span></div><div class="output-block">[0, 2, 4, 6, 8] | 208 ms   (5 × 0.2s sequential would be ~1000 ms)</div><div class="why-box"><strong>Why it matters:</strong> five 0.2s waits finish in ~0.2s because the threads wait concurrently — the GIL is released during each <code>sleep</code>/I/O call. This is the pattern for concurrent API calls, file reads, DB queries. <code>ThreadPoolExecutor</code> is almost always cleaner than raw <code>Thread</code> objects.</div>

<div class="info-box"><strong>Part 2:</strong> Golden Rules → Common Traps → Exercise</div><div class="part-header"><h2>4. Golden Rules</h2></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 1</span> Match the tool to the workload.</div><div class="body-txt"><strong>CPU-bound</strong> → <code>multiprocessing</code> (or C libs that release the GIL). <strong>I/O-bound</strong> → threads or <code>asyncio</code>.</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 2</span> Don't expect threads to speed up pure-Python CPU work.</div><div class="body-txt">The GIL serializes bytecode — threads help by <em>overlapping waits</em>, not by using more cores.</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 3</span> Guard <code>multiprocessing</code> with <code>if __name__ == "__main__":</code>.</div><div class="body-txt">On Windows/spawn the child re-imports the module; without the guard you get infinite process spawning. Worker functions must be top-level (picklable).</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 4</span> Prefer <code>concurrent.futures</code> over manual threads/processes.</div><div class="body-txt"><code>ThreadPoolExecutor</code> / <code>ProcessPoolExecutor</code> handle pooling, results, and cleanup; switching between them is a one-line change.</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 5</span> Protect shared mutable state with a <code>Lock</code>.</div><div class="body-txt">Even under the GIL, <code>counter += 1</code> isn't atomic across threads (read-modify-write). Or avoid sharing — pass data via a <code>queue.Queue</code>.</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 6</span> Never block the asyncio event loop.</div><div class="body-txt">Use <code>await asyncio.sleep(...)</code>, not <code>time.sleep</code>; offload CPU work with <code>loop.run_in_executor(pool, ...)</code>.</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 7</span> Remember NumPy/pandas/PyTorch release the GIL in C.</div><div class="body-txt">So threads (and internal BLAS threads) <em>do</em> parallelize vectorized ops — which is why the data stack scales despite the GIL.</div></div>

<div class="trap-header"><h2>5. Common Traps</h2></div><div class="trap-block"><div class="trap-title"><span class="trap-badge">Trap 1</span> Threading a CPU-bound loop for "speed".</div><div class="body-txt">No speedup (often slower from contention). <strong>Fix:</strong> <code>multiprocessing</code>.</div></div><div class="trap-block"><div class="trap-title"><span class="trap-badge">Trap 2</span> <code>multiprocessing</code> without the <code>__main__</code> guard.</div><div class="body-txt">Infinite spawning / <code>RuntimeError</code> on Windows. <strong>Fix:</strong> guard the entry point; top-level worker functions.</div></div><div class="trap-block"><div class="trap-title"><span class="trap-badge">Trap 3</span> Blocking the event loop.</div><div class="body-txt"><code>time.sleep</code> / a CPU loop inside a coroutine freezes <em>all</em> tasks. <strong>Fix:</strong> <code>asyncio.sleep</code> / <code>run_in_executor</code>.</div></div><div class="trap-block"><div class="trap-title"><span class="trap-badge">Trap 4</span> Race condition on shared state.</div><div class="body-txt"><code>counter += 1</code> from many threads loses updates. <strong>Fix:</strong> a <code>Lock</code> around the read-modify-write.</div></div><div class="trap-block"><div class="trap-title"><span class="trap-badge">Trap 5</span> Assuming processes share memory.</div><div class="body-txt">They don't — separate address spaces. <strong>Fix:</strong> pass data via <code>Queue</code>/pipes/shared memory; expect pickling overhead.</div></div><div class="trap-block"><div class="trap-title"><span class="trap-badge">Trap 6</span> Deadlock from inconsistent lock ordering.</div><div class="body-txt">Two threads grab locks A,B in opposite orders → hang. <strong>Fix:</strong> one global lock order, or message-passing.</div></div>

<div class="interview-header"><h2>8. Interview Questions</h2></div><div class="sub-header"><h3>8a — Conceptual Q&amp;A</h3></div><div class="qa-block"><div class="qa-q"><span class="q-num">1</span> What is the GIL?</div><div class="qa-a">The Global Interpreter Lock — a CPython mutex that lets only one thread execute Python bytecode at a time, so threads don't run Python code on multiple cores in parallel.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">2</span> Why does the GIL exist?</div><div class="qa-a">It simplifies CPython's memory management (reference counting isn't thread-safe without it) and keeps single-threaded code fast and C-extension integration simple.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">3</span> CPU-bound vs I/O-bound — which tool?</div><div class="qa-a">CPU-bound → <code>multiprocessing</code> (real parallelism). I/O-bound → threads or <code>asyncio</code> (overlap waits).</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">4</span> Do threads make CPU-bound Python faster?</div><div class="qa-a">No — the GIL serializes bytecode, so pure-Python CPU work sees no speedup (often slightly slower from lock overhead).</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">5</span> Why does <code>multiprocessing</code> help where threads don't?</div><div class="qa-a">Each process has its own interpreter and its own GIL, so they truly run in parallel on multiple cores. Cost: process spawn + pickling data across boundaries.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">6</span> Why the <code>if __name__ == "__main__"</code> guard with multiprocessing?</div><div class="qa-a">On spawn (Windows/macOS default), each child re-imports the module; without the guard, module-level process creation runs again → infinite spawning.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">7</span> <code>asyncio</code> vs threads for I/O?</div><div class="qa-a">Both overlap I/O. Threads are preemptive and simpler for a few tasks; <code>asyncio</code> is single-threaded cooperative scheduling that scales to thousands of concurrent connections with less overhead.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">8</span> Is <code>counter += 1</code> atomic across threads?</div><div class="qa-a">No — it's read-modify-write; the GIL can switch threads between steps, losing updates. Use a <code>Lock</code>.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">9</span> How do NumPy/PyTorch get around the GIL?</div><div class="qa-a">Their heavy work runs in C with the GIL released, so multiple threads/cores execute the numeric kernels in parallel.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">10</span> What breaks an asyncio program's concurrency?</div><div class="qa-a">Blocking the event loop with synchronous work (<code>time.sleep</code>, a CPU loop, a blocking library) — it stalls every task. Offload with <code>run_in_executor</code>.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">11</span> Concurrency vs parallelism?</div><div class="qa-a">Concurrency = dealing with many things at once (interleaving); parallelism = doing many things at the same instant (multiple cores). Threads/async give concurrency; multiprocessing gives parallelism.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">12</span> Is the GIL part of the Python language?</div><div class="qa-a">No — it's a CPython implementation detail. Jython/IronPython have none; PyPy has one; and CPython 3.13+ ships an experimental free-threaded (no-GIL) build.</div></div>
