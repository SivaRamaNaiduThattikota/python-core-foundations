<h3 class="sub">🔹 1.2 &nbsp;<code>__init__</code> initializes; <code>self</code> is the instance</h3><div class="theory-box"><code>__init__</code> runs automatically on each new instance to set up its initial attributes — it <strong>initializes</strong>, it doesn't "construct" (that's <code>__new__</code>, Chunk D). <code>self</code> is the instance the method is working on; Python passes it automatically as the first argument. Attributes set as <code>self.x = ...</code> live on that specific instance.</div><div class="code-block"><span class="ck">class</span> <span class="cm">Dog</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">def</span> <span class="cm">__init__</span>(self, name, age):   <span class="cc"># runs on each new Dog(...)</span><br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;self.name = name                <span class="cc"># store state on THIS instance</span><br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;self.age = age<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">def</span> <span class="cm">bark</span>(self):<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">return</span> <span class="cs">f"{self.name} says woof"</span><br><br>d = <span class="cm">Dog</span>(<span class="cs">"Rex"</span>, <span class="cn">3</span>)<br>d.name, d.age   <span class="cc"># ('Rex', 3)</span><br>d.bark()        <span class="cc"># 'Rex says woof'</span><br><span class="cm">type</span>(d)         <span class="cc"># &lt;class '__main__.Dog'&gt;</span></div><div class="output-block">('Rex', 3)<br>Rex says woof<br>Dog</div><div class="note-box">💡 <code>d.bark()</code> is shorthand for <code>Dog.bark(d)</code> — Python fills <code>self</code> with the instance before the dot. Verified: <code>Dog.bark(d)</code> also returns <code>'Rex says woof'</code>.</div>

<h3 class="sub">🔹 1.3 &nbsp;Instances are independent</h3><div class="theory-box">Each <code>Dog(...)</code> is a separate object with its own attribute values. <h3 class="sub">🔹 1.4 &nbsp;A class is itself an object; <code>__new__</code> vs <code>__init__</code></h3><div class="theory-box">"Everything is an object" includes classes: <code>Dog</code> is an object of type <code>type</code>. <div class="chunk-badge">Part 1 · Chunk B — Instance vs Class Attributes</div><h3 class="sub">🔹 1.5 &nbsp;Where attributes live — two namespaces</h3><div class="theory-box">Two separate namespaces hold attributes. <strong>Instance attributes</strong> live in <code>instance.__dict__</code> (set via <code>self.x = ...</code>, unique per object). <strong>Class attributes</strong> live in <code>Class.__dict__</code> (defined in the class body, shared by all instances). Methods are just class attributes that happen to be functions.</div><div class="code-block"><span class="ck">class</span> <span class="cm">Dog</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;species = <span class="cs">"Canis familiaris"</span>   <span class="cc"># class attribute (shared)</span><br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">def</span> <span class="cm">__init__</span>(self, name):<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;self.name = name              <span class="cc"># instance attribute (per object)</span><br><br>d1, d2 = <span class="cm">Dog</span>(<span class="cs">"Rex"</span>), <span class="cm">Dog</span>(<span class="cs">"Fido"</span>)<br>d1.species, d2.species   <span class="cc"># both 'Canis familiaris' - shared</span><br>d1.__dict__              <span class="cc"># {'name': 'Rex'} - ONLY instance attrs</span></div><div class="output-block">Canis familiaris | Canis familiaris<br>{'name': 'Rex'}</div><h3 class="sub">🔹 1.6 &nbsp;Attribute lookup: instance → class → bases</h3><div class="theory-box">Reading <code>obj.x</code> searches the <strong>instance</strong> first, then the <strong>class</strong>, then its <strong>base classes</strong> (7C's MRO). That's why every instance can read <code>species</code> even though only the class defines it. But <strong>writing</strong> <code>obj.x = ...</code> always creates/updates an attribute on the <em>instance</em> — never the class.</div><div class="mut-grid"><div class="mut-card mc-imm"><div class="mc-title">📖 Read <code>obj.x</code></div><div class="mc-body">instance.__dict__<br>&nbsp;&nbsp;↓ (miss)<br>Class.__dict__<br>&nbsp;&nbsp;↓ (miss)<br>base classes (MRO)</div></div><div class="mut-card mc-mut"><div class="mc-title">✍ Write <code>obj.x = v</code></div><div class="mc-body">always sets<br>instance.__dict__["x"]<br><br>never touches the<br>class attribute</div></div></div>

<h3 class="sub">🔹 1.7 &nbsp;Shadowing — an instance attribute hides the class one</h3><div class="theory-box">Because writes go to the instance, assigning <code>d1.species = ...</code> creates an instance attribute that <strong>shadows</strong> the class attribute — for that instance only. An instance can override its own copy by assignment without affecting anyone else.</div><h3 class="sub">🔹 1.8 &nbsp;The mutable class-attribute trap (S1 callback)</h3><div class="theory-box">Here's the sharp edge. independent</span></div><div class="output-block">a: ['apple'] | b: []</div><div class="why-box"><strong>Why it matters:</strong> this is Session 1's mutable-default trap and mutable class-attribute rule, now in full OOP form — the single most common "why is state leaking between my objects?" bug. Rule: <strong>class attributes for shared constants; <code>__init__</code> for per-instance mutable state.</strong></div><hr class="divider"><div style="background:#1e1e2e; border-left:4px solid #cba6f7; padding:16px 20px; border-radius:8px; font-family:monospace; color:#cdd6f4;"><h4 style="color:#cba6f7; margin:0 0 12px 0;">🔑 Chunk B — Key Takeaways</h4><ul style="margin:0; padding-left:20px; line-height:2.1"><li>Instance attrs live in <code>instance.__dict__</code>; class attrs in <code>Class.__dict__</code> (shared)</li><li>Reads search instance → class → bases; writes always land on the instance</li><li>Assigning shadows a class attr for one instance; the class stays unchanged</li><li><strong>Never a mutable class attribute</strong> — it's shared; put mutable state in <code>__init__</code> (S1)</li></ul></div>

<div class="info-box"><strong>Part 2:</strong> Golden Rules → Common Traps → Exercise</div><div class="part-header"><h2>4. Golden Rules</h2></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 1</span> Initialize all instance state in <code>__init__</code>.</div><div class="body-txt">Every attribute an instance will use should be set in <code>__init__</code> — predictable objects, no surprise <code>AttributeError</code>.</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 2</span> Class attributes for shared constants; <code>__init__</code> for per-instance mutable state.</div><div class="body-txt">Never a mutable class attribute (list/dict) — it's shared across all instances (S1 / Chunk B).</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 3</span> Every method takes <code>self</code> first (for instance methods).</div><div class="body-txt">It receives the instance automatically; omitting it is a <code>TypeError</code> on call.</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 4</span> Prefer pure methods that return new objects over ones that mutate in place — unless mutation is the point.</div><div class="body-txt">e.g. <code>Vector.add</code> returns a new vector. If a method mutates, name it clearly (<code>update</code>, <code>append</code>).</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 5</span> Use <code>getattr(obj, name, default)</code> when a field might be absent.</div><div class="body-txt">The object cousin of <code>dict.get</code> — safe, avoids <code>AttributeError</code>.</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 6</span> Keep invariants inside methods.</div><div class="body-txt">Guard state changes (like <code>withdraw</code> checking the balance) in the method, so no caller can break the object's rules.</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 7</span> Give value-objects a real <code>__eq__</code>/<code>__repr__</code> (coming in 7B).</div><div class="body-txt">Default <code>==</code> is identity and default <code>repr</code> is unhelpful; 7B fixes both.</div></div>
