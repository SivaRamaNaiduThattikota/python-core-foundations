### Exercises — Solutions

```python
# E1 — Rectangle
class Rectangle:
    def __init__(self, w, h): self.w = w; self.h = h
    def area(self): return self.w * self.h
    def perimeter(self): return 2 * (self.w + self.h)

r = Rectangle(3, 4)
print(r.area(), r.perimeter())        # 12 14
```

```python
# E2 — Circle with class constant PI
class Circle:
    PI = 3.14159
    def __init__(self, radius): self.radius = radius
    def area(self): return self.PI * self.radius ** 2

print(round(Circle(2).area(), 4))     # 12.5664
```

```python
# E3 — Person
class Person:
    def __init__(self, name, age): self.name = name; self.age = age
    def greet(self): return f"Hi, I'm {self.name} ({self.age})"

print(Person("Siva", 27).greet())
```

```python
# E4 — fix the mutable class attribute bug
class Team:
    def __init__(self): self.members = []   # per-instance, NOT a class attr
    def add(self, m): self.members.append(m)

a, b = Team(), Team(); a.add("x")
print(a.members, b.members)           # ['x'] []  - independent
```

```python
# E5 — Stack
class Stack:
    def __init__(self): self._items = []
    def push(self, x): self._items.append(x)
    def pop(self): return self._items.pop()
    def peek(self): return self._items[-1]
    def is_empty(self): return not self._items

s = Stack(); s.push(1); s.push(2)
print(s.peek(), s.pop(), s.is_empty())   # 2 2 False
```

```python
# E6 — Temperature
class Temperature:
    def __init__(self, celsius): self.celsius = celsius
    def to_fahrenheit(self): return self.celsius * 9/5 + 32
    def to_kelvin(self): return self.celsius + 273.15

t = Temperature(100)
print(t.to_fahrenheit(), t.to_kelvin())  # 212.0 373.15
```

```python
# E7 — instance counter
class Widget:
    count = 0
    def __init__(self): Widget.count += 1

Widget(); Widget(); Widget()
print(Widget.count)                   # 3
```

```python
# E8 — to_dict()
class Record:
    def __init__(self, **kw):
        for k, v in kw.items(): setattr(self, k, v)
    def to_dict(self): return dict(vars(self))

print(Record(a=1, b=2).to_dict())     # {'a': 1, 'b': 2}
```

```python
# E9 — Vector2D.add returns a NEW vector (pure)
class Vector2D:
    def __init__(self, x, y): self.x = x; self.y = y
    def add(self, other): return Vector2D(self.x + other.x, self.y + other.y)

v = Vector2D(1, 2).add(Vector2D(3, 4))
print((v.x, v.y))                     # (4, 6)
```

```python
# E10 — Grid built with a comprehension (no aliasing)
class Grid:
    def __init__(self, rows, cols): self.cells = [[0]*cols for _ in range(rows)]
    def set(self, r, c, v): self.cells[r][c] = v
    def get(self, r, c): return self.cells[r][c]

g = Grid(2, 3); g.set(0, 0, 9)
print(g.cells)                        # [[9, 0, 0], [0, 0, 0]]
```

```python
# E11 — Config: defaults (class attr) + overrides (instance)
class Config:
    defaults = {"lr": 0.01, "epochs": 100}
    def __init__(self, **overrides):
        for k, v in overrides.items(): setattr(self, k, v)
    def get(self, key): return getattr(self, key, self.defaults.get(key))

cfg = Config(lr=0.001)
print(cfg.get("lr"), cfg.get("epochs"))   # 0.001 100
```

```python
# E12 — Subject/Observer (duck typing: any object with .update)
class Subject:
    def __init__(self): self.observers = []
    def register(self, obs): self.observers.append(obs)
    def notify(self, event):
        for obs in self.observers: obs.update(event)
class Logger:
    def __init__(self): self.seen = []
    def update(self, event): self.seen.append(event)

subj = Subject(); log = Logger(); subj.register(log)
subj.notify("start"); subj.notify("stop")
print(log.seen)                       # ['start', 'stop']
```

### Code Challenges — Solutions

```python
# C1 — Toggle
class Toggle:
    def __init__(self): self.on = False
    def flip(self): self.on = not self.on; return self.on
    def is_on(self): return self.on

t = Toggle()
print(t.flip(), t.flip(), t.is_on())   # True False False
```

```python
# C2 — Queue (FIFO)
class Queue:
    def __init__(self): self._items = []
    def enqueue(self, x): self._items.append(x)
    def dequeue(self): return self._items.pop(0)   # O(n); deque is O(1)
    def is_empty(self): return not self._items

q = Queue(); q.enqueue(1); q.enqueue(2)
print(q.dequeue(), q.dequeue(), q.is_empty())   # 1 2 True
```

```python
# C3 — RunningMean (metric accumulator)
class RunningMean:
    def __init__(self): self.total = 0; self.n = 0
    def add(self, x): self.total += x; self.n += 1
    def mean(self): return self.total / self.n if self.n else 0

rm = RunningMean(); rm.add(10); rm.add(20)
print(rm.mean())                      # 15.0
```

```python
# C4 — Histogram
class Histogram:
    def __init__(self): self._c = {}
    def add(self, v): self._c[v] = self._c.get(v, 0) + 1
    def counts(self): return dict(self._c)

h = Histogram()
for ch in "aabbbc": h.add(ch)
print(h.counts())                     # {'a': 2, 'b': 3, 'c': 1}
```

```python
# C5 — Inventory (guarded remove)
class Inventory:
    def __init__(self): self.stock = {}
    def add(self, item, qty): self.stock[item] = self.stock.get(item, 0) + qty
    def remove(self, item, qty):
        if self.stock.get(item, 0) < qty: raise ValueError("not enough")
        self.stock[item] -= qty
    def total(self): return sum(self.stock.values())

inv = Inventory(); inv.add("x", 5); inv.remove("x", 2)
print(inv.total())                    # 3
try: inv.remove("x", 99)
except ValueError as e: print("raised:", e)
```

```python
# C6 — Player (clamped health invariant)
class Player:
    def __init__(self, max_health=100):
        self.max_health = max_health; self.health = max_health
    def take_damage(self, d): self.health = max(0, self.health - d)
    def heal(self, h): self.health = min(self.max_health, self.health + h)

p = Player(100); p.take_damage(150)
print(p.health)                       # 0
p.heal(30)
print(p.health)                       # 30
```

```python
# C7 — MovingAverage (sliding window)
from collections import deque
class MovingAverage:
    def __init__(self, window): self.vals = deque(maxlen=window)
    def add(self, x):
        self.vals.append(x)
        return sum(self.vals) / len(self.vals)

ma = MovingAverage(3)
print(ma.add(1), ma.add(2), ma.add(3), ma.add(4))   # 1.0 1.5 2.0 3.0
```

```python
# C8 — MiniDataset (ML Dataset/DataLoader shape)
class MiniDataset:
    def __init__(self, data): self.data = list(data)
    def size(self): return len(self.data)
    def get(self, i): return self.data[i]
    def batches(self, n): return [self.data[i:i+n] for i in range(0, len(self.data), n)]

ds = MiniDataset(range(7))
print(ds.size(), ds.get(0), ds.batches(3))   # 7 0 [[0,1,2],[3,4,5],[6]]
```
