<div class="part-header"><h2>1. Theory</h2></div><h3 class="sub">🔹 1.1 &nbsp;Dunder methods hook into Python's data model</h3><div class="theory-box"><strong>Dunder</strong> = "double underscore" (<code>__name__</code>). You already met some: <code>__init__</code> (7A), <code>__call__</code> (class-based decorators, S6), <code>__hash__</code> (why tuples work as dict keys, 2B). This session makes them systematic.</div>

<h3 class="sub">🔹 1.2 &nbsp;<code>__repr__</code> — the unambiguous developer representation</h3><div class="theory-box"><code>__repr__</code> returns a string for developers: unambiguous, and ideally looks like the code to recreate the object. <code>repr(obj)</code>, the interactive echo, and <strong>containers</strong> (a list of your objects) all use <code>__repr__</code>. Define it on <em>every</em> class — the default (<code>&lt;...object at 0x...&gt;</code>) is useless for debugging.</div><div class="code-block"><span class="ck">class</span> <span class="cm">Point</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">def</span> <span class="cm">__init__</span>(self, x, y): self.x = x; self.y = y<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">def</span> <span class="cm">__repr__</span>(self): <span class="ck">return</span> <span class="cs">f"Point({self.x}, {self.y})"</span><br><br>p = <span class="cm">Point</span>(<span class="cn">1</span>, <span class="cn">2</span>)<br><span class="cm">repr</span>(p)   <span class="cc"># 'Point(1, 2)'</span><br>[p]        <span class="cc"># [Point(1, 2)]  - containers use __repr__ for elements</span></div><div class="output-block">Point(1, 2)<br>[Point(1, 2)]</div><div class="note-box">💡 Good <code>__repr__</code> convention: return something that could be pasted back to recreate the object — <code>Point(1, 2)</code>, not <code>"a point"</code>. It makes logs, tracebacks, and a list of objects instantly readable.</div>

Golden Rules</h2></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 1</span> Define <code>__repr__</code> on every class.</div><div class="body-txt">Make it unambiguous / recreatable. <div class="interview-header"><h2>8. Interview Questions</h2></div><div class="sub-header"><h3>8a — Conceptual Q&amp;A</h3></div><div class="qa-block"><div class="qa-q"><span class="q-num">1</span> <code>__repr__</code> vs <code>__str__</code>?</div><div class="qa-a"><code>__repr__</code> is the unambiguous developer form (repr, echo, containers); <code>__str__</code> is the readable end-user form (str, print). Missing <code>__str__</code> falls back to <code>__repr__</code>.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">2</span> Why define <code>__repr__</code> on every class?</div><div class="qa-a">The default shows <code>&lt;X object at 0x...&gt;</code>, useless in logs/tracebacks/containers. A recreatable repr makes objects self-describing.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">3</span> What's the <code>__eq__</code>/<code>__hash__</code> contract?</div><div class="qa-a">Equal objects must have equal hashes. Defining <code>__eq__</code> nulls <code>__hash__</code> (object becomes unhashable), so you must define <code>__hash__</code> too — over the same fields.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">4</span> Why does a custom <code>__eq__</code> make an object unhashable?</div><div class="qa-a">Python can't guarantee the contract automatically, so it sets <code>__hash__ = None</code> to force you to define a consistent one (or leave it unhashable on purpose).</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">5</span> Why return <code>NotImplemented</code> from <code>__eq__</code>/<code>__add__</code>?</div><div class="qa-a">It tells Python to try the other operand's reflected method, then fall back. Returning <code>False</code>/raising breaks mixed-type and subclass operations.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">6</span> What does <code>@functools.total_ordering</code> do?</div><div class="qa-a">Generates the remaining comparison operators from <code>__eq__</code> + one of <code>__lt__</code>/<code>__le__</code>/<code>__gt__</code>/<code>__ge__</code>.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">7</span> Which dunders make an object sequence-like?</div><div class="qa-a"><code>__len__</code> (len), <code>__getitem__</code> (indexing/slicing/iteration), <code>__contains__</code> (in). This is the <code>Dataset</code> protocol.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">8</span> What is <code>__call__</code>?</div><div class="qa-a">Makes an instance callable — <code>obj(x)</code> runs <code>obj.__call__(x)</code>. Powers class-based decorators (S6) and callable models (<code>model(x)</code>).</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">9</span> How does <code>__getitem__</code> alone enable iteration?</div><div class="qa-a">The legacy protocol calls <code>__getitem__</code> with 0, 1, 2… until <code>IndexError</code>. (The modern way is <code>__iter__</code> — Session 8.)</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">10</span> How is truthiness determined for an object?</div><div class="qa-a"><code>__bool__</code> if defined, else <code>__len__</code> (0 → falsy), else always truthy.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">11</span> How do you overload <code>+</code> or <code>@</code>?</div><div class="qa-a">Define <code>__add__</code> / <code>__matmul__</code> (and reflected <code>__radd__</code> etc.). They should return a new object. This is how NumPy/PyTorch implement array math.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">12</span> When is it safe to make an object hashable?</div><div class="qa-a">Only when the fields used in <code>__hash__</code>/<code>__eq__</code> never change after creation — otherwise the object gets lost in dicts/sets (mutable-key trap).</div></div>
