### Exercises — Solutions

```python
import math
from functools import cached_property
```

```python
# E1 — Rectangle read-only area
class Rectangle:
    def __init__(self, w, h): self.w = w; self.h = h
    @property
    def area(self): return self.w * self.h

print(Rectangle(3, 4).area)           # 12
```

```python
# E2 — Person full_name
class Person:
    def __init__(self, first, last): self.first = first; self.last = last
    @property
    def full_name(self): return f"{self.first} {self.last}"

print(Person("Siva", "N").full_name)
```

```python
# E3 — Circle diameter (read-only)
class Circle:
    def __init__(self, r): self.radius = r
    @property
    def diameter(self): return self.radius * 2

print(Circle(5).diameter)             # 10
```

```python
# E4 — Age validated setter
class Age:
    def __init__(self, years): self.years = years
    @property
    def years(self): return self._years
    @years.setter
    def years(self, v):
        if v < 0: raise ValueError("negative age")
        self._years = v

a = Age(30); print(a.years)
try: Age(-1)
except ValueError as e: print("raised:", e)
```

```python
# E5 — Grade clamped/validated 0-100
class Grade:
    def __init__(self, score): self.score = score
    @property
    def score(self): return self._score
    @score.setter
    def score(self, v):
        if not 0 <= v <= 100: raise ValueError("0-100 only")
        self._score = v

print(Grade(85).score)
try: Grade(150)
except ValueError as e: print("raised:", e)
```

```python
# E6 — Money read-only dollars
class Money:
    def __init__(self, cents): self.cents = cents
    @property
    def dollars(self): return self.cents / 100

print(Money(250).dollars)             # 2.5
```

```python
# E7 — Point.from_tuple classmethod
class Point:
    def __init__(self, x, y): self.x = x; self.y = y
    @classmethod
    def from_tuple(cls, t): return cls(t[0], t[1])

p = Point.from_tuple((3, 4)); print(p.x, p.y)   # 3 4
```

```python
# E8 — Color.from_hex classmethod
class Color:
    def __init__(self, r, g, b): self.r = r; self.g = g; self.b = b
    def __repr__(self): return f"Color({self.r}, {self.g}, {self.b})"
    @classmethod
    def from_hex(cls, h):
        h = h.lstrip("#")
        return cls(int(h[0:2],16), int(h[2:4],16), int(h[4:6],16))

print(Color.from_hex("#ff0080"))      # Color(255, 0, 128)
```

```python
# E9 — TextUtils.word_count staticmethod
class TextUtils:
    @staticmethod
    def word_count(s): return len(s.split())

print(TextUtils.word_count("a b c d"))   # 4
```

```python
# E10 — BankAccount: validated property + methods + from_dict
class BankAccount:
    def __init__(self, balance=0): self.balance = balance
    @property
    def balance(self): return self._balance
    @balance.setter
    def balance(self, v):
        if v < 0: raise ValueError("negative")
        self._balance = v
    def deposit(self, amt): self.balance += amt
    def withdraw(self, amt):
        if amt > self.balance: raise ValueError("insufficient")
        self.balance -= amt
    @classmethod
    def from_dict(cls, d): return cls(d["balance"])

a = BankAccount.from_dict({"balance": 100}); a.deposit(50)
print(a.balance)                      # 150
try: a.withdraw(500)
except ValueError as e: print("raised:", e)
```

```python
# E11 — Vector2D with __slots__
class Vector2D:
    __slots__ = ("x", "y")
    def __init__(self, x, y): self.x = x; self.y = y

v = Vector2D(1, 2)
print(hasattr(v, "__dict__"))         # False
try: v.z = 3
except AttributeError as e: print("raised:", e)
```

```python
# E12 — Temperature: two linked read-write properties
class Temperature:
    def __init__(self, celsius=0): self._celsius = celsius
    @property
    def celsius(self): return self._celsius
    @celsius.setter
    def celsius(self, v): self._celsius = v
    @property
    def fahrenheit(self): return self._celsius * 9/5 + 32
    @fahrenheit.setter
    def fahrenheit(self, v): self._celsius = (v - 32) * 5/9

t = Temperature(0)
print(t.fahrenheit)                   # 32.0
t.fahrenheit = 212
print(t.celsius)                      # 100.0
```

### Code Challenges — Solutions

```python
# C1 — Employee annual
class Employee:
    def __init__(self, monthly): self.monthly = monthly
    @property
    def annual(self): return self.monthly * 12

print(Employee(1000).annual)          # 12000
```

```python
# C2 — Duration.from_minutes
class Duration:
    def __init__(self, seconds): self.seconds = seconds
    @classmethod
    def from_minutes(cls, m): return cls(m * 60)

print(Duration.from_minutes(2).seconds)   # 120
```

```python
# C3 — Email validated setter
class Email:
    def __init__(self, addr): self.addr = addr
    @property
    def addr(self): return self._addr
    @addr.setter
    def addr(self, v):
        if "@" not in v: raise ValueError("invalid email")
        self._addr = v

print(Email("a@b.com").addr)
try: Email("bad")
except ValueError as e: print("raised:", e)
```

```python
# C4 — Password with staticmethod validator
class Password:
    def __init__(self, pw): self.pw = pw
    @staticmethod
    def is_strong(pw): return len(pw) >= 8
    @property
    def pw(self): return self._pw
    @pw.setter
    def pw(self, v):
        if not Password.is_strong(v): raise ValueError("too weak")
        self._pw = v

Password("longenough")
try: Password("short")
except ValueError as e: print("raised:", e)
```

```python
# C5 — Widget instance counter
class Widget:
    count = 0
    def __init__(self): Widget.count += 1
    @classmethod
    def total(cls): return cls.count

Widget(); Widget()
print(Widget.total())                 # 2
```

```python
# C6 — Angle two-way degrees<->radians
import math
class Angle:
    def __init__(self, degrees=0): self._degrees = degrees
    @property
    def degrees(self): return self._degrees
    @degrees.setter
    def degrees(self, v): self._degrees = v
    @property
    def radians(self): return math.radians(self._degrees)
    @radians.setter
    def radians(self, v): self._degrees = math.degrees(v)

a = Angle(180); print(round(a.radians, 5))   # 3.14159
a.radians = math.pi / 2; print(a.degrees)    # 90.0
```

```python
# C7 — Vec: slots + magnitude property + from_tuple + repr
class Vec:
    __slots__ = ("x", "y")
    def __init__(self, x, y): self.x = x; self.y = y
    @property
    def magnitude(self): return (self.x**2 + self.y**2) ** 0.5
    @classmethod
    def from_tuple(cls, t): return cls(*t)
    def __repr__(self): return f"Vec({self.x}, {self.y})"

print(Vec(3, 4).magnitude, Vec.from_tuple((1, 2)))   # 5.0 Vec(1, 2)
```

```python
# C8 — cached_property: expensive value computed once
from functools import cached_property
class Dataset:
    def __init__(self, data): self.data = data; self.compute_calls = 0
    @cached_property
    def stats(self):
        self.compute_calls += 1
        return {"count": len(self.data), "sum": sum(self.data)}

ds = Dataset([1, 2, 3])
ds.stats; ds.stats
print(ds.stats, "| computed", ds.compute_calls, "time(s)")   # ... computed 1 time(s)
```
