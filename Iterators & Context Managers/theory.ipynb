False<br>iterator is its own iterator? True</div><div class="info-box">💡 <strong>SQL / Power BI anchor:</strong> an iterator is a <strong>cursor</strong> over a result set — it tracks position and yields one row at a time via <code>next()</code>. You open a new cursor (<code>iter()</code>) each time you want to scan again.</div><div class="mut-grid"><div class="mut-card mc-mut"><div class="mc-title">🔁 Iterable</div><div class="mc-body">has __iter__()<br>can be looped many times<br>e.g. list, dict, str, range<br>iter(x) -> a new iterator</div></div><div class="mut-card mc-imm"><div class="mc-title">➡ Iterator (cursor)</div><div class="mc-body">has __next__() + __iter__()<br>one-shot: exhausts once<br>raises StopIteration at end<br>iter(it) is it</div></div></div>

<h3 class="sub">🔹 1.2 &nbsp;How a <code>for</code> loop actually works</h3><div class="theory-box">A <code>for</code> loop is sugar over the protocol: it calls <code>iter()</code> once to get an iterator, then <code>next()</code> repeatedly, stopping when <code>StopIteration</code> is raised. Comprehensions, <code>sum()</code>, <code>list()</code>, unpacking — all do exactly this.</div><div class="code-block"><span class="cc"># for x in [10, 20]: ...  is equivalent to:</span><br>it = <span class="cm">iter</span>([<span class="cn">10</span>, <span class="cn">20</span>])<br><span class="ck">while</span> <span class="ck">True</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">try</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;x = <span class="cm">next</span>(it)<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">except</span> StopIteration:<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">break</span><br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="cm">print</span>(x)      <span class="cc"># 10, then 20</span></div><div class="output-block">10<br>20</div><div class="note-box">💡 <code>StopIteration</code> is a normal signal, not an error — the loop machinery catches it for you. That's why you never see it in ordinary <code>for</code> code.</div>

<h3 class="sub">🔹 1.5 &nbsp;Generators <em>are</em> iterators (Session 4 connection)</h3><div class="theory-box">A generator function (with <code>yield</code>) returns a generator object that <strong>already implements the iterator protocol</strong> — <code>__iter__</code> and <code>__next__</code> for free. The same countdown is a few lines, no class needed.</div><div class="code-block"><span class="ck">def</span> <span class="cm">countdown</span>(n):<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">while</span> n &gt; <span class="cn">0</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">yield</span> n; n -= <span class="cn">1</span><br><br><span class="cm">list</span>(<span class="cm">countdown</span>(<span class="cn">3</span>))          <span class="cc"># [3, 2, 1]</span><br>g = <span class="cm">countdown</span>(<span class="cn">3</span>)<br><span class="cm">iter</span>(g) <span class="ck">is</span> g, <span class="cm">type</span>(g).__name__   <span class="cc"># (True, 'generator') - it's an iterator</span></div><div class="output-block">[3, 2, 1]<br>gen is iter: True | type: generator</div><h3 class="sub">🔹 1.6 &nbsp;Re-iterable objects — make <code>__iter__</code> a generator method</h3><div class="theory-box">The fix for the one-shot problem: don't return <code>self</code>. Make <code>__iter__</code> a <strong>generator method</strong> (or return a fresh iterator) — each loop gets a brand-new generator, so the object is <strong>re-iterable</strong> like a list.</div><div class="code-block"><span class="ck">class</span> <span class="cm">Squares</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">def</span> <span class="cm">__init__</span>(self, n): self.n = n<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">def</span> <span class="cm">__iter__</span>(self):        <span class="cc"># a generator method -> fresh iterator each call</span><br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">for</span> i <span class="ck">in</span> <span class="cm">range</span>(<span class="cn">1</span>, self.n + <span class="cn">1</span>):<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">yield</span> i * i<br><br>sq = <span class="cm">Squares</span>(<span class="cn">3</span>)<br><span class="cm">list</span>(sq), <span class="cm">list</span>(sq)   <span class="cc"># ([1, 4, 9], [1, 4, 9])  - re-iterable!</span></div><div class="output-block">[1, 4, 9] [1, 4, 9]</div><div class="why-box"><strong>The design rule:</strong> a <strong>container</strong> (re-iterable) has <code>__iter__</code> return a <em>fresh</em> iterator each time (a generator method is the easy way); a <strong>one-shot iterator</strong> returns <code>self</code>. This is exactly why a <code>list</code> loops many times but its <code>iter()</code> cursor doesn't — and how you'd build a <code>Dataset</code> you can epoch over repeatedly.</div>

<h3 class="sub">🔹 1.9 &nbsp;The protocol — <code>__enter__</code> / <code>__exit__</code></h3><div class="theory-box"><code>__enter__(self)</code> runs on entry; its return value is bound to the <code>as</code> variable. <code>__exit__(self, exc_type, exc_val, tb)</code> runs on exit — <strong>always</strong>, including when the body raised (the three args describe the exception, or are <code>None</code> on clean exit). Return a truthy value from <code>__exit__</code> to <strong>suppress</strong> the exception.</div><div class="code-block"><span class="ck">class</span> <span class="cm">Managed</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">def</span> <span class="cm">__enter__</span>(self):<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span class="cm">print</span>(<span class="cs">"enter"</span>); <span class="ck">return</span> <span class="cs">"resource"</span>   <span class="cc"># bound to `as r`</span><br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">def</span> <span class="cm">__exit__</span>(self, exc_type, exc_val, tb):<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span class="cm">print</span>(<span class="cs">"exit"</span>); <span class="ck">return</span> <span class="ck">False</span>       <span class="cc"># don't suppress exceptions</span><br><br><span class="ck">with</span> <span class="cm">Managed</span>() <span class="ck">as</span> r:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="cm">print</span>(<span class="cs">"using"</span>, r)</div><div class="output-block">enter<br>using resource<br>exit</div><h3 class="sub">🔹 1.10 &nbsp;Cleanup runs on exceptions; <div class="chunk-badge">Part 1 · Chunk D — <code>contextlib</code> + Examples &amp; Edge Cases</div><h3 class="sub">🔹 1.11 &nbsp;<code>@contextmanager</code> — a context manager from a generator</h3><div class="theory-box">Writing a class for a simple manager is overkill. <code>@contextlib.contextmanager</code> turns a generator into one: code <strong>before <code>yield</code></strong> is the setup (<code>__enter__</code>), the yielded value is bound to <code>as</code>, and code <strong>after <code>yield</code></strong> is the teardown (<code>__exit__</code>). Wrap the <code>yield</code> in <code>try/finally</code> so teardown runs even on error.</div><div class="code-block"><span class="ck">from</span> contextlib <span class="ck">import</span> contextmanager<br><br><span class="cm">@contextmanager</span><br><span class="ck">def</span> <span class="cm">managed</span>():<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="cm">print</span>(<span class="cs">"setup"</span>)          <span class="cc"># __enter__</span><br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">try</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">yield</span> <span class="cs">"res"</span>          <span class="cc"># value bound to `as r`</span><br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">finally</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span class="cm">print</span>(<span class="cs">"teardown"</span>)   <span class="cc"># __exit__ (runs even on exception)</span><br><br><span class="ck">with</span> <span class="cm">managed</span>() <span class="ck">as</span> r:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="cm">print</span>(<span class="cs">"use"</span>, r)</div><div class="output-block">setup<br>use res<br>teardown</div><div class="note-box">💡 Also handy: <code>contextlib.suppress(FileNotFoundError)</code> to ignore specific exceptions in a block, and <code>contextlib.closing(obj)</code> to call <code>obj.close()</code> on exit for objects that aren't context managers themselves.</div><div class="ex-header"><h2>2. Example</h2></div><div class="ex-block"><div class="ex-title"><span class="ex-badge">Ex 1</span> The canonical file <code>with</code></div><div class="code-block"><span class="ck">with</span> <span class="cm">open</span>(<span class="cs">"data.txt"</span>) <span class="ck">as</span> f:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">for</span> line <span class="ck">in</span> f:   <span class="cc"># the file object is itself an iterator (Chunk A)!</span><br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;process(line)<br><span class="cc"># f.close() guaranteed on exit, even if process() raises</span></div><div class="output-block">(file closed automatically)</div><div class="note-box">Both protocols in one line: <code>open()</code> returns a context manager <em>and</em> an iterator — <code>with</code> manages its lifetime, <code>for</code> streams it lazily.</div></div><div class="ex-block"><div class="ex-title"><span class="ex-badge">Ex 2</span> A <code>Timer</code> context manager (class)</div><div class="code-block"><span class="ck">import</span> time<br><span class="ck">class</span> <span class="cm">Timer</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">def</span> <span class="cm">__enter__</span>(self): self.t0 = time.<span class="cm">perf_counter</span>(); <span class="ck">return</span> self<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">def</span> <span class="cm">__exit__</span>(self, *a): self.elapsed = time.<span class="cm">perf_counter</span>() - self.t0<br><br><span class="ck">with</span> <span class="cm">Timer</span>() <span class="ck">as</span> t:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="cm">sum</span>(<span class="cm">range</span>(<span class="cn">100000</span>))<br>t.elapsed   <span class="cc"># seconds (varies)</span></div><div class="output-block">t.elapsed set (timing varies)</div><div class="note-box">The standard "time a block" tool — enter records the start, exit computes the elapsed.</div></div><div class="ex-block"><div class="ex-title"><span class="ex-badge">Ex 3</span> <code>@contextmanager</code> for a temporary state change</div><div class="code-block"><span class="cm">@contextmanager</span><br><span class="ck">def</span> <span class="cm">temporary</span>(d, key, value):<br>&nbsp;&nbsp;&nbsp;&nbsp;old = d.<span class="cm">get</span>(key)<br>&nbsp;&nbsp;&nbsp;&nbsp;d[key] = value<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">try</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">yield</span> d<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">finally</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;d[key] = old   <span class="cc"># always restored</span></div><div class="output-block">config restored on exit</div><div class="note-box">Set-then-restore (config, random seed, working directory, <code>torch.no_grad()</code>-style flags) — the teardown guarantees the original is put back.</div></div><div class="ex-block"><div class="ex-title"><span class="ex-badge">Ex 4</span> Multiple managers &amp; <code>suppress</code></div><div class="code-block"><span class="ck">from</span> contextlib <span class="ck">import</span> suppress<br><span class="ck">with</span> suppress(FileNotFoundError):<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="cm">open</span>(<span class="cs">"missing.txt"</span>)   <span class="cc"># no crash if absent</span><br><br><span class="ck">with</span> <span class="cm">open</span>(<span class="cs">"a"</span>) <span class="ck">as</span> fa, <span class="cm">open</span>(<span class="cs">"b"</span>) <span class="ck">as</span> fb:   <span class="cc"># two managers; Edge Cases</h2></div><p style="color:#cdd6f4;font-family:'Segoe UI',sans-serif;font-size:0.92em;margin:0 0 12px 0">All outputs verified by running.</p><div class="edge-block"><div class="edge-title"><span class="edge-badge">Edge 1</span> <code>__exit__</code> runs even when the body raises</div><div class="code-block"><span class="ck">with</span> <span class="cm">CM</span>():<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">raise</span> ValueError(<span class="cs">"boom"</span>)   <span class="cc"># prints "cleanup ran", then re-raises</span></div><div class="output-block">cleanup ran<br>ValueError: boom</div><div class="why-box"><strong>Why:</strong> that guarantee is the whole point — resources are released on the error path too, unlike a plain assignment.</div></div><div class="edge-block"><div class="edge-title"><span class="edge-badge">Edge 2</span> Returning <code>True</code> from <code>__exit__</code> suppresses the exception</div><div class="code-block"><span class="ck">def</span> <span class="cm">__exit__</span>(self, et, ev, tb): <span class="ck">return</span> et <span class="ck">is</span> ValueError   <span class="cc"># swallow ValueError</span></div><div class="output-block">exception suppressed; code continues</div><div class="why-box"><strong>Why:</strong> powerful but dangerous — a truthy return hides the error from the caller. Reserve it for deliberate cases (like <code>suppress</code>); usually return <code>False</code>.</div></div><div class="edge-block"><div class="edge-title"><span class="edge-badge">Edge 3</span> <code>@contextmanager</code> without <code>try/finally</code> skips cleanup on error</div><div class="code-block"><span class="cm">@contextmanager</span><br><span class="ck">def</span> <span class="cm">bad</span>():<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="cm">print</span>(<span class="cs">"open"</span>)<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">yield</span><br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="cm">print</span>(<span class="cs">"close"</span>)   <span class="cc"># NOT reached if the body raises</span></div><div class="output-block">open<br>(then the exception; "close" never prints)</div><div class="why-box"><strong>Why:</strong> an exception in the body propagates <em>out through the <code>yield</code></em>, skipping the code after it. Always wrap the <code>yield</code> in <code>try/finally</code> to guarantee teardown.</div></div><div class="edge-block"><div class="edge-title"><span class="edge-badge">Edge 4</span> Multiple managers exit in reverse (LIFO) order</div><div class="code-block"><span class="ck">with</span> A() <span class="ck">as</span> a, B() <span class="ck">as</span> b:   <span class="cc"># enter A, enter B; exit B, exit A</span><br>&nbsp;&nbsp;&nbsp;&nbsp;...</div><div class="output-block">A in · B in · (body) · B out · A out</div><div class="why-box"><strong>Why:</strong> like nested <code>with</code>s — the innermost/last-entered is torn down first, so dependencies unwind correctly (e.g. close the cursor before the connection).</div></div><div class="edge-block"><div class="edge-title"><span class="edge-badge">Edge 5</span> No <code>return</code> in <code>__enter__</code> → <code>as</code> gets <code>None</code></div><div class="code-block"><span class="ck">class</span> <span class="cm">NoRet</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">def</span> <span class="cm">__enter__</span>(self): <span class="ck">pass</span>   <span class="cc"># returns None</span><br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">def</span> <span class="cm">__exit__</span>(self, *a): <span class="ck">pass</span><br><span class="ck">with</span> <span class="cm">NoRet</span>() <span class="ck">as</span> x:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="cm">print</span>(x)   <span class="cc"># None</span></div><div class="output-block">None</div><div class="why-box"><strong>Why:</strong> <code>as x</code> binds whatever <code>__enter__</code> returns. Forget to <code>return self</code> (or the resource) and you get <code>None</code> — a common "why is my context variable None?" bug.</div></div><hr class="divider"><div class="info-box">📎 <strong>End of Session 8 Part 1</strong> (Chunks A–D: iterator protocol, building iterators + generators, context managers, <code>contextlib</code>, 4 examples, 5 edge cases). <strong>Part 2</strong> — Golden Rules → Common Traps → Exercise. <strong>Part 3</strong> — ML Real-World → Interview Q&amp;A → Summary.</div>

Golden Rules</h2></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 1</span> Prefer generators over hand-written iterator classes.</div><div class="body-txt">A <code>yield</code> function is an iterator with far less boilerplate and no <code>StopIteration</code> bookkeeping.</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 2</span> Iterators are one-shot — keep the <em>iterable</em> to loop again.</div><div class="body-txt">Call <code>iter()</code> on the source anew; never pass a spent iterator to a second consumer.</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 3</span> Re-iterable class → <code>__iter__</code> returns a <em>fresh</em> iterator; one-shot → returns <code>self</code>.</div><div class="body-txt">A generator method for <code>__iter__</code> is the easy re-iterable pattern (like <code>list</code>).</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 4</span> Use <code>with</code> for anything needing cleanup.</div><div class="body-txt">Files, locks, connections, temporary state — <code>with</code> guarantees teardown even on exception.</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">Rule 5</span> Put teardown in <code>finally</code>; return <code>False</code> from <code>__exit__</code>.</div><div class="body-txt">In <code>@contextmanager</code>, wrap <code>yield</code> in <code>try/finally</code>. <div class="interview-header"><h2>8. Interview Questions</h2></div><div class="sub-header"><h3>8a — Conceptual Q&amp;A</h3></div><div class="qa-block"><div class="qa-q"><span class="q-num">1</span> Iterable vs iterator?</div><div class="qa-a">An iterable has <code>__iter__()</code> (can be looped many times); an iterator is the cursor with <code>__next__()</code> (one-shot) whose <code>__iter__</code> returns <code>self</code>.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">2</span> How does a <code>for</code> loop work under the hood?</div><div class="qa-a"><code>iter()</code> once to get an iterator, then <code>next()</code> repeatedly until <code>StopIteration</code>, which the loop catches.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">3</span> What is <code>StopIteration</code>?</div><div class="qa-a">The signal an iterator raises when it's exhausted. It's a normal control signal, caught by loop machinery — not an error you handle in ordinary code.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">4</span> Why is an iterator "one-shot"?</div><div class="qa-a">It tracks position and doesn't reset; once <code>StopIteration</code> is reached it stays exhausted. Re-<code>iter()</code> the iterable to loop again.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">5</span> How is a generator related to an iterator?</div><div class="qa-a">A generator <em>is</em> an iterator — it implements <code>__iter__</code>/<code>__next__</code> automatically. Far less code than a hand-written iterator class.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">6</span> How do you make a class re-iterable?</div><div class="qa-a"><code>__iter__</code> returns a <em>fresh</em> iterator each call (a generator method is easiest). Returning <code>self</code> makes it one-shot.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">7</span> What is a context manager / what does <code>with</code> guarantee?</div><div class="qa-a">An object with <code>__enter__</code>/<code>__exit__</code>. <code>with</code> guarantees <code>__exit__</code> runs on block exit — including when the body raises — for reliable cleanup.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">8</span> What are <code>__exit__</code>'s arguments and return value?</div><div class="qa-a"><code>(exc_type, exc_val, tb)</code> — all <code>None</code> on clean exit. Returning a truthy value <strong>suppresses</strong> the exception; otherwise it propagates.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">9</span> What does <code>@contextlib.contextmanager</code> do?</div><div class="qa-a">Turns a generator into a context manager: code before <code>yield</code> is setup, the yielded value binds to <code>as</code>, code after (in <code>finally</code>) is teardown.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">10</span> Why wrap the <code>yield</code> in <code>try/finally</code>?</div><div class="qa-a">So teardown runs even when the body raises — an exception propagates out through the <code>yield</code>, skipping any code after it otherwise.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">11</span> Name a few <code>itertools</code> tools and their use.</div><div class="qa-a"><code>islice</code> (finite slice of any iterator), <code>chain</code> (concatenate), <code>count</code>/<code>cycle</code> (infinite), <code>accumulate</code> (running totals), <code>groupby</code> (batching).</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">12</span> How do multiple managers in one <code>with</code> exit?</div><div class="qa-a">In reverse (LIFO) order — last entered, first exited — so dependencies unwind correctly.</div></div>
