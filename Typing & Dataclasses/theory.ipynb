<h1 class="main-title">🐍 Session 10 — Typing &amp; Dataclasses <span style="color:#a6e3a1;font-size:0.5em;">(finale)</span></h1><div class="info-box"><strong>Part 1:</strong> Theory → Example → Edge Cases · deep 4-chunk. The last session of the track.</div><div class="chunk-badge">Part 1 · Chunk A — Type Hints Basics</div><div class="theory-box" style="border-left-color:#cba6f7;">The capstone. <strong style="color:#89b4fa">Type hints</strong> annotate the types your code expects; <strong style="color:#89b4fa">dataclasses</strong> turn those annotations into clean data-holding classes with zero boilerplate. Together they feed the modern ML/backend stack — <strong>Pydantic</strong>, <strong>FastAPI</strong>, config objects, and typed model interfaces. Chunk A: hint basics; B: the typing toolkit (<code>Optional</code>/<code>Union</code>/generics); C: <code>@dataclass</code>; D: advanced + the Pydantic bridge.</div>

<div class="part-header"><h2>1. Theory</h2></div><h3 class="sub">🔹 1.1 &nbsp;What type hints are — and what they're for</h3><div class="theory-box">A <strong>type hint</strong> annotates the type a variable, parameter, or return value is <em>expected</em> to have. They are <strong>optional</strong> and Python stays dynamically typed. Their value is entirely in <strong>tooling and communication</strong>: IDE autocomplete &amp; error squiggles, static checkers (<code>mypy</code>/<code>pyright</code>), self-documenting signatures, and frameworks (dataclasses, Pydantic, FastAPI) that <em>read</em> them to generate behavior.</div><div class="info-box">💡 <strong>SQL / Power BI anchor:</strong> like declaring a column's data type in a schema — it documents intent and lets tools validate, but here it's advisory at runtime unless a framework chooses to enforce it (Pydantic does; Chunk D).</div>

<h3 class="sub">🔹 1.2 &nbsp;Function annotations</h3><div class="theory-box">Annotate parameters with <code>name: type</code> and the return with <code>-&gt; type</code>. They're stored in <code>fn.__annotations__</code> — which is exactly how dataclasses and Pydantic introspect your types.</div><div class="code-block"><span class="ck">def</span> <span class="cm">add</span>(a: <span class="cm">int</span>, b: <span class="cm">int</span>) -&gt; <span class="cm">int</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">return</span> a + b<br><br><span class="cm">add</span>(<span class="cn">2</span>, <span class="cn">3</span>)              <span class="cc"># 5</span><br>add.__annotations__     <span class="cc"># {'a': int, 'b': int, 'return': int}</span></div><div class="output-block">5<br>{'a': &lt;class 'int'&gt;, 'b': &lt;class 'int'&gt;, 'return': &lt;class 'int'&gt;}</div>

<h3 class="sub">🔹 1.3 &nbsp;Variable &amp; container annotations</h3><div class="theory-box">You can annotate variables too, and — importantly — <strong>parameterize container types</strong> to say what's inside: <code>list[int]</code>, <code>dict[str, int]</code>, <code>tuple[int, str]</code>. (Since Python 3.9 you use the builtins directly; older code imported <code>List</code>/<code>Dict</code> from <code>typing</code>.)</div><div class="code-block">x: <span class="cm">int</span> = <span class="cn">5</span><br>nums: <span class="cm">list</span>[<span class="cm">int</span>] = [<span class="cn">1</span>, <span class="cn">2</span>, <span class="cn">3</span>]<br>mapping: <span class="cm">dict</span>[<span class="cm">str</span>, <span class="cm">int</span>] = {<span class="cs">"a"</span>: <span class="cn">1</span>}<br>point: <span class="cm">tuple</span>[<span class="cm">int</span>, <span class="cm">int</span>] = (<span class="cn">1</span>, <span class="cn">2</span>)<br>row: <span class="cm">tuple</span>[<span class="cm">int</span>, ...] = (<span class="cn">1</span>, <span class="cn">2</span>, <span class="cn">3</span>)   <span class="cc"># variable-length homogeneous tuple</span></div><div class="output-block">5 [1, 2, 3] {'a': 1} (1, 2)</div><div class="note-box">💡 <code>list[int]</code> means "a list of ints", <code>dict[str, int]</code> is "str keys → int values". This is the vocabulary Pydantic/dataclasses use to know a field is, say, a <code>list[str]</code> of labels.</div>

<h3 class="sub">🔹 1.4 &nbsp;Hints are NOT enforced at runtime</h3><div class="theory-box">The interpreter <strong>ignores</strong> type hints when running — they don't coerce or check anything. A "wrong" type still runs. Enforcement is the job of a <strong>static type checker</strong> (<code>mypy</code>) before you run, or a validation library (Pydantic) at runtime.</div><div class="code-block"><span class="ck">def</span> <span class="cm">add</span>(a: <span class="cm">int</span>, b: <span class="cm">int</span>) -&gt; <span class="cm">int</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">return</span> a + b<br><br><span class="cm">add</span>(<span class="cs">"x"</span>, <span class="cs">"y"</span>)   <span class="cc"># 'xy' — hints ignored; str + str just works, no TypeError</span><br><br>y: <span class="cm">int</span> = <span class="cs">"actually a string"</span>   <span class="cc"># runs fine — the annotation is advisory</span></div><div class="output-block">xy<br>(no error — the int hint on y is not enforced)</div><div class="why-box"><strong>Why it matters:</strong> hints are a <em>contract you and your tools</em> agree on, not a runtime guard. <code>mypy add("x", "y")</code> would flag the error <em>before</em> running; the interpreter never will. This is the deliberate trade — keep Python dynamic, add optional static safety on top. (When you <em>do</em> want runtime enforcement, that's Pydantic — Chunk D.)</div><hr class="divider"><div style="background:#1e1e2e; border-left:4px solid #cba6f7; padding:16px 20px; border-radius:8px; font-family:monospace; color:#cdd6f4;"><h4 style="color:#cba6f7; margin:0 0 12px 0;">🔑 Chunk A — Key Takeaways</h4><ul style="margin:0; padding-left:20px; line-height:2.1"><li>Type hints annotate expected types; they're optional and Python stays dynamic</li><li>Function: <code>def f(a: int) -&gt; str</code>; stored in <code>__annotations__</code> (how frameworks read them)</li><li>Parameterize containers: <code>list[int]</code>, <code>dict[str, int]</code>, <code>tuple[int, ...]</code></li><li>NOT enforced at runtime — value comes from IDEs, <code>mypy</code>, and libraries (Pydantic) that read them</li></ul></div>

<div class="chunk-badge">Part 1 · Chunk B — The Typing Toolkit</div><h3 class="sub">🔹 1.5 &nbsp;<code>Optional[X]</code> / <code>X | None</code> — "maybe a value"</h3><div class="theory-box">A value that could be its type <em>or</em> <code>None</code>. <code>Optional[int]</code> is exactly <code>Union[int, None]</code>; the modern spelling (3.10+) is <code>int | None</code>. Use it for "might be missing" — the honest signature for a lookup that can fail.</div><div class="code-block"><span class="ck">from</span> typing <span class="ck">import</span> Optional<br><span class="ck">def</span> <span class="cm">find</span>(xs: <span class="cm">list</span>[<span class="cm">int</span>], target: <span class="cm">int</span>) -&gt; Optional[<span class="cm">int</span>]:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">return</span> xs.<span class="cm">index</span>(target) <span class="ck">if</span> target <span class="ck">in</span> xs <span class="ck">else</span> <span class="ck">None</span><br><br><span class="cm">find</span>([<span class="cn">1</span>,<span class="cn">2</span>,<span class="cn">3</span>], <span class="cn">2</span>), <span class="cm">find</span>([<span class="cn">1</span>,<span class="cn">2</span>,<span class="cn">3</span>], <span class="cn">9</span>)   <span class="cc"># (1, None)</span><br>Optional[<span class="cm">int</span>] == Union[<span class="cm">int</span>, <span class="ck">None</span>]         <span class="cc"># True</span><br><br><span class="ck">def</span> <span class="cm">g</span>(x: <span class="cm">int</span> | <span class="ck">None</span> = <span class="ck">None</span>) -&gt; <span class="cm">int</span>:   <span class="cc"># modern spelling</span><br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">return</span> x <span class="ck">if</span> x <span class="ck">is</span> <span class="ck">not</span> <span class="ck">None</span> <span class="ck">else</span> <span class="cn">0</span></div><div class="output-block">1 None<br>Optional[int] == int | None: True</div><h3 class="sub">🔹 1.6 &nbsp;<code>Union</code> / <code>|</code> and <code>Any</code></h3><div class="theory-box"><code>Union[int, str]</code> (or <code>int | str</code>) means "one of these types." <code>Any</code> is the <strong>escape hatch</strong> — it turns off type checking for that value; use it sparingly, since it defeats the purpose.</div><div class="code-block"><span class="ck">def</span> <span class="cm">parse</span>(x: <span class="cm">int</span> | <span class="cm">str</span>) -&gt; <span class="cm">str</span>:   <span class="cc"># accepts either</span><br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">return</span> <span class="cm">str</span>(x)<br><span class="cm">parse</span>(<span class="cn">5</span>), <span class="cm">parse</span>(<span class="cs">"a"</span>)   <span class="cc"># ('5', 'a')</span><br><br><span class="ck">from</span> typing <span class="ck">import</span> Any<br><span class="ck">def</span> <span class="cm">passthrough</span>(x: Any) -&gt; Any: <span class="ck">return</span> x   <span class="cc"># checker won't complain about anything</span></div><div class="output-block">5 a</div><div class="note-box">💡 <code>Any</code> silences the checker entirely — a <code>list[Any]</code> or a stray <code>Any</code> can quietly disable safety across your code. Prefer a real <code>Union</code>/generic; reach for <code>Any</code> only at genuine dynamic boundaries.</div>

<h3 class="sub">🔹 1.7 &nbsp;<code>Callable</code> — function types</h3><div class="theory-box"><code>Callable[[ArgTypes], ReturnType]</code> annotates something you'll call — a callback, a key function, a decorator target. <code>Callable[[int], int]</code> is "takes an int, returns an int."</div><div class="code-block"><span class="ck">from</span> typing <span class="ck">import</span> Callable<br><span class="ck">def</span> <span class="cm">apply</span>(f: Callable[[<span class="cm">int</span>], <span class="cm">int</span>], x: <span class="cm">int</span>) -&gt; <span class="cm">int</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">return</span> f(x)<br><br><span class="cm">apply</span>(<span class="ck">lambda</span> v: v * <span class="cn">2</span>, <span class="cn">5</span>)   <span class="cc"># 10</span></div><div class="output-block">10</div><h3 class="sub">🔹 1.8 &nbsp;Generics &amp; <code>TypeVar</code> — preserve the element type</h3><div class="theory-box">A <code>TypeVar</code> is a placeholder that links input and output types. <code>def first(xs: list[T]) -&gt; T</code> tells the checker: give it a <code>list[int]</code> and you get an <code>int</code> back; a <code>list[str]</code> → a <code>str</code>. That's <strong>generic</strong> code — type-safe without hard-coding one type.</div><div class="code-block"><span class="ck">from</span> typing <span class="ck">import</span> TypeVar<br>T = TypeVar(<span class="cs">"T"</span>)<br><span class="ck">def</span> <span class="cm">first</span>(xs: <span class="cm">list</span>[T]) -&gt; T:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">return</span> xs[<span class="cn">0</span>]<br><br><span class="cm">first</span>([<span class="cn">1</span>, <span class="cn">2</span>, <span class="cn">3</span>]), <span class="cm">first</span>([<span class="cs">"a"</span>, <span class="cs">"b"</span>])   <span class="cc"># (1, 'a') - int in -> int out, str in -> str out</span></div><div class="output-block">1 a</div><div class="info-box">💡 Generics are how <code>list[T]</code>, <code>dict[K, V]</code>, and typed containers stay precise. In modern Python you can also write <code>def first[T](xs: list[T]) -&gt; T</code> (3.12+ syntax) — no explicit <code>TypeVar</code> needed.</div>

<h3 class="sub">🔹 1.9 &nbsp;<code>Literal</code> and type aliases</h3><div class="theory-box"><code>Literal["r", "w", "a"]</code> restricts a value to a fixed set of constants — great for modes/flags. A <strong>type alias</strong> gives a complex type a readable name you can reuse.</div><div class="code-block"><span class="ck">from</span> typing <span class="ck">import</span> Literal<br><span class="ck">def</span> <span class="cm">open_file</span>(mode: Literal[<span class="cs">"r"</span>, <span class="cs">"w"</span>, <span class="cs">"a"</span>]) -&gt; <span class="cm">str</span>:   <span class="cc"># only these 3 allowed</span><br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">return</span> <span class="cs">f"mode={mode}"</span><br><span class="cm">open_file</span>(<span class="cs">"r"</span>)   <span class="cc"># 'mode=r'  (a checker rejects open_file("x"))</span><br><br>Vector = <span class="cm">list</span>[<span class="cm">float</span>]                    <span class="cc"># type alias - a readable name</span><br><span class="ck">def</span> <span class="cm">scale</span>(v: Vector, k: <span class="cm">float</span>) -&gt; Vector:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">return</span> [x * k <span class="ck">for</span> x <span class="ck">in</span> v]<br><span class="cm">scale</span>([<span class="cn">1.0</span>, <span class="cn">2.0</span>], <span class="cn">3</span>)   <span class="cc"># [3.0, 6.0]</span></div><div class="output-block">mode=r<br>[3.0, 6.0]</div><div class="note-box">💡 <code>Literal</code> is how a checker knows only certain string flags are valid — the typed version of an enum-ish constraint. Aliases (<code>Vector = list[float]</code>) keep long generic types readable and consistent across a codebase.</div><hr class="divider"><div style="background:#1e1e2e; border-left:4px solid #cba6f7; padding:14px 18px; border-radius:8px; font-family:monospace; color:#cdd6f4;"><h4 style="color:#cba6f7; margin:0 0 10px 0;">🔑 Chunk B — Key Takeaways</h4><ul style="margin:0; padding-left:20px; line-height:2.0"><li><code>Optional[X]</code> = <code>X | None</code> — a value that may be missing</li><li><code>Union[A, B]</code> / <code>A | B</code> = one of several types; <code>Any</code> = escape hatch (use sparingly)</li><li><code>Callable[[args], ret]</code> types functions; <code>TypeVar</code> writes generic, type-preserving code</li><li><code>Literal[...]</code> restricts to fixed constants; type aliases name complex types</li></ul></div>

<div class="chunk-badge">Part 1 · Chunk C — Dataclasses</div><h3 class="sub">🔹 1.10 &nbsp;<code>@dataclass</code> — kill the boilerplate</h3><div class="theory-box">A <strong>dataclass</strong> turns annotated fields into a full class: the decorator auto-generates <code>__init__</code>, <code>__repr__</code>, and <code>__eq__</code> for you. You write the <em>shape</em>; Python writes the plumbing. Perfect for config objects, records, model hyperparameters — anything that's mostly "a bundle of typed fields."</div><div class="code-block"><span class="ck">from</span> dataclasses <span class="ck">import</span> dataclass, field<br><br><span class="cn">@dataclass</span><br><span class="ck">class</span> <span class="cm">Point</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;x: <span class="cm">int</span><br>&nbsp;&nbsp;&nbsp;&nbsp;y: <span class="cm">int</span><br><br>p = <span class="cm">Point</span>(<span class="cn">1</span>, <span class="cn">2</span>)<br>p                       <span class="cc"># Point(x=1, y=2)   <- free __repr__</span><br>p == <span class="cm">Point</span>(<span class="cn">1</span>, <span class="cn">2</span>)         <span class="cc"># True             <- free __eq__ (field-by-field)</span></div><div class="output-block">repr: Point(x=1, y=2)<br>p == Point(1, 2): True&nbsp;&nbsp;|&nbsp;&nbsp;p == Point(3, 4): False</div><div class="info-box">💡 Compare to a hand-written class: no <code>def __init__(self, x, y): self.x = x; self.y = y</code>, no repr, no eq. The dataclass gives all three from four lines of field declarations — and the annotations double as real type hints.</div><h3 class="sub">🔹 1.11 &nbsp;Defaults &amp; <code>default_factory</code> — the mutable-default fix</h3><div class="theory-box">Fields can have defaults. But a <strong>mutable</strong> default (<code>list</code>, <code>dict</code>, <code>set</code>) can't be written inline — Python <em>refuses</em> it at class-creation time to prevent the shared-default bug. Use <code>field(default_factory=list)</code>: a fresh object is built per instance.</div><div class="code-block"><span class="cn">@dataclass</span><br><span class="ck">class</span> <span class="cm">Config</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;lr: <span class="cm">float</span> = <span class="cn">0.01</span>                       <span class="cc"># immutable default - fine inline</span><br>&nbsp;&nbsp;&nbsp;&nbsp;layers: <span class="cm">list</span>[<span class="cm">int</span>] = field(default_factory=<span class="cm">list</span>)  <span class="cc"># mutable -> factory</span><br><br>c1, c2 = <span class="cm">Config</span>(), <span class="cm">Config</span>()<br>c1.layers.<span class="cm">append</span>(<span class="cn">1</span>)<br>c1.layers, c2.layers    <span class="cc"># ([1], [])  <- independent, NOT shared</span></div><div class="output-block">c1.layers, c2.layers -&gt; [1] []</div><div class="trap-block">⚠️ Writing <code>layers: list = []</code> directly raises <code>ValueError: mutable default &lt;class 'list'&gt; for field ... is not allowed</code> <em>at class definition</em>. This is the same shared-mutable-default trap from <span style="color:#89dceb">Session 1</span> / <span style="color:#89dceb">7A</span> — dataclasses catch it for you and force <code>default_factory</code>.</div>

<h3 class="sub">🔹 1.12 &nbsp;<code>frozen=True</code> — immutable &amp; hashable</h3><div class="theory-box"><code>@dataclass(frozen=True)</code> makes instances <strong>read-only</strong>: assigning to a field raises <code>FrozenInstanceError</code>. As a bonus, frozen dataclasses become <strong>hashable</strong> — usable as <code>dict</code> keys and <code>set</code> members (ties back to <span style="color:#89dceb">7B</span>: immutability enables a stable hash).</div><div class="code-block"><span class="cn">@dataclass</span>(frozen=<span class="ck">True</span>)<br><span class="ck">class</span> <span class="cm">Coord</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;x: <span class="cm">int</span><br>&nbsp;&nbsp;&nbsp;&nbsp;y: <span class="cm">int</span><br><br>c = <span class="cm">Coord</span>(<span class="cn">1</span>, <span class="cn">2</span>)<br>{c: <span class="cs">"home"</span>}[<span class="cm">Coord</span>(<span class="cn">1</span>, <span class="cn">2</span>)]   <span class="cc"># 'home'  <- hashable, works as a dict key</span><br>c.x = <span class="cn">9</span>                    <span class="cc"># FrozenInstanceError</span></div><div class="output-block">used as dict key -&gt; home<br>c.x = 9 -&gt; FrozenInstanceError</div><h3 class="sub">🔹 1.13 &nbsp;<code>order=True</code> — free comparisons &amp; sorting</h3><div class="theory-box"><code>@dataclass(order=True)</code> generates <code>__lt__</code>, <code>__le__</code>, <code>__gt__</code>, <code>__ge__</code> that compare instances <strong>field-by-field, top to bottom</strong> (like a tuple). Instant sortability — no hand-written <code>__lt__</code>.</div><div class="code-block"><span class="cn">@dataclass</span>(order=<span class="ck">True</span>)<br><span class="ck">class</span> <span class="cm">Version</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;major: <span class="cm">int</span><br>&nbsp;&nbsp;&nbsp;&nbsp;minor: <span class="cm">int</span><br><br><span class="cm">sorted</span>([<span class="cm">Version</span>(<span class="cn">2</span>,<span class="cn">0</span>), <span class="cm">Version</span>(<span class="cn">1</span>,<span class="cn">5</span>), <span class="cm">Version</span>(<span class="cn">1</span>,<span class="cn">2</span>)])   <span class="cc"># sorts by (major, minor)</span></div><div class="output-block">[Version(major=1, minor=2), Version(major=1, minor=5), Version(major=2, minor=0)]</div><div class="note-box">💡 Field order matters: comparison walks fields top-to-bottom, so <code>major</code> dominates <code>minor</code> — exactly tuple ordering. Combine <code>frozen=True, order=True</code> for an immutable, hashable, sortable value object.</div><hr class="divider"><div style="background:#1e1e2e; border-left:4px solid #cba6f7; padding:14px 18px; border-radius:8px; font-family:monospace; color:#cdd6f4;"><h4 style="color:#cba6f7; margin:0 0 10px 0;">🔑 Chunk C — Key Takeaways</h4><ul style="margin:0; padding-left:20px; line-height:2.0"><li><code>@dataclass</code> auto-writes <code>__init__</code>, <code>__repr__</code>, <code>__eq__</code> from typed fields</li><li>Mutable defaults need <code>field(default_factory=list)</code> — inline <code>= []</code> is a <code>ValueError</code></li><li><code>frozen=True</code> → read-only + hashable (valid dict key / set member)</li><li><code>order=True</code> → tuple-style field-by-field comparisons → instant sorting</li></ul></div>

<div class="chunk-badge">Part 1 · Chunk D — Advanced Typing, the Pydantic Bridge, Examples &amp; Edge Cases</div><h3 class="sub">🔹 1.14 &nbsp;<code>TypedDict</code> — a typed shape for a dict</h3><div class="theory-box">JSON payloads and config blobs arrive as <code>dict</code>s. <code>TypedDict</code> lets you declare the expected keys and their types — the checker verifies you access real keys with the right types. At runtime it's still a plain <code>dict</code> (zero overhead).</div><div class="code-block"><span class="ck">from</span> typing <span class="ck">import</span> TypedDict<br><span class="ck">class</span> <span class="cm">Movie</span>(TypedDict):<br>&nbsp;&nbsp;&nbsp;&nbsp;title: <span class="cm">str</span><br>&nbsp;&nbsp;&nbsp;&nbsp;year: <span class="cm">int</span><br><br>m: Movie = {<span class="cs">"title"</span>: <span class="cs">"Arrival"</span>, <span class="cs">"year"</span>: <span class="cn">2016</span>}<br>m[<span class="cs">"title"</span>], m[<span class="cs">"year"</span>]     <span class="cc"># Arrival 2016</span><br><span class="cm">isinstance</span>(m, <span class="cm">dict</span>)          <span class="cc"># True - it IS a dict at runtime</span></div><div class="output-block">Arrival 2016&nbsp;&nbsp;|&nbsp;&nbsp;isinstance(m, dict): True</div><h3 class="sub">🔹 1.15 &nbsp;<code>Protocol</code> — structural (duck) typing, checked</h3><div class="theory-box">A <code>Protocol</code> says "anything with these methods qualifies" — no inheritance needed. It's <a style="color:#89dceb">duck typing</a> that a type checker can actually verify: if it has <code>__len__</code>, it's <code>Sized</code>. This is how you type "any file-like / any model-like object" without a shared base class.</div><div class="code-block"><span class="ck">from</span> typing <span class="ck">import</span> Protocol<br><span class="ck">class</span> <span class="cm">Sized</span>(Protocol):<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">def</span> <span class="cm">__len__</span>(self) -&gt; <span class="cm">int</span>: ...<br><br><span class="ck">def</span> <span class="cm">total</span>(x: <span class="cm">Sized</span>) -&gt; <span class="cm">int</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">return</span> <span class="cm">len</span>(x)<br><span class="cm">total</span>([<span class="cn">1</span>,<span class="cn">2</span>,<span class="cn">3</span>]), <span class="cm">total</span>(<span class="cs">"abcd"</span>)   <span class="cc"># 3 4 - list & str both satisfy it</span></div><div class="output-block">3 4</div><h3 class="sub">🔹 1.16 &nbsp;<code>NamedTuple</code> — typed, immutable record</h3><div class="theory-box">A typed <code>tuple</code> with named fields — immutable, hashable, indexable. Lighter than a dataclass when you just need a small fixed record and want tuple behavior (unpacking, indexing).</div><div class="code-block"><span class="ck">from</span> typing <span class="ck">import</span> NamedTuple<br><span class="ck">class</span> <span class="cm">Point</span>(NamedTuple):<br>&nbsp;&nbsp;&nbsp;&nbsp;x: <span class="cm">int</span><br>&nbsp;&nbsp;&nbsp;&nbsp;y: <span class="cm">int</span><br><br>p = <span class="cm">Point</span>(<span class="cn">1</span>, <span class="cn">2</span>)<br>p.x, p[<span class="cn">0</span>], <span class="cm">tuple</span>(p)     <span class="cc"># 1 1 (1, 2) - attr access AND tuple behavior</span></div><div class="output-block">1 1 (1, 2)</div><div class="info-box">💡 Rule of thumb: <code>NamedTuple</code> = immutable tuple-like record; <code>@dataclass</code> = mutable/rich object with methods and options; <code>TypedDict</code> = a plain dict you want typed (e.g. JSON). All three are "typed bundles of fields" for different runtime shapes.</div>

<h3 class="sub">🔹 1.17 &nbsp;Dataclass helpers — <code>asdict</code>, <code>replace</code>, <code>__post_init__</code>, <code>field</code> options</h3><div class="theory-box">The <code>dataclasses</code> module ships utilities: <code>asdict</code>/<code>astuple</code> serialize an instance, <code>replace</code> makes a modified <em>copy</em> (great for frozen objects), <code>__post_init__</code> runs after <code>__init__</code> for validation or derived fields, and <code>field(...)</code> tunes per-field behavior (<code>init=False</code>, <code>compare=False</code>, <code>repr=False</code>).</div><div class="code-block"><span class="ck">from</span> dataclasses <span class="ck">import</span> dataclass, field, asdict, replace<br><span class="cn">@dataclass</span><br><span class="ck">class</span> <span class="cm">Circle</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;r: <span class="cm">float</span><br>&nbsp;&nbsp;&nbsp;&nbsp;area: <span class="cm">float</span> = field(init=<span class="ck">False</span>)   <span class="cc"># derived, not a constructor arg</span><br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">def</span> <span class="cm">__post_init__</span>(self):<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;self.area = <span class="cn">3.14159</span> * self.r ** <span class="cn">2</span><br><br>c = <span class="cm">Circle</span>(<span class="cn">2</span>)<br>c.area                  <span class="cc"># 12.56636  (computed in __post_init__)</span><br>asdict(c)               <span class="cc"># {'r': 2, 'area': 12.56636}</span><br>replace(c, r=<span class="cn">3</span>)          <span class="cc"># NEW Circle(r=3) - original untouched</span></div><div class="output-block">c.area -&gt; 12.56636<br>asdict(c) -&gt; {'r': 2, 'area': 12.56636}</div><h3 class="sub">🔹 1.18 &nbsp;The Pydantic bridge — hints that <em>actually enforce</em></h3><div class="theory-box">Recall <a style="color:#89dceb">1.4</a>: standard hints are <strong>not</strong> checked at runtime. <strong>Pydantic</strong> closes that gap. A <code>BaseModel</code> reads your annotations and <em>validates &amp; coerces</em> incoming data at construction — the foundation of FastAPI request bodies, LLM structured-output parsing, and tool-call schemas. This is where the whole session pays off for ML/GenAI work.</div><div class="warn-box">⚠️ <code>pydantic</code> is a third-party package (<code>pip install pydantic</code>) and is <strong>not</strong> installed in this course environment — so the block below is shown for study, not executed. Run it in a project venv to see it live.</div><div class="code-block"><span class="cc"># pip install pydantic</span><br><span class="ck">from</span> pydantic <span class="ck">import</span> BaseModel, ValidationError<br><br><span class="ck">class</span> <span class="cm">User</span>(BaseModel):<br>&nbsp;&nbsp;&nbsp;&nbsp;name: <span class="cm">str</span><br>&nbsp;&nbsp;&nbsp;&nbsp;age: <span class="cm">int</span><br><br><span class="cm">User</span>(name=<span class="cs">"Ada"</span>, age=<span class="cs">"30"</span>)   <span class="cc"># age coerced "30" -> 30 : User(name='Ada', age=30)</span><br><br><span class="ck">try</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="cm">User</span>(name=<span class="cs">"Ada"</span>, age=<span class="cs">"not-a-number"</span>)   <span class="cc"># raises at construction</span><br><span class="ck">except</span> ValidationError <span class="ck">as</span> e:<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="cm">print</span>(<span class="cs">"rejected: bad age"</span>)</div><div class="info-box">💡 <strong>The mental model:</strong> <code>@dataclass</code> = zero-cost typed container, no runtime checks. <code>pydantic.BaseModel</code> = same field-declaration style, but validates + coerces at the boundary. Use dataclasses for internal objects; use Pydantic where untrusted data enters (API payloads, LLM outputs, config files).</div>

<div class="ex-header"><h2>2. Examples</h2></div><h3 class="sub">📘 Example A — a typed, self-validating ML training config</h3><div class="theory-box">Frozen (immutable, hashable → usable as a cache/registry key), a <code>Literal</code> to constrain the optimizer, and <code>__post_init__</code> to reject bad hyperparameters at construction.</div><div class="code-block"><span class="ck">from</span> dataclasses <span class="ck">import</span> dataclass, asdict<br><span class="ck">from</span> typing <span class="ck">import</span> Literal<br><br><span class="cn">@dataclass</span>(frozen=<span class="ck">True</span>)<br><span class="ck">class</span> <span class="cm">TrainConfig</span>:<br>&nbsp;&nbsp;&nbsp;&nbsp;lr: <span class="cm">float</span> = <span class="cn">1e-3</span><br>&nbsp;&nbsp;&nbsp;&nbsp;optimizer: Literal[<span class="cs">"adam"</span>, <span class="cs">"sgd"</span>] = <span class="cs">"adam"</span><br>&nbsp;&nbsp;&nbsp;&nbsp;layers: <span class="cm">tuple</span>[<span class="cm">int</span>, ...] = (<span class="cn">128</span>, <span class="cn">64</span>)<br>&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">def</span> <span class="cm">__post_init__</span>(self):<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span class="ck">if</span> self.lr &lt;= <span class="cn">0</span>: <span class="ck">raise</span> <span class="cm">ValueError</span>(<span class="cs">"lr must be positive"</span>)<br><br>cfg = <span class="cm">TrainConfig</span>(lr=<span class="cn">0.01</span>, optimizer=<span class="cs">"sgd"</span>)<br>asdict(cfg)                 <span class="cc"># serialize for logging / experiment tracking</span><br><span class="cm">TrainConfig</span>(lr=-<span class="cn">1</span>)             <span class="cc"># ValueError: lr must be positive</span></div><div class="output-block">cfg.optimizer=sgd&nbsp;&nbsp;lr=0.01<br>asdict -&gt; {'lr': 0.01, 'optimizer': 'sgd', 'layers': (128, 64)}<br>TrainConfig(lr=-1) -&gt; ValueError: lr must be positive</div><h3 class="sub">📘 Example B — a <code>Protocol</code> as a model interface</h3><div class="theory-box">Any object with a <code>predict</code> method satisfies <code>Model</code> — no shared base class. <div class="warn-header"><h2>3. Edge Cases</h2></div><div class="edge-block"><p><strong>EC1 — <code>get_type_hints</code> resolves string/forward-ref annotations.</strong> Annotations written as strings (<code>x: "int"</code>) or under <code>from __future__ import annotations</code> stay as strings in <code>__annotations__</code>. <code>get_type_hints(f)</code> evaluates them into real types — which is why frameworks call it instead of reading <code>__annotations__</code> raw.<br><code>get_type_hints(f)</code> → <code>{'x': &lt;class 'int'&gt;, 'y': list[str], 'return': &lt;class 'NoneType'&gt;}</code></p></div><div class="edge-block"><p><strong>EC2 — <code>__post_init__</code> for derived fields.</strong> Fields marked <code>field(init=False)</code> aren't constructor args; compute them in <code>__post_init__</code>. <code>Circle(2).area</code> → <code>12.56636</code>, even though you never pass <code>area</code>.</p></div><div class="edge-block"><p><strong>EC3 — a plain <code>@dataclass</code> is <em>not</em> hashable.</strong> Because it gets <code>__eq__</code> but no <code>__hash__</code>, <code>hash(M1(1))</code> raises <code>TypeError: unhashable type: 'M1'</code>. You need <code>frozen=True</code> (or <code>eq=False</code>) to make it hashable — the same eq-without-hash rule from <a style="color:#89dceb">7B</a>.</p></div><div class="edge-block"><p><strong>EC4 — <code>field(compare=False)</code> excludes a field from <code>==</code>/ordering.</strong> <code>Rec(1, 100) == Rec(1, 999)</code> → <code>True</code> when <code>ts</code> is <code>compare=False</code>: two records with the same <code>id</code> are "equal" regardless of timestamp.</p></div><div class="edge-block"><p><strong>EC5 — dataclasses do NOT enforce types (the big gotcha).</strong> <code>@dataclass class Wrong: n: int</code> then <code>Wrong("not an int")</code> → <code>Wrong(n='not an int')</code>, no error. Annotations are documentation the runtime ignores (<a style="color:#89dceb">1.4</a>). If you need enforcement, that's exactly what <strong>Pydantic</strong> (1.18) is for.</p></div>

<div class="part-header"><h2>Part 2 · §4 — Golden Rules</h2></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">RULE 1</span>Hints document intent; they don't enforce it.</div><div class="body-txt">Python never checks annotations at runtime. Their value comes from IDEs, <code>mypy</code>/<code>pyright</code>, and libraries that read them. Treat them as machine-checkable documentation — and actually run a checker.</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">RULE 2</span>Prefer modern built-in generics.</div><div class="body-txt">Use <code>list[int]</code>, <code>dict[str, int]</code>, <code>int | None</code> (3.9+/3.10+) over the old <code>List</code>, <code>Dict</code>, <code>Optional</code> imports. Reach into <code>typing</code> only for <code>Callable</code>, <code>Protocol</code>, <code>TypeVar</code>, <code>Literal</code>, etc.</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">RULE 3</span>Reach for a dataclass when a class is "mostly fields."</div><div class="body-txt">If you're hand-writing <code>__init__</code>, <code>__repr__</code>, and <code>__eq__</code> to shuttle values around, a <code>@dataclass</code> does it in a few typed lines. Keep regular classes for rich behavior/complex invariants.</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">RULE 4</span>Never use a mutable default — use <code>default_factory</code>.</div><div class="body-txt">Dataclasses forbid inline <code>= []</code> for exactly the shared-default reason from Session 1. <code>field(default_factory=list)</code> builds a fresh object per instance.</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">RULE 5</span><code>frozen=True</code> when you want a value object.</div><div class="body-txt">Immutable → hashable → safe as dict keys, set members, and cache keys. Combine with <code>order=True</code> for sortable value objects. Update via <code>replace(obj, field=new)</code>, which returns a copy.</div></div><div class="rule-block"><div class="rule-title"><span class="rule-badge">RULE 6</span>Validate at the boundary with Pydantic; use dataclasses inside.</div><div class="body-txt">Untrusted input (API bodies, LLM output, config files) → <code>pydantic.BaseModel</code> (it validates + coerces). Trusted internal data → <code>@dataclass</code> (zero-cost). Don't rely on hints alone to catch bad data.</div></div>

<div class="trap-header"><h2>Part 2 · §5 — Common Traps</h2></div><div class="trap-block"><div class="trap-title"><span class="trap-badge">TRAP 1</span>Expecting hints to validate at runtime.</div><div class="body-txt"><code>def f(x: int)</code> then <code>f("hi")</code> runs happily — no error. Same for dataclass fields (Edge Case 5). If you need enforcement, add explicit checks or use Pydantic.</div></div><div class="trap-block"><div class="trap-title"><span class="trap-badge">TRAP 2</span>Mutable dataclass default → <code>ValueError</code> at import.</div><div class="body-txt"><code>items: list = []</code> doesn't wait for an instance — it raises when the class is <em>defined</em>. The fix is always <code>field(default_factory=list)</code>.</div></div><div class="trap-block"><div class="trap-title"><span class="trap-badge">TRAP 3</span>Assuming every dataclass is hashable.</div><div class="body-txt">A default dataclass has <code>__eq__</code> but no <code>__hash__</code>, so it's <strong>unhashable</strong> — <code>set</code>/dict-key use raises <code>TypeError</code>. Add <code>frozen=True</code> to get a hash.</div></div><div class="trap-block"><div class="trap-title"><span class="trap-badge">TRAP 4</span>Field order changes <code>order=True</code> sorting silently.</div><div class="body-txt">Comparison walks fields top-to-bottom like a tuple. Put a low-priority field above a high-priority one and your sort is wrong — with no error. Order fields deliberately (or use <code>field(compare=False)</code> to exclude one).</div></div><div class="trap-block"><div class="trap-title"><span class="trap-badge">TRAP 5</span><code>Any</code> silently disables checking.</div><div class="body-txt">One <code>Any</code> (or an untyped function) propagates: values flowing through it are unchecked downstream. It reads as "typed" but isn't. Prefer a real <code>Union</code>/generic; reserve <code>Any</code> for genuine dynamic boundaries.</div></div><div class="trap-block"><div class="trap-title"><span class="trap-badge">TRAP 6</span>Mutating a "frozen" object's mutable field.</div><div class="body-txt"><code>frozen=True</code> blocks <em>rebinding</em> a field, but if a field holds a <code>list</code> you can still <code>.append</code> to it — and that also breaks the hash contract. Store immutable values (tuples) in frozen dataclasses.</div></div>

<div class="ml-header"><h2>Part 3 · §7 — ML / GenAI in the Real World</h2></div><div class="ml-block"><div class="ml-title"><span class="ml-badge">ML 1</span>Pydantic is the schema layer of modern GenAI</div><div class="body-txt">FastAPI request/response bodies, LangChain/LlamaIndex structured output, and LLM <em>tool-call arguments</em> are all defined as Pydantic <code>BaseModel</code>s. The model returns JSON; Pydantic validates &amp; coerces it into typed objects. The annotation skills from this session are exactly what you write there.</div></div><div class="ml-block"><div class="ml-title"><span class="ml-badge">ML 2</span>Dataclasses for configs &amp; experiment tracking</div><div class="body-txt">Hyperparameter configs are naturally frozen dataclasses: immutable (reproducibility), hashable (cache/registry key), <code>asdict()</code>-serializable (log to MLflow/W&amp;B). HuggingFace <code>TrainingArguments</code> is a dataclass; PyTorch Lightning and Hydra lean on them heavily.</div></div><div class="ml-block"><div class="ml-title"><span class="ml-badge">ML 3</span><code>Protocol</code> for framework-agnostic model code</div><div class="body-txt">A <code>Predictor</code> Protocol (<code>def predict(self, x) -&gt; ...</code>) lets your evaluation/serving code accept a scikit-learn estimator, a torch module wrapper, or a custom class interchangeably — duck typing the checker can verify, no shared base class.</div></div><div class="ml-block"><div class="ml-title"><span class="ml-badge">ML 4</span>Typed tensors &amp; <code>Literal</code> for modes</div><div class="body-txt">Type aliases (<code>Vector = list[float]</code>, or jaxtyping/nptyping for shapes) document data flow; <code>Literal["train", "eval"]</code> and <code>Literal["adam", "sgd"]</code> constrain mode/optimizer flags so a checker catches typos before a 12-hour run does.</div></div><div class="ml-block"><div class="ml-title"><span class="ml-badge">ML 5</span>Hints power the tooling you already use</div><div class="body-txt">IDE autocomplete on a <code>DataFrame</code> or model object, <code>mypy</code> in CI catching a <code>None</code> before prod, and auto-generated API docs all read your annotations. On a team, typed code is the difference between "readable" and "guess-and-check."</div></div>

<div class="interview-header"><h2>Part 3 · §8 — Interview Q&amp;A</h2></div><div class="sub-header"><h3>§8a — Concept Questions</h3></div><div class="qa-block"><div class="qa-q"><span class="q-num">1</span>Are Python type hints enforced at runtime?</div><div class="qa-a">No. The interpreter ignores them for execution — <code>f("x")</code> on <code>def f(x: int)</code> runs fine. They're read by IDEs, static checkers (<code>mypy</code>, <code>pyright</code>), and libraries (Pydantic, dataclasses). Enforcement requires an explicit check or a validating library.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">2</span>What does <code>@dataclass</code> generate for you?</div><div class="qa-a"><code>__init__</code>, <code>__repr__</code>, and <code>__eq__</code> by default, derived from the annotated fields. Options add more: <code>order=True</code> → comparison methods, <code>frozen=True</code> → immutability + <code>__hash__</code>.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">3</span>Why can't you write <code>x: list = []</code> as a dataclass default?</div><div class="qa-a">A single mutable default would be shared across all instances (the Session-1 trap). Dataclasses detect this and raise <code>ValueError</code> at class definition, forcing <code>field(default_factory=list)</code>, which builds a fresh list per instance.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">4</span>When is a dataclass hashable?</div><div class="qa-a">When it has a <code>__hash__</code>. A default dataclass has <code>__eq__</code> but no hash → unhashable. <code>frozen=True</code> (with default <code>eq=True</code>) generates a hash based on the fields, making it usable as a dict key / set member.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">5</span><code>Optional[int]</code> vs <code>int | None</code>?</div><div class="qa-a">Identical meaning — a value that's an <code>int</code> or <code>None</code>. <code>Optional[int] == Union[int, None]</code>. <code>int | None</code> is the modern (3.10+) spelling and needs no import.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">6</span>What is a <code>Protocol</code>, and how does it differ from an ABC?</div><div class="qa-a">A <code>Protocol</code> defines structural typing: any object with the right methods/attributes conforms — no inheritance needed (checked statically). An ABC uses nominal typing: a class must explicitly subclass it. Protocols type "duck-typed" interfaces.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">7</span>dataclass vs <code>NamedTuple</code> vs <code>TypedDict</code> — when each?</div><div class="qa-a"><code>NamedTuple</code>: immutable tuple-like record. <code>@dataclass</code>: mutable/rich object with methods and options (<code>frozen</code>, <code>order</code>). <code>TypedDict</code>: a plain <code>dict</code> (e.g. JSON) you want typed — it's still a dict at runtime.</div></div><div class="qa-block"><div class="qa-q"><span class="q-num">8</span>Where does Pydantic fit relative to dataclasses?</div><div class="qa-a">Same field-declaration style, but Pydantic <em>validates and coerces</em> data at construction. Use dataclasses for trusted internal objects (zero cost); use Pydantic at boundaries where untrusted data enters — API payloads, LLM structured output, config files.</div></div><div class="sub-header"><h3>§8b — Code Challenges (8)</h3></div><div class="info-box">Solve in <code>01_typing.ipynb</code>; verified answers in <code>solutions.ipynb</code>.</div><div class="cc-block"><div class="cc-title"><span class="cc-badge badge-easy">C1 · Easy</span>Frozen + hashable de-dup</div><div class="body-txt"><code>Money(amount, currency)</code> as a frozen dataclass; put several (with duplicates) in a <code>set</code> and confirm de-duplication.</div></div><div class="cc-block"><div class="cc-title"><span class="cc-badge badge-easy">C2 · Easy</span>Type alias + function</div><div class="body-txt">Alias <code>JSONNum = list[float]</code>; write <code>normalize(v: JSONNum) -&gt; JSONNum</code> that divides each element by the sum.</div></div><div class="cc-block"><div class="cc-title"><span class="cc-badge badge-med">C3 · Med</span>Sort ignoring a field</div><div class="body-txt"><code>@dataclass(order=True) Task(priority, label)</code> with <code>label = field(compare=False)</code>; sort by priority and show two tasks compare equal regardless of label.</div></div><div class="cc-block"><div class="cc-title"><span class="cc-badge badge-med">C4 · Med</span>Validate in <code>__post_init__</code></div><div class="body-txt"><code>Percentage(value)</code> must be 0–100 — raise <code>ValueError</code> otherwise.</div></div><div class="cc-block"><div class="cc-title"><span class="cc-badge badge-med">C5 · Med</span>A generic <code>Stack[T]</code></div><div class="body-txt">Class with <code>push(x: T)</code> / <code>pop() -&gt; T</code> using <code>Generic[T]</code>; instantiate as <code>Stack[int]</code>.</div></div><div class="cc-block"><div class="cc-title"><span class="cc-badge badge-med">C6 · Med</span><code>Protocol</code> over many models</div><div class="body-txt"><code>SupportsPredict</code> Protocol; <code>score(models: list[SupportsPredict], x)</code> returns each model's prediction — works for unrelated classes.</div></div><div class="cc-block"><div class="cc-title"><span class="cc-badge badge-hard">C7 · Hard</span>Immutable update with <code>replace</code></div><div class="body-txt">Frozen <code>HyperParams(lr, epochs)</code>; <code>with_lr(cfg, lr)</code> returns a NEW config (original untouched) via <code>dataclasses.replace</code>.</div></div><div class="cc-block"><div class="cc-title"><span class="cc-badge badge-hard">C8 · Hard</span>Hand-rolled schema validation (Pydantic-lite)</div><div class="body-txt">Given a <code>TypedDict</code>, write <code>validate(d, schema)</code> that checks every declared key is present with the right <code>isinstance</code> type — the idea Pydantic industrializes.</div></div>
