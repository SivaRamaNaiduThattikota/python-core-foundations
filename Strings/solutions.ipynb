### Exercises — Solutions

```python
# E1 (Easy) — Reverse word order
def reverse_words(s):
    return " ".join(s.split()[::-1])

print(reverse_words("the cat sat"))   # sat cat the
```

```python
# E2 (Easy) — Count vowels (case-insensitive)
def count_vowels(s):
    return sum(1 for c in s.lower() if c in "aeiou")

print(count_vowels("Education"))       # 5
```

```python
# E3 (Easy) — Title-case a name (avoids str.title() apostrophe quirks)
def title_name(s):
    return " ".join(w.capitalize() for w in s.split())

print(title_name("siva rama naidu"))   # Siva Rama Naidu
```

```python
# E4 (Easy) — File extension
def extension(name):
    return name.rsplit(".", 1)[-1]      # split from the right, last piece

print(extension("report.final.csv"))   # csv
```

```python
# E5 (Medium) — Clean CSV fields
def clean_fields(s):
    return [x.strip() for x in s.split(",")]

print(clean_fields("a, b ,c , d"))      # ['a', 'b', 'c', 'd']
```

```python
# E6 (Medium) — Valid palindrome (alnum, case-insensitive)
def is_palindrome(s):
    t = [c.lower() for c in s if c.isalnum()]
    return t == t[::-1]

print(is_palindrome("A man, a plan, a canal: Panama"))  # True
print(is_palindrome("race a car"))                      # False
```

```python
# E7 (Medium) — Most frequent character (ties -> first-seen)
from collections import Counter
def most_frequent(s):
    return Counter(s).most_common(1)[0][0]

print(most_frequent("mississippi"))    # i
```

```python
# E8 (Medium) — Aligned report row
def report_row(name, score):
    return f"{name:<10}{score:>6.1%}"

print(report_row("precision", 0.8734)) # precision  87.3%
```

```python
# E9 (Medium) — Caesar cipher (wrap, keep case, leave non-letters)
def caesar(s, k):
    out = []
    for ch in s:
        if ch.isalpha():
            base = ord("A") if ch.isupper() else ord("a")
            out.append(chr((ord(ch) - base + k) % 26 + base))
        else:
            out.append(ch)
    return "".join(out)

print(caesar("xyz", 3), caesar("Hello, World!", 3))  # abc  Khoor, Zruog!
```

```python
# E10 (Hard) — Run-length encode
def rle(s):
    if not s:
        return ""
    out, prev, n = [], s[0], 1
    for ch in s[1:]:
        if ch == prev:
            n += 1
        else:
            out.append(prev + str(n)); prev, n = ch, 1
    out.append(prev + str(n))            # flush the final run
    return "".join(out)

print(rle("aaabbc"))                     # a3b2c1
```

```python
# E11 (Hard) — Longest repeated-character run
def longest_run(s):
    if not s:
        return ("", 0)
    best_ch, best_n = s[0], 1
    ch, n = s[0], 1
    for c in s[1:]:
        n = n + 1 if c == ch else 1
        if c != ch:
            ch = c
        if n > best_n:
            best_ch, best_n = ch, n
    return (best_ch, best_n)

print(longest_run("aaabbbbcc"))          # ('b', 4)
```

```python
# E12 (Hard) — Reverse only the vowels (two pointers)
def reverse_vowels(s):
    vowels = set("aeiouAEIOU")
    chars = list(s)
    i, j = 0, len(chars) - 1
    while i < j:
        if chars[i] not in vowels:
            i += 1
        elif chars[j] not in vowels:
            j -= 1
        else:
            chars[i], chars[j] = chars[j], chars[i]
            i += 1; j -= 1
    return "".join(chars)

print(reverse_vowels("leetcode"))        # leotcede
```

### Code Challenges — Solutions

```python
# C1 (Easy) — Reverse a string
def reverse(s):
    return s[::-1]

print(reverse("hello"))                  # olleh
```

```python
# C2 (Easy) — Anagram check
def is_anagram(a, b):
    return sorted(a) == sorted(b)        # or Counter(a) == Counter(b) for O(n)

print(is_anagram("listen", "silent"))    # True
```

```python
# C3 (Medium) — First non-repeating character (2D seam: counts + order)
from collections import Counter
def first_unique(s):
    c = Counter(s)
    for ch in s:
        if c[ch] == 1:
            return ch
    return None

print(first_unique("leetcode"), first_unique("aabb"))   # l None
```

```python
# C4 (Medium) — Word frequency
from collections import Counter
def word_freq(text):
    return Counter(text.split())

print(word_freq("the cat the dog the"))  # Counter({'the': 3, 'cat': 1, 'dog': 1})
```

```python
# C5 (Medium) — Valid palindrome
def is_palindrome(s):
    t = [c.lower() for c in s if c.isalnum()]
    return t == t[::-1]

print(is_palindrome("A man, a plan, a canal: Panama"))  # True
```

```python
# C6 (Medium) — Run-length encode
def rle(s):
    if not s:
        return ""
    out, prev, n = [], s[0], 1
    for ch in s[1:]:
        if ch == prev:
            n += 1
        else:
            out.append(prev + str(n)); prev, n = ch, 1
    out.append(prev + str(n))
    return "".join(out)

print(rle("aaabbc"))                     # a3b2c1
```

```python
# C7 (Hard) — Longest substring without repeating characters (sliding window + dict)
def length_of_longest(s):
    seen, start, best = {}, 0, 0
    for i, ch in enumerate(s):
        if ch in seen and seen[ch] >= start:
            start = seen[ch] + 1         # jump past the repeat
        seen[ch] = i
        best = max(best, i - start + 1)
    return best

print(length_of_longest("abcabcbb"), length_of_longest("bbbbb"), length_of_longest("pwwkew"))  # 3 1 3
```

```python
# C8 (Hard) — Group anagrams (sorted signature -> dict key)
from collections import defaultdict
def group_anagrams(words):
    g = defaultdict(list)
    for w in words:
        g["".join(sorted(w))].append(w)
    return list(g.values())

print(group_anagrams(["eat","tea","tan","ate","nat","bat"]))
```
