<h1 class="main-title">🐍 Session 2B — Tuple (Deep)</h1>

<div class="info-box">
  <strong>Part 1:</strong> Theory → Example → Edge Cases<br>
  This is the second sequence sub-session. It's the mirror image of 2A: where a list is a mutable dynamic array, a tuple is the immutable counterpart — and almost everything interesting about tuples flows from that one difference. Because tuples are simpler than lists (two methods, not eleven), this is a sharper, shorter session.
</div>

<hr class="divider">

<!-- ══ SECTION 1: THEORY ══ -->
<div class="part-header"><h2>1. Theory</h2></div>

<h3 class="sub">1.1 What a tuple is — a fixed-size, immutable array of references</h3>

<div class="theory-box">
  <p style="margin:0 0 10px 0">A tuple is an ordered collection of object references, just like a list — but with one defining constraint: once created, you cannot change which objects its slots point at. No append, no insert, no reassignment of a slot, no removal. <strong style="color:#cba6f7">The sequence of pointers is frozen at creation.</strong></p>
  <p style="margin:0 0 10px 0"><strong>Connecting to Session 1:</strong> a tuple is immutable the same way an int or a string is. "Changing" a tuple doesn't exist — any operation that looks like a change (<span class="ic">+</span>, <span class="ic">*</span>) builds a new tuple and rebinds your name to it. The original is untouched.</p>
  <p style="margin:0"><strong>SQL/Power BI anchor:</strong> think of a tuple like a single row returned from a query — a fixed set of fields in a fixed order. You read <span class="ic">row[0]</span>, <span class="ic">row[1]</span>, but you don't "append a column" to one returned row. The shape is fixed; that's the contract. A list is more like a working table you keep modifying.</p>
</div>

<h3 class="sub">1.2 Why immutability matters — three concrete payoffs</h3>

<p class="body-p">Immutability isn't a restriction for its own sake. It buys three things lists can't offer:</p>

<div class="payoff-box">
  <div class="payoff-row">
    <span class="payoff-num">1</span>
    <span class="payoff-title">Hashability.</span>
    &nbsp;Because a tuple's contents can't change, Python can compute a stable hash for it that never goes stale. That makes a tuple usable as a <strong>dict key</strong> and a <strong>set member</strong> — lists cannot do this. This is the single biggest practical reason to reach for a tuple. Verified:
    <div class="code-block" style="margin:8px 0 0 0"><pre>d = {(<span class="cn">1</span>, <span class="cn">2</span>): <span class="cs">"point A"</span>}        <span class="cc"># tuple key — works</span>
s = {(<span class="cn">0</span>,<span class="cn">0</span>), (<span class="cn">1</span>,<span class="cn">1</span>), (<span class="cn">0</span>,<span class="cn">0</span>)}      <span class="cc"># tuple in a set, dedups to 2 elements</span></pre></div>
    <span style="color:#6c7086;font-size:0.88em">(Caveat in Edge cases: this only holds if everything inside the tuple is also hashable.)</span>
  </div>
  <div class="payoff-row">
    <span class="payoff-num">2</span>
    <span class="payoff-title">Memory efficiency.</span>
    &nbsp;A tuple is smaller than the equivalent list, because it's fixed-size — no over-allocation, no spare capacity, no growth bookkeeping. Verified on this machine:
    <div class="code-block" style="margin:8px 0 0 0"><pre>[<span class="cn">1</span>,<span class="cn">2</span>,<span class="cn">3</span>]  -&gt; <span class="cn">88</span> bytes
(<span class="cn">1</span>,<span class="cn">2</span>,<span class="cn">3</span>)  -&gt; <span class="cn">64</span> bytes    <span class="cc"># ~27% smaller</span></pre></div>
    For millions of small fixed records (coordinates, RGB values, feature pairs), tuples save real memory.
  </div>
  <div class="payoff-row">
    <span class="payoff-num">3</span>
    <span class="payoff-title">Safety / intent.</span>
    &nbsp;A tuple signals "this collection is not meant to change." Passing a tuple to a function guarantees the function can't accidentally mutate it (the silent-corruption trap from 2A's Trap 7 can't happen with a tuple's structure). It documents intent: these values belong together and are fixed.
  </div>
</div>

<h3 class="sub">1.3 The shallow-immutability caveat (Session 1 callback)</h3>

<p class="body-p">A tuple's immutability is <strong style="color:#f38ba8">shallow</strong> — it freezes which objects the slots point at, not the objects themselves. If a slot points at a mutable object (a list), that inner object can still be mutated:</p>

<div class="code-block"><pre>t = (<span class="cn">1</span>, [<span class="cn">2</span>, <span class="cn">3</span>], <span class="cs">"x"</span>)
t[<span class="cn">1</span>].<span class="cm">append</span>(<span class="cn">4</span>)        <span class="cc"># ALLOWED — mutating the inner list</span>
<span class="cm">print</span>(t)              <span class="cc"># (1, [2, 3, 4], 'x')</span></pre></div>

<div class="verified-box">✅ Verified. You can't reassign <code>t[1]</code> to a different object, but you can change the object it already points at. This is exactly the tuple-containing-a-list case from Session 1's Edge 4 — and it's why such a tuple is <strong>not hashable</strong> (its contents can effectively change).</div>

<h3 class="sub">1.4 Packing and unpacking — the tuple superpower</h3>

<p class="body-p">Tuples power some of Python's most-loved syntax, often invisibly.</p>

<div class="unpack-grid">
  <div class="unpack-card uc-1">
    <div class="uc-title">Packing</div>
    <div class="uc-body">Commas create a tuple — parentheses optional:</div>
  </div>
  <div class="unpack-card uc-2">
    <div class="uc-title">Unpacking</div>
    <div class="uc-body">Assign a tuple's elements to multiple names at once:</div>
  </div>
</div>

<div class="code-block"><pre>packed = <span class="cn">1</span>, <span class="cn">2</span>, <span class="cn">3</span>        <span class="cc"># this IS a tuple: (1, 2, 3)</span>
a, b, c = packed        <span class="cc"># a=1, b=2, c=3</span></pre></div>

<p class="body-p"><strong style="color:#89dceb">The swap idiom</strong> — the famous no-temp-variable swap is tuple packing + unpacking:</p>

<div class="code-block"><pre>x, y = y, x             <span class="cc"># right side packs (y,x), left side unpacks it</span></pre></div>

<p class="body-p"><strong style="color:#89dceb">Star unpacking</strong> — capture "the rest" into a list (you previewed this in 2A):</p>

<div class="code-block"><pre>first, *rest = (<span class="cn">1</span>, <span class="cn">2</span>, <span class="cn">3</span>, <span class="cn">4</span>)    <span class="cc"># first=1, rest=[2,3,4]</span></pre></div>

<p class="body-p"><strong style="color:#89dceb">Multiple return values</strong> — when a function "returns several things," it returns one tuple:</p>

<div class="code-block"><pre><span class="ck">def</span> <span class="cm">minmax</span>(seq):
    <span class="ck">return</span> <span class="cm">min</span>(seq), <span class="cm">max</span>(seq)   <span class="cc"># returns the tuple (min, max)</span>

lo, hi = <span class="cm">minmax</span>([<span class="cn">3</span>,<span class="cn">1</span>,<span class="cn">4</span>,<span class="cn">1</span>,<span class="cn">5</span>])    <span class="cc"># unpacks it</span></pre></div>

<div class="verified-box">✅ All verified. This is everywhere in Python — <code>enumerate</code> yields <code>(index, value)</code> tuples, <code>dict.items()</code> yields <code>(key, value)</code> tuples, and you unpack them in for loops without thinking about it.</div>

<h3 class="sub">1.5 namedtuple — readable records without a class</h3>

<p class="body-p">A plain tuple has a weakness: <span class="ic">p[0]</span> and <span class="ic">p[1]</span> don't tell you what they mean. <span class="ic">collections.namedtuple</span> fixes that — same immutable, hashable, memory-light tuple, but with named fields:</p>

<div class="code-block"><pre><span class="ck">from</span> collections <span class="ck">import</span> namedtuple
Point = <span class="cm">namedtuple</span>(<span class="cs">"Point"</span>, [<span class="cs">"x"</span>, <span class="cs">"y"</span>])
p = <span class="cm">Point</span>(<span class="cn">3</span>, <span class="cn">4</span>)
p.x        <span class="cc"># 3   — access by name</span>
p[<span class="cn">0</span>]       <span class="cc"># 3   — still works by index</span>
px, py = p <span class="cc"># 3 4 — still unpacks</span></pre></div>

<p class="body-p">Plus handy methods: <span class="ic">p._asdict()</span> → <span class="ic">{'x':3,'y':4}</span>, and <span class="ic">p._replace(x=99)</span> → a new <span class="ic">Point(x=99, y=4)</span> (immutable, so it returns a copy). Verified.</p>

<div class="info-box">
  💡 <strong>SQL anchor:</strong> a namedtuple is a named row — <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1;font-family:'Courier New',monospace">row.customer_id</code> instead of <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1;font-family:'Courier New',monospace">row[3]</code>. It's the lightweight choice for fixed records before you reach for a full class or a dataclass (Session 10).
</div>

<h3 class="sub">1.6 The Two Methods</h3>

<p class="body-p">A tuple has exactly <strong style="color:#cba6f7">two methods</strong> (vs. a list's eleven) — because you can't mutate it, so all the mutating methods are gone:</p>

<div class="method-block">
  <div class="method-title"><span class="method-badge">count(x)</span><span class="method-ret-val">returns int</span><span class="complexity">O(n)</span></div>
  <div class="method-body">How many times x appears in the tuple.</div>
  <div class="code-block" style="margin:8px 0 0 0"><pre>t = (<span class="cn">1</span>, <span class="cn">2</span>, <span class="cn">2</span>, <span class="cn">3</span>, <span class="cn">2</span>)
t.<span class="cm">count</span>(<span class="cn">2</span>)    <span class="cc"># 3  — verified</span></pre></div>
</div>

<div class="method-block">
  <div class="method-title"><span class="method-badge">index(x)</span><span class="method-ret-val">returns index</span><span class="complexity">O(n)</span></div>
  <div class="method-body">First position of x. Raises <span class="ic">ValueError</span> if not found.</div>
  <div class="code-block" style="margin:8px 0 0 0"><pre>t.<span class="cm">index</span>(<span class="cn">3</span>)    <span class="cc"># 3  — verified</span></pre></div>
</div>

<div class="note-box">💡 That's the entire method surface. The simplicity is the feature.</div>

<div class="ex-header"><h2>2. Example (runnable, step by step)</h2></div>

<div class="ex-block">
  <div class="ex-title"><span class="ex-badge">Example 1</span> Immutability: identity is fixed, slots can't be rebound</div>
  <div class="code-block"><pre>t = (<span class="cn">1</span>, <span class="cn">2</span>, <span class="cn">3</span>)
<span class="cm">print</span>(<span class="cm">id</span>(t))        <span class="cc"># some address</span>
t[<span class="cn">0</span>] = <span class="cn">99</span>           <span class="cc"># attempt to change slot 0</span></pre></div>
  <div class="output-block">
140119012556416<br>
TypeError: 'tuple' object does not support item assignment
  </div>
  <div class="step-box">
    <strong>Step by step:</strong> the tuple exists at a fixed address. Trying to rebind slot 0 raises <code>TypeError</code> — the slot-pointer sequence is frozen. Compare to a list, where <code>lst[0] = 99</code> happily rebinds the slot. This is the core behavioral difference in one line.
  </div>
</div>

<div class="ex-block">
  <div class="ex-title"><span class="ex-badge">Example 2</span> Hashability in action</div>
  <div class="code-block"><pre><span class="cc"># tuples can be dict keys and set members; lists cannot</span>
locations = {}
locations[(<span class="cn">40.7</span>, -<span class="cn">74.0</span>)] = <span class="cs">"New York"</span>     <span class="cc"># coordinate tuple as key</span>
locations[(<span class="cn">51.5</span>, -<span class="cn">0.1</span>)]  = <span class="cs">"London"</span>
<span class="cm">print</span>(locations[(<span class="cn">40.7</span>, -<span class="cn">74.0</span>)])            <span class="cc"># New York</span>

seen_points = {(<span class="cn">0</span>,<span class="cn">0</span>), (<span class="cn">1</span>,<span class="cn">1</span>), (<span class="cn">0</span>,<span class="cn">0</span>)}         <span class="cc"># set of tuples</span>
<span class="cm">print</span>(seen_points)                          <span class="cc"># {(0,0),(1,1)} — deduped</span></pre></div>
  <div class="output-block">New York<br>{(1, 1), (0, 0)}</div>
  <div class="step-box">
    This is the payoff that matters most. A coordinate pair, an <code>(row, col)</code> index, a <code>(user_id, item_id)</code> pair — anything you want to use as a dictionary key or track in a set must be hashable, and a tuple of immutables is. A list here would raise <code>TypeError: unhashable type: 'list'</code>.
  </div>
</div>

<div class="ex-block">
  <div class="ex-title"><span class="ex-badge">Example 3</span> Packing, unpacking, swap, multiple return</div>
  <div class="code-block"><pre>packed = <span class="cn">1</span>, <span class="cn">2</span>, <span class="cn">3</span>             <span class="cc"># packing (parens optional)</span>
a, b, c = packed             <span class="cc"># unpacking</span>
x, y = <span class="cn">10</span>, <span class="cn">20</span>
x, y = y, x                  <span class="cc"># swap → x=20, y=10</span>

<span class="ck">def</span> <span class="cm">divmod_demo</span>(n, d):
    <span class="ck">return</span> n // d, n % d     <span class="cc"># returns a tuple (quotient, remainder)</span>

q, r = <span class="cm">divmod_demo</span>(<span class="cn">17</span>, <span class="cn">5</span>)    <span class="cc"># q=3, r=2</span></pre></div>
  <div class="output-block">swapped: 20 10 &nbsp;&nbsp;&nbsp; multi-return: q=3, r=2</div>
  <div class="step-box">
    Every "multiple values" pattern in Python is a tuple underneath. The left-side unpacking is what makes it read naturally.
  </div>
</div>

<div class="ex-block">
  <div class="ex-title"><span class="ex-badge">Example 4</span> namedtuple as a readable record</div>
  <div class="code-block"><pre><span class="ck">from</span> collections <span class="ck">import</span> namedtuple
Employee = <span class="cm">namedtuple</span>(<span class="cs">"Employee"</span>, [<span class="cs">"name"</span>, <span class="cs">"role"</span>, <span class="cs">"salary"</span>])
e = <span class="cm">Employee</span>(<span class="cs">"Siva"</span>, <span class="cs">"ML Engineer"</span>, <span class="cn">2000000</span>)
<span class="cm">print</span>(e.role)              <span class="cc"># ML Engineer  — readable, not e[1]</span>
<span class="cm">print</span>(e.<span class="cm">_asdict</span>())         <span class="cc"># {'name':'Siva','role':'ML Engineer','salary':2000000}</span>
raise_e = e.<span class="cm">_replace</span>(salary=<span class="cn">2500000</span>)   <span class="cc"># new record, original unchanged</span></pre></div>
  <div class="output-block">
ML Engineer<br>
{'name': 'Siva', 'role': 'ML Engineer', 'salary': 2000000}
  </div>
  <div class="step-box">
    <code>e.role</code> is self-documenting where <code>e[1]</code> is a mystery. <code>_replace</code> returns a new record because tuples are immutable — you never mutate, you derive.
  </div>
</div>

<!-- ══ SECTION 3: EDGE CASES ══ -->
<div class="warn-header"><h2>3. Edge Cases</h2></div>
<p class="body-p">Six cases, all verified.</p>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 1</span> The single-element tuple trap</div>
  <div class="code-block"><pre>not_tuple = (<span class="cn">5</span>)      <span class="cc"># NOT a tuple — just int 5 in parentheses</span>
real_tuple = (<span class="cn">5</span>,)    <span class="cc"># the comma makes it a tuple</span>
<span class="cm">print</span>(<span class="cm">type</span>(not_tuple).__name__)   <span class="cc"># int</span>
<span class="cm">print</span>(<span class="cm">type</span>(real_tuple).__name__)  <span class="cc"># tuple</span></pre></div>
  <div class="output-block">(5) is: int &nbsp;&nbsp;|&nbsp;&nbsp; (5,) is: tuple</div>
  <div class="why-box">
    <strong>Why it matters:</strong> parentheses don't make a tuple — <strong>the comma does</strong>. <code>(5)</code> is just 5 with redundant parens. You need the trailing comma: <code>(5,)</code>. This bites constantly when building a one-element tuple, and silently — <code>x = (5)</code> gives you an int, and the bug surfaces later when you try to iterate or concatenate.
  </div>
</div>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 2</span> Shallow immutability: inner mutables can change</div>
  <div class="code-block"><pre>t = (<span class="cn">1</span>, [<span class="cn">2</span>, <span class="cn">3</span>], <span class="cs">"x"</span>)
t[<span class="cn">1</span>].<span class="cm">append</span>(<span class="cn">4</span>)
<span class="cm">print</span>(t)             <span class="cc"># (1, [2, 3, 4], 'x')</span></pre></div>
  <div class="output-block">(1, [2, 3, 4], 'x')</div>
  <div class="why-box">
    <strong>Why it matters:</strong> the tuple is "immutable" but the list inside it isn't. You froze the structure (which objects the slots point at), not the contents of those objects. This is the Session 1 Edge 4 case and the reason such a tuple can't be hashed.
  </div>
</div>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 3</span> A tuple with a mutable inside is unhashable</div>
  <div class="code-block"><pre><span class="cm">hash</span>((<span class="cn">1</span>, <span class="cn">2</span>))        <span class="cc"># fine — some integer hash</span>
<span class="cm">hash</span>((<span class="cn">1</span>, [<span class="cn">2</span>]))      <span class="cc"># TypeError: unhashable type: 'list'</span></pre></div>
  <div class="output-block">tuple-with-list unhashable: unhashable type: 'list'</div>
  <div class="why-box">
    <strong>Why it matters:</strong> hashability requires everything inside to be hashable. One list element poisons the whole tuple — you can't use it as a dict key or set member. <strong>The rule: a tuple is hashable only if all its contents are.</strong>
  </div>
</div>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 4</span> + and * create new tuples (no in-place anything)</div>
  <div class="code-block"><pre>e = (<span class="cn">1</span>, <span class="cn">2</span>)
<span class="cm">print</span>(<span class="cm">id</span>(e))        <span class="cc"># address A</span>
e += (<span class="cn">3</span>,)
<span class="cm">print</span>(<span class="cm">id</span>(e))        <span class="cc"># address B — DIFFERENT, it's a new object</span>
<span class="cm">print</span>((<span class="cn">0</span>,) * <span class="cn">3</span>)     <span class="cc"># (0, 0, 0)</span></pre></div>
  <div class="output-block">id before ≠ id after += &nbsp;&nbsp;&nbsp; (0,)*3: (0, 0, 0)</div>
  <div class="why-box">
    <strong>Why it matters:</strong> there's no <code>+=</code> mutate-in-place for tuples (unlike lists, 2A Edge 7). <code>e += (3,)</code> builds a brand-new tuple and rebinds <code>e</code>. This is Session 1's rebind-only behavior for immutables — <code>+=</code> on a tuple is pure rebinding, always a new object. (Repeated <code>+=</code> in a loop is therefore O(n²), same trap as list <code>+</code>.)
  </div>
</div>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 5</span> Tuple unpacking must match length exactly</div>
  <div class="code-block"><pre>a, b, c = (<span class="cn">1</span>, <span class="cn">2</span>, <span class="cn">3</span>)      <span class="cc"># fine</span>
a, b = (<span class="cn">1</span>, <span class="cn">2</span>, <span class="cn">3</span>)         <span class="cc"># ValueError: too many values to unpack</span>
a, b, c = (<span class="cn">1</span>, <span class="cn">2</span>)         <span class="cc"># ValueError: not enough values to unpack</span></pre></div>
  <div class="why-box">
    <strong>Why it matters:</strong> unlike star-unpacking (<code>a, *b</code>), plain unpacking demands the counts match exactly, or you get a <code>ValueError</code>. Use <code>*</code> to absorb a variable number: <code>a, *b = (1,2,3,4)</code> gives <code>a=1</code>, <code>b=[2,3,4]</code>. This is a common bug when a function's return shape changes and old unpacking code breaks loudly.
  </div>
</div>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 6</span> Tuples are faster to create and iterate than lists</div>
  <div class="code-block"><pre><span class="cc"># Constructing a tuple literal is faster than a list literal,</span>
<span class="cc"># and tuples have a smaller memory footprint (64 vs 88 bytes for 3 elements).</span></pre></div>
  <div class="why-box">
    <strong>Why it matters:</strong> because a tuple is fixed-size with no growth machinery, Python can construct it slightly faster and store it more compactly. For large collections of fixed records, or as constant lookup data, tuples are the efficient default. (Not a reason to micro-optimize everything into tuples — use them where immutability is meaningful, and the speed is a free bonus.)
  </div>
</div>

<div class="part-header"><h2>4. Golden Rules</h2></div>

<h3 class="sub">The decision framework: tuple or list? (full depth — this is the real skill)</h3>

<p class="body-p">This is the question that actually matters, so let me give it proper treatment rather than a one-liner. The choice isn't about preference — it's about <strong style="color:#cba6f7">what the data is</strong>.</p>

<div class="decision-box">
  <div class="d-row">
    <span class="d-tag-tuple">tuple</span>
    Use when the collection has a <strong>fixed structure where position has meaning</strong>. A coordinate <code style="background:#313244;padding:1px 5px;border-radius:4px;color:#a6e3a1;font-family:'Courier New',monospace">(x, y)</code>, an RGB color <code style="background:#313244;padding:1px 5px;border-radius:4px;color:#a6e3a1;font-family:'Courier New',monospace">(255, 0, 0)</code>, a database row <code style="background:#313244;padding:1px 5px;border-radius:4px;color:#a6e3a1;font-family:'Courier New',monospace">(id, name, email)</code>, a function returning <code style="background:#313244;padding:1px 5px;border-radius:4px;color:#a6e3a1;font-family:'Courier New',monospace">(min, max)</code>. The number of elements is fixed and each position means something specific. You'd never "append a fourth thing" to a coordinate — that would make it a different kind of thing entirely. This is sometimes called a <strong>record</strong> or <strong>struct</strong> use.
  </div>
  <div class="d-row">
    <span class="d-tag-list">list</span>
    Use when the collection is a <strong>variable-length sequence of like items</strong>. A list of user IDs, a list of feature rows, a list of log lines. Elements are homogeneous in role, the count grows and shrinks, and position is incidental (item 3 isn't fundamentally different from item 4). This is a <strong>collection</strong> use.
  </div>
</div>

<div class="test-box">
  <strong style="color:#89b4fa">A useful test:</strong><br>
  "Would it make sense to loop over this and process each element the same way?"<br>
  → If <strong>yes</strong> → <span class="ic" style="color:#89b4fa">list</span> (homogeneous collection).<br><br>
  "Do the positions mean different things?"<br>
  → If <strong>yes</strong> → <span class="ic" style="color:#cba6f7">tuple</span> (heterogeneous record).<br><br>
  <span style="color:#6c7086"><code style="background:#313244;padding:1px 5px;border-radius:4px;color:#a6e3a1;font-family:'Courier New',monospace">(name, age, city)</code> fails the loop test — you wouldn't process name, age, and city identically — so it's a tuple. <code style="background:#313244;padding:1px 5px;border-radius:4px;color:#a6e3a1;font-family:'Courier New',monospace">[name1, name2, name3]</code> passes it — so it's a list.</span>
</div>

<h3 class="sub">The three concrete reasons to actively prefer a tuple when the data fits:</h3>

<div class="reason-block">
  <div class="reason-title"><span class="reason-badge">Reason 1</span> You need it as a dict key or set member.</div>
  <div class="reason-body">
    This is <strong>non-negotiable, not a preference</strong>. Lists are unhashable; if you need <code>{(row, col): value}</code> or a set of coordinate pairs, you must use tuples. This alone decides many cases.
  </div>
</div>

<div class="reason-block">
  <div class="reason-title"><span class="reason-badge">Reason 2</span> You want to guarantee immutability.</div>
  <div class="reason-body">
    Passing a tuple to a function means it cannot be mutated — the silent-corruption trap (2A Trap 7) is structurally impossible. Returning a tuple of config values means callers can't accidentally change your internal state.
  </div>
</div>

<div class="reason-block">
  <div class="reason-title"><span class="reason-badge">Reason 3</span> Memory and speed at scale.</div>
  <div class="reason-body">
    A tuple is ~27% smaller (64 vs 88 bytes for 3 elements, verified) and faster to construct. For millions of fixed records, that's real.
  </div>
</div>

<h3 class="sub">Other rules, kept tight:</h3>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 4</span> Always use the trailing comma for a single-element tuple.</div>
  <div class="rule-body"><code>(5,)</code>, never <code>(5)</code>. The comma makes the tuple; the parens don't.</div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 5</span> Reach for namedtuple when positions get hard to remember.</div>
  <div class="rule-body">
    If you find yourself writing comments like <code># [0] is name, [1] is age</code>, that's the signal to upgrade to a namedtuple for self-documenting <code>.name</code> / <code>.age</code> access.
  </div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 6</span> To "change" a tuple, rebuild it.</div>
  <div class="rule-body">
    Since you can't mutate, slice-and-concatenate a new one: <code>t = t[:i] + (new,) + t[i+1:]</code>. If you're doing this often, you probably wanted a list.
  </div>
</div>

<!-- ══ SECTION 5: COMMON TRAPS ══ -->
<div class="trap-header"><h2>5. Common Traps</h2></div>

<div class="trap-block">
  <div class="trap-title">
    <span class="trap-badge">Trap 1</span> The missing comma (silent type bug)
    <span class="depth-tag">full depth — it's the #1 tuple bug</span>
  </div>
  <div class="code-block"><pre><span class="ck">def</span> <span class="cm">get_tags</span>():
    <span class="ck">return</span> (<span class="cs">"python"</span>)     <span class="cc"># BUG: returns the STRING "python", not a tuple</span></pre></div>
  <p class="body-p" style="margin:6px 0">You intended a one-element tuple; you got a string, because parentheses alone don't make a tuple. The bug is silent — no error at the return. It surfaces later and confusingly:</p>
  <div class="code-block"><pre><span class="ck">for</span> tag <span class="ck">in</span> <span class="cm">get_tags</span>():    <span class="cc"># iterates the STRING</span>
    <span class="cm">print</span>(tag)            <span class="cc"># prints 'p','y','t','h','o','n' — one char each!</span></pre></div>
  <div class="verified-box">✅ Verified — the buggy version yields six characters.</div>
  <div class="tell-box">
    <strong>Tell:</strong> you're iterating what you think is a single-item collection and getting individual characters (string) or a <code>TypeError</code> (int/float).
  </div>
  <div class="fix-box">Fix: <code>return ("python",)</code></div>
</div>

<div class="trap-block">
  <div class="trap-title">
    <span class="trap-badge">Trap 2</span> The accidental tuple from a stray comma.
    <span class="depth-tag">tight</span>
  </div>
  <div class="code-block"><pre>x = <span class="cn">5</span>,        <span class="cc"># x is (5,), a tuple — not the int 5</span></pre></div>
  <div class="verified-box">✅ Verified: <code>x = 5,</code> gives <code>(5,)</code>.</div>
  <p class="body-p" style="margin:6px 0">A trailing comma you didn't mean to type silently creates a one-element tuple. The inverse of Trap 1 — here the comma is the problem.</p>
  <div class="tell-box"><strong>Tell:</strong> a variable you expected to be a number/string is mysteriously a tuple.</div>
</div>

<div class="trap-block">
  <div class="trap-title">
    <span class="trap-badge">Trap 3</span> Unhashable tuple as a key.
    <span class="depth-tag">tight</span>
  </div>
  <div class="code-block"><pre>cache[(<span class="cn">1</span>, [<span class="cn">2</span>])] = <span class="cs">"x"</span>     <span class="cc"># TypeError: unhashable type: 'list'</span></pre></div>
  <div class="verified-box">✅ Verified.</div>
  <p class="body-p" style="margin:6px 0">A tuple containing a mutable can't be a dict key or set member. The tuple being "immutable" isn't enough — its contents must be hashable too.</p>
  <div class="fix-box">Fix: ensure all elements are immutable (convert inner lists to tuples).</div>
</div>

<div class="trap-block">
  <div class="trap-title">
    <span class="trap-badge">Trap 4</span> Assuming "immutable" means "deeply frozen."
    <span class="depth-tag">tight</span>
  </div>
  <div class="code-block"><pre>config = (<span class="cn">1</span>, [<span class="cn">2</span>, <span class="cn">3</span>])
config[<span class="cn">1</span>].<span class="cm">append</span>(<span class="cn">4</span>)       <span class="cc"># works! inner list mutated → (1, [2,3,4])</span></pre></div>
  <p class="body-p" style="margin:6px 0">The structure is frozen, the inner objects aren't. If you handed this tuple out expecting it to be tamper-proof, the inner list is still a back door.</p>
  <div class="fix-box">Fix: use only immutable elements if you need true immutability.</div>
</div>

<div class="ml-header"><h2>7. ML Real-World Connection</h2></div>

<p class="body-p">Tuples are quieter than lists in ML code, but they're <strong style="color:#a6e3a1">load-bearing</strong> in ways you use constantly without noticing.</p>

<div class="ml-block">
  <div class="ml-title"><span class="ml-badge">ML Concept 1</span> Array shapes are tuples — and you unpack them all the time</div>
  <p class="body-p" style="margin:0 0 8px 0">Every NumPy array, every PyTorch/TensorFlow tensor reports its dimensions as a tuple:</p>
  <div class="code-block"><pre>arr = np.<span class="cm">zeros</span>((<span class="cn">3</span>, <span class="cn">4</span>, <span class="cn">5</span>))
arr.shape          <span class="cc"># (3, 4, 5)  — a tuple</span>
batch, height, width = arr.shape    <span class="cc"># unpacking dims</span></pre></div>
  <div class="verified-box">✅ Verified. <code>shape</code> is a tuple precisely because dimensions are a fixed-structure record, not a growable list — you don't "append a dimension" to an existing array. Reshaping (<code>arr.reshape((6, 10))</code>) takes a tuple too. When you write <code>model.input_shape</code>, debug a dimension-mismatch error, or unpack <code>n, c, h, w = tensor.shape</code>, you're working with tuples.</div>
</div>

<div class="ml-block">
  <div class="ml-title"><span class="ml-badge">ML Concept 2</span> dict.items() yields (key, value) tuples — every config loop uses this</div>
  <div class="code-block"><pre><span class="ck">for</span> name, value <span class="ck">in</span> hyperparams.<span class="cm">items</span>():    <span class="cc"># each item is a tuple</span>
    <span class="cm">print</span>(<span class="cs">f"</span>{name}<span class="cs">: </span>{value}<span class="cs">"</span>)</pre></div>
  <div class="verified-box">✅ Verified. You unpack the tuple in the loop header without thinking about it. Same with <code>enumerate</code> yielding <code>(index, value)</code> — tuples power the most common iteration idioms in ML preprocessing.</div>
</div>

<div class="ml-block">
  <div class="ml-title"><span class="ml-badge">ML Concept 3</span> Tuple keys for compound lookups — the recommender-system staple</div>
  <p class="body-p" style="margin:0 0 8px 0">When you need to look up by two or more things at once:</p>
  <div class="code-block"><pre>ratings = {}
ratings[(<span class="cs">"user1"</span>, <span class="cs">"movieA"</span>)] = <span class="cn">4.5</span>      <span class="cc"># (user, item) pair as key</span>
ratings[(<span class="cs">"user2"</span>, <span class="cs">"movieA"</span>)] = <span class="cn">5.0</span>
ratings[(<span class="cs">"user1"</span>, <span class="cs">"movieA"</span>)]            <span class="cc"># 4.5</span></pre></div>
  <div class="verified-box">✅ Verified. A <code>(user_id, item_id)</code> tuple key is the natural representation for a sparse ratings matrix, co-occurrence counts, or any "value indexed by a combination." This is the hashability payoff doing real work — directly relevant to recommendation and the sparse-matrix patterns you'll hit. It's also exactly your OpsRAG dedup territory: a <code>(question_id, answer_id)</code> or content-hash tuple makes a clean dedup key.</div>
</div>

<div class="ml-block">
  <div class="ml-title"><span class="ml-badge">ML Concept 4</span> Immutable hyperparameter configs</div>
  <p class="body-p" style="margin:0 0 8px 0">Storing a fixed config as a tuple (or namedtuple) guarantees no part of your pipeline accidentally mutates it mid-run — structurally preventing the "my learning rate changed between experiments" reproducibility bug from 2A's ML section:</p>
  <div class="code-block"><pre>HYPERPARAMS = (<span class="cs">"adam"</span>, <span class="cn">0.001</span>, <span class="cn">32</span>)       <span class="cc"># frozen</span>
optimizer, lr, batch_size = HYPERPARAMS</pre></div>
  <div class="verified-box">✅ Verified. A namedtuple version (<code>Config(optimizer="adam", lr=0.001, ...)</code>) is even better — immutable and self-documenting.</div>
</div>

<!-- ══ SECTION 8: INTERVIEW QUESTIONS ══ -->
<div class="interview-header"><h2>8. Interview Questions</h2></div>

<div class="sub-header"><h3>8a — Conceptual Q&amp;A (10 questions)</h3></div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">1</span><span class="qa-badge badge-basic">core</span> "Tuple vs list — the core difference?"</div>
  <div class="qa-a">
    <span class="ans-label">A</span>
    Tuples are <strong>immutable</strong> (fixed structure, can't add/remove/reassign slots), lists are <strong>mutable</strong>. From that one difference flows hashability, memory efficiency, and the record-vs-collection distinction.
  </div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">2</span><span class="qa-badge badge-basic">hashability</span> "Why can a tuple be a dict key but a list can't?"</div>
  <div class="qa-a">
    <span class="ans-label">A</span>
    Dict keys must be <strong>hashable</strong>, requiring a stable hash over the object's lifetime. A tuple's immutability guarantees a stable hash; a list can change, so it's unhashable.
  </div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">3</span><span class="qa-badge badge-trap">trap</span> "Is a tuple always hashable?"</div>
  <div class="qa-a">
    <span class="ans-label">A</span>
    No — only if <strong>all its contents are hashable</strong>. A tuple containing a list is unhashable, because the inner list could change.
  </div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">4</span><span class="qa-badge badge-medium">decision</span> "When would you choose a tuple over a list?"</div>
  <div class="qa-a">
    <span class="ans-label">A</span>
    When the data is a <strong>fixed-structure record</strong> (coordinate, RGB, row), when you need a <strong>dict key/set member</strong>, when you want <strong>guaranteed immutability</strong>, or for <strong>memory/speed at scale</strong> with fixed records.
  </div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">5</span><span class="qa-badge badge-trap">classic trap</span> "What does (5) evaluate to? How do you make a one-element tuple?"</div>
  <div class="qa-a">
    <span class="ans-label">A</span>
    <code>(5)</code> is just the int <code>5</code> — parentheses don't make a tuple. <strong>The comma does:</strong> <code>(5,)</code>.
  </div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">6</span><span class="qa-badge badge-medium">shallow</span> "Are tuples completely immutable?"</div>
  <div class="qa-a">
    <span class="ans-label">A</span>
    The <strong>structure</strong> is immutable (which objects the slots point at can't change), but if a slot holds a mutable object like a list, that inner object can still be mutated. <strong>Immutability is shallow.</strong>
  </div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">7</span><span class="qa-badge badge-basic">mutation</span> "How do you 'modify' a tuple?"</div>
  <div class="qa-a">
    <span class="ans-label">A</span>
    You can't — you <strong>build a new one</strong> via slicing and concatenation: <code>t = t[:i] + (new,) + t[i+1:]</code>. If you do this often, you probably wanted a list.
  </div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">8</span><span class="qa-badge badge-medium">namedtuple</span> "What's a namedtuple and when use it?"</div>
  <div class="qa-a">
    <span class="ans-label">A</span>
    An immutable tuple with named fields (<code>p.x</code> instead of <code>p[0]</code>). Use it for readable fixed records — lighter than a class, self-documenting, still hashable and unpackable. <code>_replace</code> derives a modified copy.
  </div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">9</span><span class="qa-badge badge-basic">returns</span> "How does Python return multiple values?"</div>
  <div class="qa-a">
    <span class="ans-label">A</span>
    It returns a <strong>single tuple</strong>. <code>return a, b</code> packs <code>(a, b)</code>; the caller unpacks with <code>x, y = f()</code>. There's no actual "multiple return" — it's one tuple.
  </div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">10</span><span class="qa-badge badge-medium">swap</span> "Why is x, y = y, x a valid swap?"</div>
  <div class="qa-a">
    <span class="ans-label">A</span>
    The right side is <strong>fully evaluated</strong> into a tuple <code>(y, x)</code> before any assignment, then unpacked into <code>x, y</code>. No temp variable needed.
  </div>
</div>

<div class="ml-header"><h2>7. ML Real-World Connection</h2></div>

<p class="body-p">Tuples are quieter than lists in ML code, but they're <strong style="color:#a6e3a1">load-bearing</strong> in ways you use constantly without noticing.</p>

<div class="ml-block">
  <div class="ml-title"><span class="ml-badge">ML Concept 1</span> Array shapes are tuples — and you unpack them all the time</div>
  <p class="body-p" style="margin:0 0 8px 0">Every NumPy array, every PyTorch/TensorFlow tensor reports its dimensions as a tuple:</p>
  <div class="code-block"><pre>arr = np.<span class="cm">zeros</span>((<span class="cn">3</span>, <span class="cn">4</span>, <span class="cn">5</span>))
arr.shape          <span class="cc"># (3, 4, 5)  — a tuple</span>
batch, height, width = arr.shape    <span class="cc"># unpacking dims</span></pre></div>
  <div class="verified-box">✅ Verified. <code>shape</code> is a tuple precisely because dimensions are a fixed-structure record, not a growable list — you don't "append a dimension" to an existing array. Reshaping (<code>arr.reshape((6, 10))</code>) takes a tuple too. When you write <code>model.input_shape</code>, debug a dimension-mismatch error, or unpack <code>n, c, h, w = tensor.shape</code>, you're working with tuples.</div>
</div>

<div class="ml-block">
  <div class="ml-title"><span class="ml-badge">ML Concept 2</span> dict.items() yields (key, value) tuples — every config loop uses this</div>
  <div class="code-block"><pre><span class="ck">for</span> name, value <span class="ck">in</span> hyperparams.<span class="cm">items</span>():    <span class="cc"># each item is a tuple</span>
    <span class="cm">print</span>(<span class="cs">f"</span>{name}<span class="cs">: </span>{value}<span class="cs">"</span>)</pre></div>
  <div class="verified-box">✅ Verified. You unpack the tuple in the loop header without thinking about it. Same with <code>enumerate</code> yielding <code>(index, value)</code> — tuples power the most common iteration idioms in ML preprocessing.</div>
</div>

<div class="ml-block">
  <div class="ml-title"><span class="ml-badge">ML Concept 3</span> Tuple keys for compound lookups — the recommender-system staple</div>
  <p class="body-p" style="margin:0 0 8px 0">When you need to look up by two or more things at once:</p>
  <div class="code-block"><pre>ratings = {}
ratings[(<span class="cs">"user1"</span>, <span class="cs">"movieA"</span>)] = <span class="cn">4.5</span>      <span class="cc"># (user, item) pair as key</span>
ratings[(<span class="cs">"user2"</span>, <span class="cs">"movieA"</span>)] = <span class="cn">5.0</span>
ratings[(<span class="cs">"user1"</span>, <span class="cs">"movieA"</span>)]            <span class="cc"># 4.5</span></pre></div>
  <div class="verified-box">✅ Verified. A <code>(user_id, item_id)</code> tuple key is the natural representation for a sparse ratings matrix, co-occurrence counts, or any "value indexed by a combination." This is the hashability payoff doing real work — directly relevant to recommendation and the sparse-matrix patterns you'll hit. It's also exactly your OpsRAG dedup territory: a <code>(question_id, answer_id)</code> or content-hash tuple makes a clean dedup key.</div>
</div>

<div class="ml-block">
  <div class="ml-title"><span class="ml-badge">ML Concept 4</span> Immutable hyperparameter configs</div>
  <p class="body-p" style="margin:0 0 8px 0">Storing a fixed config as a tuple (or namedtuple) guarantees no part of your pipeline accidentally mutates it mid-run — structurally preventing the "my learning rate changed between experiments" reproducibility bug from 2A's ML section:</p>
  <div class="code-block"><pre>HYPERPARAMS = (<span class="cs">"adam"</span>, <span class="cn">0.001</span>, <span class="cn">32</span>)       <span class="cc"># frozen</span>
optimizer, lr, batch_size = HYPERPARAMS</pre></div>
  <div class="verified-box">✅ Verified. A namedtuple version (<code>Config(optimizer="adam", lr=0.001, ...)</code>) is even better — immutable and self-documenting.</div>
</div>

<div class="interview-header"><h2>8. Interview Questions</h2></div>

<div class="sub-header"><h3>8a — Conceptual Q&amp;A (10 questions)</h3></div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">1</span><span class="qa-badge badge-basic">core</span> "Tuple vs list — the core difference?"</div>
  <div class="qa-a"><span class="ans-label">A</span> Tuples are <strong>immutable</strong> (fixed structure, can't add/remove/reassign slots), lists are <strong>mutable</strong>. From that one difference flows hashability, memory efficiency, and the record-vs-collection distinction.</div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">2</span><span class="qa-badge badge-basic">hashability</span> "Why can a tuple be a dict key but a list can't?"</div>
  <div class="qa-a"><span class="ans-label">A</span> Dict keys must be <strong>hashable</strong>, requiring a stable hash over the object's lifetime. A tuple's immutability guarantees a stable hash; a list can change, so it's unhashable.</div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">3</span><span class="qa-badge badge-trap">trap</span> "Is a tuple always hashable?"</div>
  <div class="qa-a"><span class="ans-label">A</span> No — only if <strong>all its contents are hashable</strong>. A tuple containing a list is unhashable, because the inner list could change.</div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">4</span><span class="qa-badge badge-medium">decision</span> "When would you choose a tuple over a list?"</div>
  <div class="qa-a"><span class="ans-label">A</span> When the data is a <strong>fixed-structure record</strong> (coordinate, RGB, row), when you need a <strong>dict key/set member</strong>, when you want <strong>guaranteed immutability</strong>, or for <strong>memory/speed at scale</strong> with fixed records.</div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">5</span><span class="qa-badge badge-trap">classic trap</span> "What does (5) evaluate to? How do you make a one-element tuple?"</div>
  <div class="qa-a"><span class="ans-label">A</span> <code>(5)</code> is just the int <code>5</code> — parentheses don't make a tuple. <strong>The comma does:</strong> <code>(5,)</code>.</div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">6</span><span class="qa-badge badge-medium">shallow</span> "Are tuples completely immutable?"</div>
  <div class="qa-a"><span class="ans-label">A</span> The <strong>structure</strong> is immutable (which objects the slots point at can't change), but if a slot holds a mutable object like a list, that inner object can still be mutated. <strong>Immutability is shallow.</strong></div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">7</span><span class="qa-badge badge-basic">mutation</span> "How do you 'modify' a tuple?"</div>
  <div class="qa-a"><span class="ans-label">A</span> You can't — you <strong>build a new one</strong> via slicing and concatenation: <code>t = t[:i] + (new,) + t[i+1:]</code>. If you do this often, you probably wanted a list.</div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">8</span><span class="qa-badge badge-medium">namedtuple</span> "What's a namedtuple and when use it?"</div>
  <div class="qa-a"><span class="ans-label">A</span> An immutable tuple with named fields (<code>p.x</code> instead of <code>p[0]</code>). Use it for readable fixed records — lighter than a class, self-documenting, still hashable and unpackable. <code>_replace</code> derives a modified copy.</div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">9</span><span class="qa-badge badge-basic">returns</span> "How does Python return multiple values?"</div>
  <div class="qa-a"><span class="ans-label">A</span> It returns a <strong>single tuple</strong>. <code>return a, b</code> packs <code>(a, b)</code>; the caller unpacks with <code>x, y = f()</code>. There's no actual "multiple return" — it's one tuple.</div>
</div>

<div class="qa-block">
  <div class="qa-q"><span class="q-num">10</span><span class="qa-badge badge-medium">swap</span> "Why is x, y = y, x a valid swap?"</div>
  <div class="qa-a"><span class="ans-label">A</span> The right side is <strong>fully evaluated</strong> into a tuple <code>(y, x)</code> before any assignment, then unpacked into <code>x, y</code>. No temp variable needed.</div>
</div>

<div class="challenge-block">
  <div class="ch-title"><span class="ch-badge-medium">C4 · Medium</span><span class="ch-medium">Find the first pair of numbers summing to target, return as a tuple.</span></div>
  <div class="ch-prompt">Given a list of numbers and a target, return the first <code>(a, b)</code> pair where <code>a + b == target</code>. Use O(1) lookup.</div>
  <div class="code-block"><pre><span class="ck">def</span> <span class="cm">find_pair</span>(nums, target):

    <span class="ck">for</span> n <span class="ck">in</span> nums:
        <span class="ck">if</span> target - n <span class="ck">in</span> seen:
            <span class="ck">return</span> (target - n, n)
        seen.<span class="cm">add</span>(n)
    <span class="ck">return</span> <span class="ck">None</span>

<span class="cc"># find_pair([2,7,11,15], 9) -> (2, 7)</span></pre></div>
  <div class="why-box">The two-sum pattern (O(1) set lookup) returning a tuple. Set membership + tuple result.</div>
</div>

<div class="challenge-block">
  <div class="ch-title"><span class="ch-badge-hard">C5 · Hard</span><span class="ch-hard">Group points by quadrant, using a tuple as the grouping key.</span></div>
  <div class="ch-prompt">Given a list of <code>(x, y)</code> points, group them into quadrants using a <code>(bool, bool)</code> tuple as the dict key.</div>
  <div class="code-block"><pre><span class="ck">def</span> <span class="cm">quadrant</span>(points):
    result = {}
    <span class="ck">for</span> x, y <span class="ck">in</span> points:                 <span class="cc"># unpack each point tuple</span>
        q = (x >= <span class="cn">0</span>, y >= <span class="cn">0</span>)            <span class="cc"># a (bool, bool) tuple as the key</span>
        result.<span class="cm">setdefault</span>(q, []).<span class="cm">append</span>((x, y))
    <span class="ck">return</span> result

<span class="cc"># -> {(True,True):[(1,1),(2,2)], (False,True):[(-1,1)], ...}</span></pre></div>
  <div class="verified-box">✅ Verified. Two things at once: unpacking each point in the loop, and using a <code>(bool, bool)</code> tuple as a dict key — four quadrants, four keys. <code>setdefault</code> initializes the list on first sight.</div>
</div>

<div class="challenge-block">
  <div class="ch-title"><span class="ch-badge-medium">C6 · Medium</span><span class="ch-medium">Transpose a matrix represented as a list of tuples.</span></div>
  <div class="ch-prompt">Given <code>[(1,2,3), (4,5,6)]</code>, return the transposed matrix as a list of tuples.</div>
  <div class="code-block"><pre>matrix = [(<span class="cn">1</span>,<span class="cn">2</span>,<span class="cn">3</span>), (<span class="cn">4</span>,<span class="cn">5</span>,<span class="cn">6</span>)]
<span class="cm">list</span>(<span class="cm">zip</span>(*matrix))    <span class="cc"># [(1,4), (2,5), (3,6)]</span></pre></div>
  <div class="why-box"><code>zip(*matrix)</code> unpacks the rows as arguments and pairs columns — and <code>zip</code> naturally yields tuples, so the result is a list of tuples (no <code>list()</code> wrapping of each needed if tuples are fine).</div>
</div>

<div class="challenge-block">
  <div class="ch-title"><span class="ch-badge-hard">C7 · Hard</span><span class="ch-hard">Find the most common coordinate pair and its count.</span></div>
  <div class="ch-prompt">Given a list of <code>(x, y)</code> coordinate tuples with duplicates, return the most common coordinate and how many times it appeared.</div>
  <div class="code-block"><pre><span class="ck">from</span> collections <span class="ck">import</span> Counter

<span class="ck">def</span> <span class="cm">most_common_point</span>(coords):
    <span class="ck">return</span> <span class="cm">Counter</span>(coords).<span class="cm">most_common</span>(<span class="cn">1</span>)[<span class="cn">0</span>]

<span class="cc"># most_common_point([(0,0),(1,1),(0,0),(0,0),(1,1)]) -> ((0,0), 3)</span></pre></div>
  <div class="verified-box">✅ Verified. This only works because tuples are hashable — <code>Counter</code> uses them as keys. A list of lists would raise <code>unhashable type</code>. The result <code>((0,0), 3)</code> is itself a tuple of <code>(item, count)</code>.</div>
</div>
