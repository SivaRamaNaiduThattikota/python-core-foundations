```python
# 4 (Medium) — Sparse matrix with tuple keys. Use a dict with (row, col) tuple keys to store only the non-zero entries of a sparse matrix. 
# Store value 5 at (0,0) and value 9 at (2,3), then look up (2,3).

# Hint: this is the canonical "why tuples must be hashable" use. sparse[(0,0)] = 5. Why would a list key fail here?
sparse = {}
sparse[(0, 0)] = 5
sparse[(2, 3)] = 9
print(sparse)
print(sparse.get((2,3)))
```

```python
# 5 (Hard) — Dedup coordinate pairs preserving order. Given coords = [(0,0),(1,1),(0,0),(2,2),(1,1)], remove duplicate pairs while keeping first-seen order → [(0,0),(1,1),(2,2)].

# Hint: this combines 2A's dedup pattern with tuple hashability — tuples can go in a set (unlike lists), so the seen-set approach works directly. Or the dict.fromkeys one-liner. 
# Why does this work for tuples but would fail if the elements were lists?
coords = [(0,0),(1,1),(0,0),(2,2),(1,1)]
# One -liner:
print(list(dict.fromkeys(coords)))

# Explicit seen-set version:
seen = set()
out = []
for c in coords:
    if c not in seen:
        seen.add(c)
        out.append(c)
print(out)
```

<div style="background:#1e1e2e; padding:16px 20px; border-radius:8px; font-family:monospace; color:#cdd6f4; line-height:1.8">
  <p style="margin:0 0 12px 0;">
    Both give <code>[(0,0),(1,1),(2,2)]</code>. The key insight: this works because tuples are <strong style="color:#89b4fa">hashable</strong>, so they can go into a <code>set</code> (or be <code>dict</code> keys) for O(1) membership.
  </p>
  <p style="margin:0 0 12px 0; border-left:4px solid #f38ba8; padding-left:12px">
    If the elements were lists — <code>[[0,0],[1,1],...]</code> — <code>seen.add([0,0])</code> would raise <code>unhashable type: 'list'</code> and the whole pattern collapses.
  </p>
  <p style="margin:0; border-left:4px solid #a6e3a1; padding-left:12px">
    That's the practical difference between <strong style="color:#a6e3a1">tuple and list</strong> in one example.
  </p>
</div>
