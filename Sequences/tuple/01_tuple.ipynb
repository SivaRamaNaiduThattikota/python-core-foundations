```python
from collections import Counter
def freq(seq):
    return sorted(Counter(seq).items(), key=lambda x: -x[1])
seq = ("a","b","a","c","a","b")
print(freq(seq))
```

```python
def count_frequency(arr):
    freq = {}

    # Step 1: Count manually
    for item in arr:
        if item in freq:
            freq[item] += 1
        else:
            freq[item] = 1

    # Step 2: Convert dict to list of tuples
    result = []
    for key in freq:
        result.append((key, freq[key]))

    # Step 3: Manual sorting (selection sort - descending by count)
    n = len(result)
    for i in range(n):
        max_idx = i
        for j in range(i + 1, n):
            if result[j][1] > result[max_idx][1]:
                max_idx = j
        
        # swap
        result[i], result[max_idx] = result[max_idx], result[i]

    return result
```

```python
# arr = [1, 2, 2, 3, 3, 3, 1]
arr = ("a","b","a","c","a","b")
print(count_frequency(arr))
```

Time Complexity

Counting → O(n)

Selection sort → O(n²) (n = unique elements)

Total: O(n²) (due to manual sorting)

#### Merge Sort Code

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)


def merge(left, right):
    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i][1] >= right[j][1]:   # compare count (ing)
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    # remaining elements
    while i < len(left):
        result.append(left[i])
        i += 1

    while j < len(right):
        result.append(right[j])
        j += 1

    return result
```

### Top K Frequent Elements
1. Frequency counting
2. Merging sort
3. Extract top K

```python
def top_k_frequent(arr, k):
    # Step 1: Count frequency manually
    freq = {}
    for x in arr:
        if x in freq:
            freq[x] += 1
        else:
            freq[x] = 1

    # Step 2: Convert to (item, count)
    freq_list = []
    for key in freq:
        freq_list.append((key, freq[key]))

    # Step 3: Sort using merge sort
    sorted_list = merge_sort(freq_list)

    # Step 4: Extract top k items
    result = []
    for i in range(min(k, len(sorted_list))):
        result.append(sorted_list[i][0])  # only item

    return result
```

```python
# arr = [1, 2, 2, 3, 3, 3, 1]
arr = ("a","b","a","c","a","b")
print(top_k_frequent(arr, 2))
```

<div class="part-header"><h2>📦 Use a Tuple — Fixed Structure, Position Has Meaning</h2></div>

<div class="theory-box">
  Use a tuple when the collection has a <strong>fixed structure where position has meaning</strong>. The number of elements is fixed and each position means something specific. You'd never "append a fourth thing" to a coordinate — that would make it a different kind of thing entirely. This is sometimes called a <strong>record</strong> or <strong>struct</strong> use.
</div>

<h3 style="color:#89dceb;font-family:'Segoe UI',sans-serif;margin-top:20px;margin-bottom:10px;font-size:1em">Real examples</h3>

<div class="record-box">
  <div class="rec-row">
    <span class="rec-type">(x, y)</span>
    <span class="rec-meaning">A coordinate — slot 0 is always x, slot 1 is always y</span>
    <span class="rec-tag">coordinate</span>
  </div>
  <div class="rec-row">
    <span class="rec-type">(255, 0, 0)</span>
    <span class="rec-meaning">An RGB color — slot 0 is red, slot 1 is green, slot 2 is blue</span>
    <span class="rec-tag">color</span>
  </div>
  <div class="rec-row">
    <span class="rec-type">(id, name, email)</span>
    <span class="rec-meaning">A database row — each position is a specific field</span>
    <span class="rec-tag">db row</span>
  </div>

<div class="why-box">
  <strong>Why not a list?</strong> A list signals "this is a variable-length collection of like things." A coordinate isn't a collection — it's a fixed record where each position has a distinct semantic role. Using a tuple <em>communicates that intent</em> to anyone reading the code, and structurally prevents accidental mutation or growth.
</div>

<div class="note-box">
  💡 <strong>The test:</strong> "Would it make sense to append more items to this?" If the answer is no (you'd never add a third number to an <code>(x, y)</code> pair), it's a tuple. If yes (a list of user IDs that grows), it's a list.
</div>
