<div class="part-header"><h2>1. Theory</h2></div>

<h3 class="sub">1.1 What a set is — a hash table of unique, hashable elements</h3>

<div class="theory-box">
  A set is an <strong>unordered collection of unique, hashable elements</strong>, built on a hash table. Three defining properties, each load-bearing:
</div>

<div class="prop-box">
  <div class="prop-row">
    <span class="prop-num">1</span>
    <span class="prop-title">Unique</span>
    &nbsp;— duplicates collapse automatically. <span class="ic">{3, 1, 2, 1, 3}</span> becomes <span class="ic">{1, 2, 3}</span>. This isn't a feature you invoke; it's structural — a hash table can't hold the same key twice.
  </div>
  <div class="prop-row">
    <span class="prop-num">2</span>
    <span class="prop-title">Hashable elements only</span>
    &nbsp;— every element must be hashable (immutable-enough to have a stable hash), for exactly the reason tuples can be dict keys (2B). Numbers, strings, tuples-of-immutables: fine. Lists, dicts, sets: not — they're mutable. <span class="ic">{[1,2]}</span> raises <span class="ic">unhashable type: 'list'</span>.
  </div>
  <div class="prop-row">
    <span class="prop-num">3</span>
    <span class="prop-title">Unordered, no indexing</span>
    &nbsp;— there's no <span class="ic">s[0]</span>. Elements are stored by where their hash lands in the table, not by position. <span class="ic">s[0]</span> raises <span class="ic">'set' object is not subscriptable</span>. (CPython sets have no insertion-order guarantee, unlike dicts since 3.7.)
  </div>
</div>

<h3 class="sub">1.2 How a hash table gives O(1) membership — the core mechanism</h3>

<p class="body-p">This is the heart of the session. When you ask <span class="ic">x in my_set</span>, Python does <strong>not</strong> scan elements one by one (that's what a list does, O(n)). Instead:</p>

<div class="step-box">
  <div class="step-row">
    <span class="step-num">1</span>
    <span class="step-title">Hash</span>
    &nbsp;— It computes <span class="ic">hash(x)</span> — a number derived from x's value.
  </div>
  <div class="step-row">
    <span class="step-num">2</span>
    <span class="step-title">Direct Address</span>
    &nbsp;— That hash points directly to a bucket (a slot in the table) where x would live if it's present.
  </div>
  <div class="step-row">
    <span class="step-num">3</span>
    <span class="step-title">Instant Check</span>
    &nbsp;— Python jumps straight to that bucket and checks. One jump, not a scan.
  </div>
</div>

<p class="body-p">Because the location is <strong>computed from the value</strong> rather than searched for, membership is <strong style="color:#a6e3a1">O(1) on average</strong> — constant time regardless of set size. This is the single most important performance fact about sets, and the live proof is dramatic:</p>

<div class="code-block"><pre>big_list = <span class="cm">list</span>(<span class="cm">range</span>(<span class="cn">1_000_000</span>))
big_set  = <span class="cm">set</span>(big_list)
(<span class="cn">999999</span>) <span class="ck">in</span> big_list   <span class="cc"># 7.49 ms   — scans ~1M elements</span>
(<span class="cn">999999</span>) <span class="ck">in</span> big_set    <span class="cc"># 2.65 µs   — one hash lookup</span></pre></div>

<div class="perf-box">
  <div class="perf-header">⚡ ~2,800× speedup on this machine — same lookup, opposite strategies</div>
  <div class="perf-row"><span class="perf-slow">list: 7.49 ms</span> &nbsp;— scans up to 1,000,000 elements</div>
  <div class="perf-row"><span class="perf-fast">set:  2.65 µs</span> &nbsp;— one hash computation, one bucket check</div>
</div>

<div class="info-box">
  💡 <strong>SQL/Power BI anchor:</strong> a set is like a column with a hash index. <span class="ic">x in set</span> is an indexed lookup (instant); <span class="ic">x in list</span> is a full table scan (reads every row). Same idea you rely on for query performance.<br><br>
  (Worst case is O(n) if many elements hash-collide into the same bucket, but Python's hashing makes that rare in practice — treat membership as O(1).)
</div>

<h3 class="sub">1.3 Set algebra — the operations lists can't do cleanly</h3>

<p class="body-p">Sets give you mathematical set operations as first-class, fast operations. This is the other reason to reach for a set (beyond dedup and membership):</p>

<div class="code-block"><pre>a = {<span class="cn">1</span>, <span class="cn">2</span>, <span class="cn">3</span>, <span class="cn">4</span>}
b = {<span class="cn">3</span>, <span class="cn">4</span>, <span class="cn">5</span>, <span class="cn">6</span>}
a | b      <span class="cc"># union        -> {1,2,3,4,5,6}   (in either)</span>
a &amp; b      <span class="cc"># intersection -> {3,4}           (in both)</span>
a - b      <span class="cc"># difference   -> {1,2}           (in a, not b)</span>
a ^ b      <span class="cc"># symmetric    -> {1,2,5,6}       (in exactly one)</span></pre></div>

<p class="body-p">All outputs confirmed. And relational tests:</p>

<div class="code-block"><pre>{<span class="cn">1</span>,<span class="cn">2</span>} &lt;= a          <span class="cc"># subset?    True</span>
a &gt;= {<span class="cn">1</span>,<span class="cn">2</span>}          <span class="cc"># superset?  True</span>
a.<span class="cm">isdisjoint</span>({<span class="cn">9</span>})   <span class="cc"># no overlap? True</span></pre></div>

<div class="algebra-box">
  <div class="alg-row"><span class="alg-op">a | b</span> <span class="alg-name">union</span> <span class="alg-result">{1,2,3,4,5,6}</span> &nbsp;<span class="alg-desc">— in either</span></div>
  <div class="alg-row"><span class="alg-op">a &amp; b</span> <span class="alg-name">intersection</span> <span class="alg-result">{3,4}</span> &nbsp;<span class="alg-desc">— in both</span></div>
  <div class="alg-row"><span class="alg-op">a - b</span> <span class="alg-name">difference</span> <span class="alg-result">{1,2}</span> &nbsp;<span class="alg-desc">— in a, not b</span></div>
  <div class="alg-row"><span class="alg-op">a ^ b</span> <span class="alg-name">symmetric diff</span> <span class="alg-result">{1,2,5,6}</span> &nbsp;<span class="alg-desc">— in exactly one</span></div>
</div>

<div class="info-box">
  💡 <strong>Why this matters:</strong> doing these with lists requires manual loops and membership checks (O(n·m)); with sets they're concise and fast. "Which users are in both segment A and segment B?" → <span class="ic">a &amp; b</span>. "Which items are new this week?" → <span class="ic">this_week - last_week</span>.<br><br>
  <strong>SQL anchor:</strong> union/intersect/difference are literally SQL set operators (<span class="ic">UNION</span>, <span class="ic">INTERSECT</span>, <span class="ic">EXCEPT</span>) — same operations, same names.
</div>

<h3 class="sub">1.4 Operators vs methods — a useful distinction</h3>

<p class="body-p">Each set operation has both an operator (<span class="ic">|</span>) and a method (<span class="ic">.union()</span>). The difference is what they accept:</p>

<div class="op-compare">
  <div class="opc-row">
    <span class="opc-label opc-bad">a | b</span>
    <span class="opc-tag-bad">set only</span>
    <span class="opc-desc">operator: BOTH sides must be sets — <code style="background:#3a1e1e;padding:1px 5px;border-radius:4px;color:#f38ba8;font-family:'Courier New',monospace">a | [5,6]</code> raises TypeError</span>
  </div>
  <div class="opc-row">
    <span class="opc-label opc-good">a.union([5,6,7])</span>
    <span class="opc-tag-good">any iterable</span>
    <span class="opc-desc">method: takes ANY iterable — list, generator, range — and works fine</span>
  </div>
</div>

<div class="note-box">💡 When you want to union a set with a list or generator, use the method — it'll iterate any iterable. The operator is stricter (set-only) but reads cleaner when both are already sets.</div>

<h3 class="sub">1.5 Mutating methods</h3>

<p class="body-p">Sets are mutable (you can add/remove), with one subtlety around missing elements:</p>

<div class="method-block">
  <div class="method-title"><span class="method-badge">add(x)</span><span class="method-ret-none">returns None</span></div>
  <div class="method-body">Add one element.</div>
  <div class="code-block" style="margin:8px 0 0 0"><pre>s.<span class="cm">add</span>(<span class="cn">4</span>)</pre></div>
</div>

<div class="method-block">
  <div class="method-title"><span class="method-badge">discard(x)</span><span class="method-ret-none">returns None</span></div>
  <div class="method-body">Remove if present — <strong>no error if missing</strong>. Safe choice for uncertain data.</div>
  <div class="code-block" style="margin:8px 0 0 0"><pre>s.<span class="cm">discard</span>(<span class="cn">10</span>)       <span class="cc"># nothing happens if 10 not in s</span></pre></div>
</div>

<div class="method-block">
  <div class="method-title"><span class="method-badge">remove(x)</span><span class="method-ret-none">returns None</span></div>
  <div class="method-body">Remove — <strong>KeyError if missing</strong>. Use when the element must be present; error loudly if not.</div>
  <div class="code-block" style="margin:8px 0 0 0"><pre>s.<span class="cm">remove</span>(<span class="cn">10</span>)        <span class="cc"># KeyError if 10 not in s</span></pre></div>
</div>

<div class="method-block">
  <div class="method-title"><span class="method-badge">pop()</span></div>
  <div class="method-body">Remove and return an <strong>arbitrary</strong> element (unordered — not "the last"). Don't rely on which one you get.</div>
</div>

<div class="method-block">
  <div class="method-title"><span class="method-badge">clear()</span><span class="method-ret-none">returns None</span></div>
  <div class="method-body">Empty the set.</div>
</div>

<div class="method-block">
  <div class="method-title"><span class="method-badge">update(iterable)</span> <span style="color:#6c7086;font-size:0.85em">also |=</span><span class="method-ret-none">returns None</span></div>
  <div class="method-body">Add many elements at once from any iterable.</div>
</div>

<div class="warn-box">
  ⚠️ <strong>Key distinction:</strong> <code style="background:#3a2a00;padding:1px 5px;border-radius:4px;color:#f9e2af;font-family:'Courier New',monospace">discard</code> is safe (silent on missing), <code style="background:#3a2a00;padding:1px 5px;border-radius:4px;color:#f9e2af;font-family:'Courier New',monospace">remove</code> raises KeyError on missing. And <code style="background:#3a2a00;padding:1px 5px;border-radius:4px;color:#f9e2af;font-family:'Courier New',monospace">pop()</code> returns an arbitrary element — not "the last" — there's no order, so don't rely on which one.
</div>

<h3 class="sub">1.6 frozenset — the immutable, hashable set</h3>

<p class="body-p">Here's the elegant parallel to list→tuple. A <span class="ic">frozenset</span> is to a set what a tuple is to a list: the immutable version. Once created, you can't add or remove. And because it's immutable, it's hashable — which means a frozenset can do what a regular set cannot:</p>

<div class="code-block"><pre>fs = <span class="cm">frozenset</span>([<span class="cn">1</span>, <span class="cn">2</span>, <span class="cn">3</span>])
d = {fs: <span class="cs">"value"</span>}                       <span class="cc"># frozenset as a DICT KEY</span>
nested = {<span class="cm">frozenset</span>([<span class="cn">1</span>,<span class="cn">2</span>]), <span class="cm">frozenset</span>([<span class="cn">3</span>,<span class="cn">4</span>])}   <span class="cc"># a SET of sets</span>
fs.<span class="cm">add</span>(<span class="cn">5</span>)                               <span class="cc"># AttributeError — immutable</span></pre></div>

<div class="note-box">
  💡 <strong>Why it exists:</strong> a regular set can't be an element of another set, or a dict key, because it's mutable (unhashable) — the same reason a list can't. <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1;font-family:'Courier New',monospace">frozenset</code> solves that. Use it when you need "a set as a key" or "a set of sets": grouping by a collection of tags, caching results keyed by a set of parameters, representing graph edges as unordered pairs.
</div>

<h3 class="sub">1.7 Set comprehension and construction</h3>

<p class="body-p">Sets have a comprehension form (like lists/dicts):</p>

<div class="code-block"><pre>{x*x <span class="ck">for</span> x <span class="ck">in</span> <span class="cm">range</span>(<span class="cn">6</span>)}    <span class="cc"># {0, 1, 4, 9, 16, 25}</span></pre></div>

<div class="warn-box">
  ⚠️ <strong>Construction:</strong> <code style="background:#3a2a00;padding:1px 5px;border-radius:4px;color:#f9e2af;font-family:'Courier New',monospace">{1, 2, 3}</code> (literal), <code style="background:#3a2a00;padding:1px 5px;border-radius:4px;color:#f9e2af;font-family:'Courier New',monospace">set(iterable)</code> (from any iterable — the dedup idiom), <code style="background:#3a2a00;padding:1px 5px;border-radius:4px;color:#f9e2af;font-family:'Courier New',monospace">set()</code> (empty — <strong>not</strong> <code style="background:#3a2a00;padding:1px 5px;border-radius:4px;color:#f9e2af;font-family:'Courier New',monospace">{}</code>, which is a dict — see Edge cases).
</div>

<div class="ex-header"><h2>2. Example (runnable, step by step)</h2></div>

<div class="ex-block">
  <div class="ex-title"><span class="ex-badge">Example 1</span> Dedup: the everyday set use</div>
  <div class="code-block"><pre>records = [<span class="cn">1</span>, <span class="cn">2</span>, <span class="cn">2</span>, <span class="cn">3</span>, <span class="cn">1</span>, <span class="cn">4</span>, <span class="cn">3</span>]
unique  = <span class="cm">set</span>(records)                    <span class="cc"># {1, 2, 3, 4} — order NOT preserved</span>
ordered = <span class="cm">list</span>(<span class="cm">dict</span>.<span class="cm">fromkeys</span>(records))   <span class="cc"># [1, 2, 3, 4] — order preserved</span></pre></div>
  <div class="output-block">{1, 2, 3, 4}<br>[1, 2, 3, 4]</div>
  <div class="step-explain">
    <strong>Step by step:</strong> <code>set(records)</code> collapses duplicates instantly (O(n)) but loses order. When you need order and dedup, <code>dict.fromkeys</code> (2A/2B callback) is the idiom. This is your OpsRAG dedup at its core — collapsing 27,760 unique records from a larger raw set is exactly <code>set()</code> (or a seen-set with a custom key when the dedup key is derived).
  </div>
</div>

<div class="ex-block">
  <div class="ex-title"><span class="ex-badge">Example 2</span> Membership: the O(1) speedup in practice</div>
  <div class="code-block"><pre>valid_ids = <span class="cm">set</span>(<span class="cm">load_all_user_ids</span>())     <span class="cc"># build once, O(n)</span>
<span class="ck">for</span> record <span class="ck">in</span> incoming:                  <span class="cc"># then check membership O(1) each</span>
    <span class="ck">if</span> record.user_id <span class="ck">in</span> valid_ids:      <span class="cc"># instant, even for millions of ids</span>
        <span class="cm">process</span>(record)</pre></div>
  <div class="step-explain">
    The pattern: build a set once from the lookup collection, then do fast repeated membership checks. If <code>valid_ids</code> were a list, each <code>in</code> check would scan it — turning the loop into O(n·m). The set makes it O(n+m). This is 2A's Rule 2, now you understand the hash-table reason.
  </div>
</div>

<div class="ex-block">
  <div class="ex-title"><span class="ex-badge">Example 3</span> Set algebra for real questions</div>
  <div class="code-block"><pre>last_week = {<span class="cs">"alice"</span>, <span class="cs">"bob"</span>, <span class="cs">"carol"</span>}
this_week = {<span class="cs">"bob"</span>, <span class="cs">"carol"</span>, <span class="cs">"dave"</span>}

new_users = this_week - last_week    <span class="cc"># {'dave'}        — joined</span>
churned   = last_week - this_week    <span class="cc"># {'alice'}       — left</span>
retained  = last_week &amp; this_week    <span class="cc"># {'bob','carol'} — stayed</span>
all_seen  = last_week | this_week    <span class="cc"># everyone ever</span></pre></div>
  <div class="step-explain">
    Each question — new, churned, retained — is one clean set operation. The list equivalent would be nested loops with membership tests. This is genuinely how you'd compute cohort changes, feature overlaps, or vocabulary differences.
  </div>
</div>

<div class="ex-block">
  <div class="ex-title"><span class="ex-badge">Example 4</span> frozenset as a dict key</div>
  <div class="code-block"><pre><span class="cc"># cache results keyed by a SET of parameters (order-independent)</span>
cache = {}
cache[<span class="cm">frozenset</span>([<span class="cs">"lr"</span>, <span class="cs">"batch"</span>])] = <span class="cs">"config A"</span>
cache[<span class="cm">frozenset</span>([<span class="cs">"batch"</span>, <span class="cs">"lr"</span>])]            <span class="cc"># same key! -> "config A"</span></pre></div>
  <div class="step-explain">
    Because a frozenset is unordered and hashable, <code>{"lr", "batch"}</code> and <code>{"batch", "lr"}</code> are the same key — useful when the key is "a collection of things where order doesn't matter."
  </div>
</div>

<!-- ══ SECTION 3: EDGE CASES ══ -->
<div class="warn-header"><h2>3. Edge Cases</h2></div>
<p class="body-p">Six cases, all outputs confirmed.</p>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 1</span> {} is an empty dict, not an empty set</div>
  <div class="code-block"><pre><span class="cm">type</span>({})        <span class="cc"># dict   ← NOT a set!</span>
<span class="cm">type</span>(<span class="cm">set</span>())     <span class="cc"># set    ← the only way to make an empty set</span></pre></div>
  <div class="output-block">{} is actually: dict &nbsp;&nbsp;&nbsp; empty set MUST be set()</div>
  <div class="why-box">
    <strong>Why it matters:</strong> the <code>{}</code> literal was claimed by dicts first, so an empty set has no literal — you must write <code>set()</code>. Writing <code>s = {}</code> thinking you made a set gives you a dict, and <code>s.add(x)</code> then raises <code>AttributeError</code>. A genuinely common beginner bug.
  </div>
</div>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 2</span> Sets are unordered; don't rely on iteration order</div>
  <div class="code-block"><pre>s = {<span class="cn">3</span>, <span class="cn">1</span>, <span class="cn">2</span>}
<span class="cm">list</span>(s)         <span class="cc"># may be [1,2,3] or any order — NOT guaranteed</span>
s.<span class="cm">pop</span>()         <span class="cc"># removes an ARBITRARY element</span></pre></div>
  <div class="why-box">
    <strong>Why it matters:</strong> unlike dicts (insertion-ordered since 3.7) and lists, sets have no order guarantee. Iterating a set, or <code>pop()</code>-ing from it, gives elements in hash-table order, which can vary. Never write code that depends on set ordering. If you need order, use a list or <code>dict.fromkeys</code>.
  </div>
</div>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 3</span> Elements must be hashable</div>
  <div class="code-block"><pre>{<span class="cn">1</span>, <span class="cs">"a"</span>, (<span class="cn">2</span>, <span class="cn">3</span>)}    <span class="cc"># fine — all hashable</span>
{[<span class="cn">1</span>, <span class="cn">2</span>]}            <span class="cc"># TypeError: unhashable type: 'list'</span></pre></div>
  <div class="output-block">list in set: unhashable type: 'list'</div>
  <div class="why-box">
    <strong>Why it matters:</strong> you can't put lists, dicts, or sets into a set (they're mutable/unhashable). To store collections in a set, convert them to tuples or frozensets first. This is the same hashability rule from 2B, now governing set membership.
  </div>
</div>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 4</span> remove raises, discard is silent</div>
  <div class="code-block"><pre>s = {<span class="cn">1</span>, <span class="cn">2</span>, <span class="cn">3</span>}
s.<span class="cm">discard</span>(<span class="cn">10</span>)   <span class="cc"># no error, nothing happens</span>
s.<span class="cm">remove</span>(<span class="cn">10</span>)    <span class="cc"># KeyError: 10</span></pre></div>
  <div class="output-block">discard missing (no error) &nbsp;&nbsp;&nbsp; remove missing raises KeyError</div>
  <div class="why-box">
    <strong>Why it matters:</strong> choose based on intent. <code>discard</code> when "remove it if it's there." <code>remove</code> when "it should be there, error loudly if not." Using <code>remove</code> in a loop over uncertain data is a common source of unexpected <code>KeyError</code>.
  </div>
</div>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 5</span> Operator needs a set; method takes any iterable</div>
  <div class="code-block"><pre>a = {<span class="cn">1</span>, <span class="cn">2</span>, <span class="cn">3</span>}
a | [<span class="cn">4</span>, <span class="cn">5</span>]           <span class="cc"># TypeError — operator needs a set</span>
a.<span class="cm">union</span>([<span class="cn">4</span>, <span class="cn">5</span>])      <span class="cc"># {1,2,3,4,5} — method takes any iterable</span></pre></div>
  <div class="output-block">a | list fails: TypeError &nbsp;&nbsp;&nbsp; a.union([5,6,7]) works</div>
  <div class="why-box">
    <strong>Why it matters:</strong> if one operand is a list/generator, you must use the method form (<code>.union</code>, <code>.intersection</code>, etc.), not the operator. Reaching for <code>|</code> with a non-set raises <code>TypeError</code>.
  </div>
</div>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 6</span> A regular set can't be a set element or dict key; frozenset can</div>
  <div class="code-block"><pre>{{<span class="cn">1</span>, <span class="cn">2</span>}, {<span class="cn">3</span>, <span class="cn">4</span>}}                       <span class="cc"># TypeError: unhashable type: 'set'</span>
{<span class="cm">frozenset</span>([<span class="cn">1</span>,<span class="cn">2</span>]), <span class="cm">frozenset</span>([<span class="cn">3</span>,<span class="cn">4</span>])}   <span class="cc"># fine</span></pre></div>
  <div class="output-block">set of frozensets: works (confirmed)</div>
  <div class="why-box">
    <strong>Why it matters:</strong> a set of sets is impossible with regular sets (mutable → unhashable), exactly like a list-of-lists-as-keys fails. <code>frozenset</code> is the fix, the same way tuple fixes it for lists. This is the list→tuple / set→frozenset symmetry made concrete.
  </div>
</div>

<div class="part-header"><h2>4. Golden Rules</h2></div>

<h3 class="sub">When to reach for a set — the decision framework (full depth)</h3>

<p class="body-p">A set earns its place for exactly three jobs. If your problem is one of these, a set is almost always the right tool; if it isn't, a set is probably the wrong one.</p>

<div class="job-block">
  <div class="job-title"><span class="job-badge">Job 1</span> Fast membership testing</div>
  <div class="job-body">
    The moment you're checking "is x in this collection?" repeatedly — especially inside a loop — convert the collection to a set once and check against it. This is the highest-value use. The reasoning is the hash-table O(1) from Part 1: a list makes each check O(n), so a loop of m checks against n items is O(n·m); a set makes it O(m). The live proof was 2,800×.
  </div>
  <div class="signal-box">🔔 <strong>Signal:</strong> you see <code>if x in some_list</code> inside a loop, and the list doesn't change during the loop.</div>
</div>

<div class="job-block">
  <div class="job-title"><span class="job-badge">Job 2</span> Deduplication</div>
  <div class="job-body">
    When you need unique elements, <code>set(iterable)</code> collapses duplicates in one O(n) pass. Use plain <code>set()</code> when order doesn't matter; use <code>dict.fromkeys()</code> when you need to keep first-seen order (sets don't preserve order — Job 2's one caveat).
  </div>
  <div class="signal-box">🔔 <strong>Signal:</strong> "I only want the distinct values."</div>
</div>

<div class="job-block">
  <div class="job-title"><span class="job-badge">Job 3</span> Set algebra</div>
  <div class="job-body">
    When your question is naturally about overlap, difference, or combination between two collections — union, intersection, difference, symmetric difference. These are one clean operator each (<code>|</code>, <code>&amp;</code>, <code>-</code>, <code>^</code>) versus nested loops with lists.
  </div>
  <div class="signal-box">🔔 <strong>Signal:</strong> questions phrased as "in both," "in either," "in A but not B," "which are new/gone."</div>
</div>

<div class="warn-box">
  ⚠️ <strong>When NOT to use a set:</strong> when you need order (use list or <code>dict.fromkeys</code>), when you need indexing/positional access (<code>s[0]</code> doesn't exist), when you need duplicates (sets can't hold them), or when elements aren't hashable (lists/dicts as elements — use tuples/frozensets, or a different structure).
</div>

<h3 class="sub">Other rules, kept tight:</h3>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 4</span> Empty set is <code style="background:#313244;padding:1px 5px;border-radius:4px;color:#a6e3a1;font-family:'Courier New',monospace">set()</code>, never <code style="background:#313244;padding:1px 5px;border-radius:4px;color:#a6e3a1;font-family:'Courier New',monospace">{}</code>.</div>
  <div class="rule-body"><code>{}</code> is an empty dict. This is a hard syntax fact, not a style choice.</div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 5</span> Build the set once, outside the loop.</div>
  <div class="rule-body"><code>valid = set(big_list)</code> before the loop, then check <code>x in valid</code> inside. Rebuilding the set each iteration throws away the whole advantage.</div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 6</span> Use the method form when an operand is a list/generator.</div>
  <div class="rule-body">Operators (<code>|</code>, <code>&amp;</code>) require both sides to be sets; methods (<code>.union</code>, <code>.intersection</code>) accept any iterable.</div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 7</span> <code style="background:#313244;padding:1px 5px;border-radius:4px;color:#a6e3a1;font-family:'Courier New',monospace">discard</code> to remove-if-present, <code style="background:#313244;padding:1px 5px;border-radius:4px;color:#a6e3a1;font-family:'Courier New',monospace">remove</code> to remove-or-error.</div>
  <div class="rule-body">Pick by intent; <code>discard</code> avoids <code>KeyError</code> on uncertain data.</div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 8</span> Reach for <code style="background:#313244;padding:1px 5px;border-radius:4px;color:#a6e3a1;font-family:'Courier New',monospace">frozenset</code> when you need a set as a dict key or a set element.</div>
  <div class="rule-body">Same as choosing tuple over list for hashability.</div>
</div>

<!-- ══ SECTION 5: COMMON TRAPS ══ -->
<div class="trap-header"><h2>5. Common Traps</h2></div>

<div class="trap-block">
  <div class="trap-title">
    <span class="trap-badge">Trap 1</span> {} is a dict, not a set.
    <span class="depth-tag">the classic</span>
  </div>
  <div class="code-block"><pre>s = {}
s.<span class="cm">add</span>(<span class="cn">1</span>)     <span class="cc"># AttributeError: 'dict' object has no attribute 'add'</span></pre></div>
  <div class="tell-box"><strong>Tell:</strong> <code>AttributeError</code> about <code>add</code>/<code>discard</code> on something you thought was a set.</div>
  <div class="fix-box">Fix: <code>s = set()</code></div>
</div>

<div class="trap-block">
  <div class="trap-title">
    <span class="trap-badge">Trap 2</span> Relying on set iteration order.
    <span class="depth-tag">full depth — subtle and non-deterministic</span>
  </div>
  <div class="code-block"><pre>s = {<span class="cn">10</span>, <span class="cn">3</span>, <span class="cn">7</span>, <span class="cn">1</span>, <span class="cn">22</span>}
<span class="cm">list</span>(s)      <span class="cc"># [1, 3, 22, 7, 10]  -- NOT sorted, NOT insertion order</span></pre></div>
  <div class="output-block">[1, 3, 22, 7, 10]</div>
  <div class="why-box">
    The order is neither sorted nor the order you added them — it's hash-table layout. The danger is that it <em>looks</em> stable in small tests, so code that accidentally depends on set order (taking "the first element," building output from set iteration) passes locally and breaks unpredictably with different data or Python versions. This is stricter than the list case — with lists, order is guaranteed; with sets, assuming order is always a latent bug.
  </div>
  <div class="tell-box"><strong>Tell:</strong> output whose order changes between runs or environments.</div>
  <div class="fix-box">Fix: if order matters, sort explicitly (<code>sorted(s)</code>) or use <code>dict.fromkeys</code> to preserve insertion order.</div>
</div>

<div class="trap-block">
  <div class="trap-title">
    <span class="trap-badge">Trap 3</span> Modifying a set while iterating raises.
    <span class="depth-tag">full depth — and it's an error, unlike lists</span>
  </div>
  <div class="code-block"><pre>s = {<span class="cn">1</span>, <span class="cn">2</span>, <span class="cn">3</span>, <span class="cn">4</span>, <span class="cn">5</span>}
<span class="ck">for</span> x <span class="ck">in</span> s:
    <span class="ck">if</span> x % <span class="cn">2</span> == <span class="cn">0</span>:
        s.<span class="cm">discard</span>(x)      <span class="cc"># RuntimeError: Set changed size during iteration</span></pre></div>
  <div class="output-block">RuntimeError: Set changed size during iteration</div>
  <div class="why-box">
    This raises <code>RuntimeError</code> immediately. Recall 2A Trap 2: mutating a list while iterating fails silently (skips elements, no error). A set is stricter — it detects the size change mid-iteration and raises loudly. That's arguably better (fails fast), but it's a different behavior to know.
  </div>
  <div class="fix-box">Fix: build a new set with a comprehension:</div>
  <div class="code-block"><pre>s = {x <span class="ck">for</span> x <span class="ck">in</span> s <span class="ck">if</span> x % <span class="cn">2</span> != <span class="cn">0</span>}   <span class="cc"># {1, 3, 5}</span></pre></div>
  <div class="note-box">Same principle as lists — don't mutate the thing you're iterating; construct a new one.</div>
</div>

<div class="trap-block">
  <div class="trap-title">
    <span class="trap-badge">Trap 4</span> set() on a string splits it into characters.
    <span class="depth-tag">tight</span>
  </div>
  <div class="code-block"><pre><span class="cm">set</span>(<span class="cs">"hello"</span>)     <span class="cc"># {'h', 'e', 'l', 'o'}  -- not {"hello"}</span></pre></div>
  <div class="output-block">{'h', 'e', 'l', 'o'}</div>
  <div class="why-box">
    <code>set()</code> takes an iterable and a string iterates into characters (same as <code>list("hello")</code> and <code>extend</code> from 2A). To make a one-element set, use the literal: <code>{"hello"}</code>.
  </div>
  <div class="tell-box"><strong>Tell:</strong> a set full of single characters when you expected one whole string.</div>
</div>

<!-- SECTION 4: GOLDEN RULES -->
<div class="part-header" style="border-left-color:#a6e3a1"><h2 style="color:#a6e3a1">4. Golden Rules</h2></div>

<p class="body-p"><strong style="color:#cba6f7">When to reach for a set — the decision framework (full depth)</strong></p>
<p class="body-p">A set earns its place for exactly three jobs. If your problem is one of these, a set is almost always the right tool; if it isn't, a set is probably the wrong one.</p>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Job 1</span> Fast membership testing</div>
  <p class="body-p" style="margin:0 0 8px 0">The moment you're checking "is x in this collection?" repeatedly — especially inside a loop — convert the collection to a set once and check against it. This is the highest-value use. The reasoning is the hash-table O(1) from Part 1: a list makes each check O(n), so a loop of m checks against n items is O(n·m); a set makes it O(m). The live proof was 2,800×.</p>
  <div class="note-box"><strong>Signal you need this:</strong> you see <code>if x in some_list</code> inside a loop, and the list doesn't change during the loop.</div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Job 2</span> Deduplication</div>
  <p class="body-p" style="margin:0 0 8px 0">When you need unique elements, <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">set(iterable)</code> collapses duplicates in one O(n) pass. Use plain <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">set()</code> when order doesn't matter; use <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">dict.fromkeys()</code> when you need to keep first-seen order (sets don't preserve order — Job 2's one caveat).</p>
  <div class="note-box"><strong>Signal:</strong> "I only want the distinct values."</div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Job 3</span> Set algebra</div>
  <p class="body-p" style="margin:0 0 8px 0">When your question is naturally about overlap, difference, or combination between two collections — union, intersection, difference, symmetric difference. These are one clean operator each (<code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">|</code>, <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">&</code>, <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">-</code>, <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">^</code>) versus nested loops with lists.</p>
  <div class="note-box"><strong>Signal:</strong> questions phrased as "in both," "in either," "in A but not B," "which are new/gone."</div>
</div>

<div class="warn-box">
  <strong>When NOT to use a set:</strong> when you need order (use list or <code style="background:#3a2a00;padding:2px 6px;border-radius:4px;color:#f9e2af">dict.fromkeys</code>), when you need indexing/positional access (<code style="background:#3a2a00;padding:2px 6px;border-radius:4px;color:#f9e2af">s[0]</code> doesn't exist), when you need duplicates (sets can't hold them), or when elements aren't hashable (lists/dicts as elements — use tuples/frozensets, or a different structure).
</div>

<p class="body-p"><strong style="color:#cba6f7">Other rules, kept tight:</strong></p>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 4</span> Empty set is <code style="background:#1a2e1a;padding:2px 6px;border-radius:4px;color:#a6e3a1">set()</code>, never <code style="background:#1a2e1a;padding:2px 6px;border-radius:4px;color:#a6e3a1">{}</code></div>
  <p class="body-p" style="margin:0"><code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">{}</code> is an empty dict. This is a hard syntax fact, not a style choice.</p>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 5</span> Build the set once, outside the loop</div>
  <p class="body-p" style="margin:0"><code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">valid = set(big_list)</code> before the loop, then check <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">x in valid</code> inside. Rebuilding the set each iteration throws away the whole advantage.</p>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 6</span> Use the method form when an operand is a list/generator</div>
  <p class="body-p" style="margin:0">Use <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">.union</code>, <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">.intersection</code> when an operand is a list/generator. Operators (<code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">|</code>, <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">&</code>) require both sides to be sets; methods accept any iterable.</p>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 7</span> <code style="background:#1a2e1a;padding:2px 6px;border-radius:4px;color:#a6e3a1">discard</code> to remove-if-present, <code style="background:#1a2e1a;padding:2px 6px;border-radius:4px;color:#a6e3a1">remove</code> to remove-or-error</div>
  <p class="body-p" style="margin:0">Pick by intent; <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">discard</code> avoids <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">KeyError</code> on uncertain data.</p>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 8</span> Reach for <code style="background:#1a2e1a;padding:2px 6px;border-radius:4px;color:#a6e3a1">frozenset</code> when you need a set as a dict key or a set element</div>
  <p class="body-p" style="margin:0">Same as choosing tuple over list for hashability.</p>
</div>

<!-- SECTION 5: COMMON TRAPS -->
<div class="part-header" style="border-left-color:#f38ba8"><h2 style="color:#f38ba8">5. Common Traps</h2></div>

<div class="trap-block">
  <div class="trap-title"><span class="trap-badge">Trap 1</span> {} is a dict, not a set — the classic</div>
  <div class="code-block"><pre>s = {}
s.add(<span class="cn">1</span>)     <span class="cc"># AttributeError: 'dict' object has no attribute 'add'</span></pre></div>
  <div class="why-box">You meant an empty set, you got a dict, and it fails the moment you call a set method. The tell: <code>AttributeError</code> about <code>add</code>/<code>discard</code> on something you thought was a set. Fix: <code>s = set()</code>.</div>
</div>

<div class="trap-block">
  <div class="trap-title"><span class="trap-badge">Trap 2</span> Relying on set iteration order — subtle and non-deterministic</div>
  <div class="code-block"><pre>s = {<span class="cn">10</span>, <span class="cn">3</span>, <span class="cn">7</span>, <span class="cn">1</span>, <span class="cn">22</span>}
<span class="cm">list</span>(s)      <span class="cc"># [1, 3, 22, 7, 10]  -- NOT sorted, NOT insertion order</span></pre></div>
  <div class="why-box">The order is neither sorted nor the order you added them; it's hash-table layout. The danger is that it looks stable in small tests, so code that accidentally depends on set order (taking "the first element," building output from set iteration) passes locally and breaks unpredictably with different data or Python versions. The tell: output whose order changes between runs or environments. Fix: if order matters, sort explicitly (<code>sorted(s)</code>) or use <code>dict.fromkeys</code> to preserve insertion order. This is stricter than the list case — with lists, order is guaranteed; with sets, assuming order is always a latent bug.</div>
</div>

<div class="trap-block">
  <div class="trap-title"><span class="trap-badge">Trap 3</span> Modifying a set while iterating raises — and it's an error, unlike lists</div>
  <div class="code-block"><pre>s = {<span class="cn">1</span>, <span class="cn">2</span>, <span class="cn">3</span>, <span class="cn">4</span>, <span class="cn">5</span>}
<span class="ck">for</span> x <span class="ck">in</span> s:
    <span class="ck">if</span> x % <span class="cn">2</span> == <span class="cn">0</span>:
        s.discard(x)      <span class="cc"># RuntimeError: Set changed size during iteration</span></pre></div>
  <div class="why-box">This raises <code>RuntimeError</code> immediately. Recall 2A Trap 2: mutating a list while iterating fails silently (skips elements, no error). A set is stricter — it detects the size change mid-iteration and raises loudly. That's arguably better (fails fast), but it's a different behavior to know. Fix: build a new set with a comprehension:</div>
  <div class="code-block"><pre>s = {x <span class="ck">for</span> x <span class="ck">in</span> s <span class="ck">if</span> x % <span class="cn">2</span> != <span class="cn">0</span>}   <span class="cc"># {1, 3, 5}</span></pre></div>
  <div class="why-box">Same principle as lists — don't mutate the thing you're iterating; construct a new one.</div>
</div>

<div class="trap-block">
  <div class="trap-title"><span class="trap-badge">Trap 4</span> set() on a string splits it into characters</div>
  <div class="code-block"><pre><span class="cm">set</span>(<span class="cs">"hello"</span>)     <span class="cc"># {'h', 'e', 'l', 'o'}  -- not {"hello"}</span></pre></div>
  <div class="why-box"><code>set()</code> takes an iterable and a string iterates into characters (same as <code>list("hello")</code> and extend from 2A). To make a one-element set, use the literal: <code>{"hello"}</code>. The tell: a set full of single characters when you expected one whole string.</div>
</div>

<div class="part-header" style="border-left-color:#a6e3a1"><h2 style="color:#a6e3a1">7. ML Real-World Connection</h2></div>

<p class="body-p">Sets are workhorses in ML data prep — dedup, vocabulary, fast lookups, and dataset comparison are all set operations. This section is the most directly relevant to your OpsRAG work of the whole sub-session.</p>

<div class="ml-block">
  <div class="ml-title"><span class="ml-badge">Use 1</span> Deduplication pipeline — this is your OpsRAG dedup</div>
  <p class="body-p" style="margin:0 0 8px 0">The seen-set pattern with a derived key is exactly how you collapsed raw records to 27,760 clean ones:</p>
  <div class="code-block"><pre>seen = <span class="cm">set</span>()
unique = []
<span class="ck">for</span> r <span class="ck">in</span> records:
    key = r[<span class="cs">"text"</span>]              <span class="cc"># or a content hash, or (title, body)</span>
    <span class="ck">if</span> key <span class="ck">not in</span> seen:          <span class="cc"># O(1) membership</span>
        seen.add(key)
        unique.append(r)</pre></div>
  <div class="why-box">4 records → 2 unique. The set tracks what you've already kept (O(1) checks), the list preserves order and holds the full records. When the dedup key is derived (content hash, normalized text, accepted-answer-vs-highest-score logic), this is the pattern — a plain <code>set(records)</code> won't work because dicts/records aren't hashable and you need custom key logic. This is precisely your data-pipeline dedup, and now you understand its O(n) complexity and why the set is what makes it fast.</div>
</div>

<div class="ml-block">
  <div class="ml-title"><span class="ml-badge">Use 2</span> Vocabulary building — the first step of any NLP/embedding pipeline</div>
  <div class="code-block"><pre>vocab = <span class="cm">set</span>()
<span class="ck">for</span> doc <span class="ck">in</span> corpus:
    vocab.update(doc.split())          <span class="cc"># accumulate unique tokens</span>
word_to_idx = {w: i <span class="ck">for</span> i, w <span class="ck">in</span> <span class="cm">enumerate</span>(<span class="cm">sorted</span>(vocab))}   <span class="cc"># token -> integer id</span></pre></div>
  <div class="why-box">3 docs → vocab of 6 → a word→index mapping. Every tokenizer, every embedding layer, every bag-of-words model starts by building the set of unique tokens, then assigning each an integer index. <code>set.update()</code> (add many) accumulates the vocabulary across documents; <code>sorted()</code> then gives a deterministic ordering for the index map (note: you must sort, because set order isn't stable — Trap 2 in action). Directly relevant to the retrieval side of OpsRAG.</div>
</div>

<div class="ml-block">
  <div class="ml-title"><span class="ml-badge">Use 3</span> Fast label/whitelist lookup</div>
  <div class="code-block"><pre>valid_labels = {<span class="cs">"cat"</span>, <span class="cs">"dog"</span>, <span class="cs">"bird"</span>}
valid = [p <span class="ck">for</span> p <span class="ck">in</span> predictions <span class="ck">if</span> p <span class="ck">in</span> valid_labels]   <span class="cc"># O(1) per check</span></pre></div>
  <div class="why-box">Filtering predictions against an allowed set, validating categories, checking IDs against a whitelist — build the set once, check membership fast. The alternative (list membership) turns filtering m predictions against n labels into O(n·m).</div>
</div>

<div class="ml-block">
  <div class="ml-title"><span class="ml-badge">Use 4</span> Feature overlap between train and test — catching leakage and drift</div>
  <div class="code-block"><pre>train_features & test_features    <span class="cc"># {'age','income','score'}  - common</span>
train_features - test_features    <span class="cc"># {'region'}   - train-only (potential leakage)</span>
test_features - train_features    <span class="cc"># {'country'}  - missing in train (will break inference)</span></pre></div>
  <div class="why-box">Set algebra answers real dataset-integrity questions in one line each: which features are shared, which exist only in train (a leakage/mismatch signal), which the model will encounter at inference but never saw in training. This is genuinely how you'd audit a feature pipeline before training.</div>
</div>

<div class="part-header" style="border-left-color:#f9e2af"><h2 style="color:#f9e2af">8. Interview Questions</h2></div>

<div class="part-header" style="border-left-color:#89b4fa; margin-top:10px"><h2 style="color:#89b4fa; font-size:1.1em;">8a. Conceptual Q&amp;A</h2></div>

<div class="qa-block">
  <div class="qa-q">Q1. How is a set implemented, and why is membership O(1)?</div>
  <p class="qa-a">As a hash table. <code>x in s</code> computes <code>hash(x)</code>, jumps directly to the bucket where x would be, and checks — no scan. Average O(1), independent of size.</p>
</div>

<div class="qa-block">
  <div class="qa-q">Q2. Set vs list — when do you choose a set?</div>
  <p class="qa-a">For fast membership testing, deduplication, or set algebra (union/intersection/difference). Not when you need order, indexing, duplicates, or unhashable elements.</p>
</div>

<div class="qa-block">
  <div class="qa-q">Q3. What can go into a set?</div>
  <p class="qa-a">Only hashable (effectively immutable) elements: numbers, strings, tuples-of-immutables, frozensets. Not lists, dicts, or sets.</p>
</div>

<div class="qa-block">
  <div class="qa-q">Q4. How much faster is set membership than list membership?</div>
  <p class="qa-a">List is O(n) (linear scan), set is O(1) (hash lookup). At scale it's dramatic — often hundreds to thousands of times faster for large collections.</p>
</div>

<div class="qa-block">
  <div class="qa-q">Q5. What's the difference between remove and discard?</div>
  <p class="qa-a">Both remove an element; <code>remove</code> raises <code>KeyError</code> if it's missing, <code>discard</code> does nothing. Use <code>discard</code> for "remove if present."</p>
</div>

<div class="qa-block">
  <div class="qa-q">Q6. Why is {} not an empty set?</div>
  <p class="qa-a"><code>{}</code> is an empty dict — the literal was taken by dicts. An empty set is <code>set()</code>.</p>
</div>

<div class="qa-block">
  <div class="qa-q">Q7. What's a frozenset and why does it exist?</div>
  <p class="qa-a">An immutable, hashable set. Because it can't change, it can be a dict key or an element of another set — things a mutable set can't do. It's the set→frozenset analog of list→tuple.</p>
</div>

<div class="qa-block">
  <div class="qa-q">Q8. Do sets preserve insertion order?</div>
  <p class="qa-a">No. Sets are unordered (hash-table layout); iteration order isn't guaranteed and shouldn't be relied on. Dicts, by contrast, preserve insertion order since Python 3.7.</p>
</div>

<div class="qa-block">
  <div class="qa-q">Q9. How do you dedup a list while preserving order?</div>
  <p class="qa-a">Not with a plain set (loses order). Use <code>dict.fromkeys(seq)</code> or a seen-set with an ordered output list.</p>
</div>

<div class="qa-block">
  <div class="qa-q">Q10. What are the four main set operations and their operators?</div>
  <p class="qa-a">Union <code>|</code> (in either), intersection <code>&amp;</code> (in both), difference <code>-</code> (in first not second), symmetric difference <code>^</code> (in exactly one).</p>
</div>
