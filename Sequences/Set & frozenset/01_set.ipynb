<div class="part-header"><h2>🔒 Hashability Rule — What Can Go Inside a Set</h2></div>

<div class="theory-box">
  You can't put <strong>lists</strong>, <strong>dicts</strong>, or <strong>sets</strong> into a set — they're mutable and therefore unhashable. To store collections in a set, convert them to <strong>tuples</strong> or <strong>frozensets</strong> first. This is the same hashability rule from Session 2B, now governing set membership.
</div>

<hr class="divider">

<div class="hash-grid">
  <div class="hash-card hc-good">
    <div class="hc-title">✅ Hashable — allowed in sets</div>
    <div class="hc-body">
      int, float, bool<br>
      str<br>
      tuple (of immutables)<br>
      frozenset
    </div>
  </div>
  <div class="hash-card hc-bad">
    <div class="hc-title">❌ Unhashable — will crash</div>
    <div class="hc-body">
      list<br>
      dict<br>
      set<br>
      tuple containing a list
    </div>
  </div>
</div>

<hr class="divider">

<div class="danger-box">
  ❌ <strong>These all raise TypeError: unhashable type</strong>
</div>

<div class="code-block"><pre>{[<span class="cn">1</span>, <span class="cn">2</span>], [<span class="cn">3</span>, <span class="cn">4</span>]}          <span class="cc"># TypeError: unhashable type: 'list'</span>
{{<span class="cs">"a"</span>: <span class="cn">1</span>}, {<span class="cs">"b"</span>: <span class="cn">2</span>}}       <span class="cc"># TypeError: unhashable type: 'dict'</span>
{{<span class="cn">1</span>, <span class="cn">2</span>}, {<span class="cn">3</span>, <span class="cn">4</span>}}          <span class="cc"># TypeError: unhashable type: 'set'</span>
{(<span class="cn">1</span>, [<span class="cn">2</span>, <span class="cn">3</span>])}             <span class="cc"># TypeError: unhashable type: 'list' (list inside tuple)</span></pre></div>

<div class="output-block">TypeError: unhashable type: 'list'</div>

<hr class="divider">

<div class="fix-block">
  <div class="fix-title"><span class="fix-badge">Fix 1</span> Convert lists → tuples</div>
  <div class="code-block"><pre>rows = [[<span class="cn">1</span>, <span class="cn">2</span>], [<span class="cn">3</span>, <span class="cn">4</span>], [<span class="cn">1</span>, <span class="cn">2</span>]]   <span class="cc"># list of lists — can't put in a set</span>

unique_rows = {<span class="cm">tuple</span>(row) <span class="ck">for</span> row <span class="ck">in</span> rows}
<span class="cc"># → {(1, 2), (3, 4)}  — deduped, now hashable</span></pre></div>
  <div class="step-box">Converting each inner list to a tuple makes it hashable. The set then deduplicates automatically. This is the canonical pattern for deduplicating a list of lists.</div>
</div>

<div class="fix-block">
  <div class="fix-title"><span class="fix-badge">Fix 2</span> Convert sets → frozensets</div>
  <div class="code-block"><pre>tag_groups = [
    {<span class="cs">"ml"</span>, <span class="cs">"python"</span>},
    {<span class="cs">"sql"</span>, <span class="cs">"data"</span>},
    {<span class="cs">"ml"</span>, <span class="cs">"python"</span>}    <span class="cc"># duplicate</span>
]

unique_groups = {<span class="cm">frozenset</span>(g) <span class="ck">for</span> g <span class="ck">in</span> tag_groups}
<span class="cc"># → {frozenset({'ml','python'}), frozenset({'sql','data'})}</span></pre></div>
  <div class="step-box">A <code>frozenset</code> is the immutable, hashable version of a set. Converting each set to a frozenset lets you store it inside another set — the set-of-sets pattern.</div>
</div>

<hr class="divider">

<div class="callback-box">
  <strong>Session 2B callback — the same rule, restated:</strong><br><br>
  In 2B: a tuple is only hashable if <strong>all its contents are hashable</strong> — one mutable element poisons the whole thing.<br><br>
  Here: a set element must be hashable for the same reason — the hash table needs a <strong>stable hash</strong> for every key it stores. Mutable objects can change after insertion, which would corrupt the table's invariants.<br><br>
  The fix is always the same: swap the mutable type for its immutable counterpart — <code>list → tuple</code>, <code>set → frozenset</code>.
</div>

<div class="note-box">
  💡 <strong>Quick decision rule:</strong> need to store a collection <em>inside</em> a set or use it as a dict key? Ask "is it immutable all the way down?" If not — convert. <code>list → tuple</code>, <code>set → frozenset</code>, and make sure nothing inside those contains a mutable either.
</div>

```python
# 5. First non-repeating element
# Given a sequence, return the first element that appears 
# exactly once. [2,3,2,4,3,5] → 4.
from collections import Counter
def first_unique(seq):
    counts = Counter(seq)          # count every element: {2:2, 3:2, 4:1, 5:1}
    for x in seq:                  # scan in ORIGINAL order
        if counts[x] == 1:
            return x
    return None
# [2,3,2,4,3,5] -> 4
```

a set alone can't solve this, for two reasons. A set can't count (it only knows present/absent, not how many times), and a set has no order (so even if it could, "first" is meaningless). You need a counter (dict-based) to count, then a scan of the original sequence to get order. This is precisely the seam where sets stop and dicts begin — which is why 2D (Dict) is next: a dict does what a set does plus associates a value (like a count) with each key, and preserves insertion order.
