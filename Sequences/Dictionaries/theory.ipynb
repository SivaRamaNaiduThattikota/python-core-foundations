<div class="part-header"><h2>Session 2D — Dict (Deep) &nbsp;·&nbsp; Part 1, Chunk A — Theory: The Model + Method Surface</h2></div>

<p class="body-p">The finale of Session 2, and the most important data structure in Python. The dict is how Python represents objects, namespaces, keyword arguments, JSON, and most real-world data. It ties together everything from Session 2: it's a hash table like a set (2C), its keys must be hashable like tuple/frozenset keys (2B), and its values can be anything including lists (2A).</p>

<hr class="divider">

<!-- SECTION 1: THEORY -->
<div class="part-header" style="border-left-color:#89b4fa"><h2 style="color:#89b4fa">1. Theory — Chunk A</h2></div>

<!-- 1.1 -->
<div class="part-header" style="border-left-color:#a6e3a1; margin-top:10px"><h2 style="color:#a6e3a1; font-size:1.1em;">1.1 What a dict is — a hash table mapping keys to values</h2></div>

<p class="body-p">A dict is an unordered-by-hash but insertion-ordered collection of key→value pairs, built on a hash table. If a set (2C) is a hash table of keys, a dict is a hash table of keys where each key also carries an associated value. Everything you learned about set performance applies to dict keys, plus you get a value attached.</p>

<p class="body-p"><strong style="color:#cba6f7">Three defining properties:</strong></p>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Property 1</span> O(1) key access</div>
  <p class="body-p" style="margin:0 0 8px 0"><code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">d[key]</code> computes <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">hash(key)</code>, jumps to the bucket, returns the value. Constant time regardless of dict size — same mechanism as set membership. The live proof was dramatic:</p>
  <div class="code-block"><pre>d = {i: i <span class="ck">for</span> i <span class="ck">in</span> <span class="cm">range</span>(<span class="cn">1_000_000</span>)}
d[<span class="cn">999999</span>]        <span class="cc"># 7.07 µs  - one hash lookup</span>
<span class="cc"># vs scanning a list of (key,value) pairs for the same key: 56 ms</span></pre></div>
  <div class="why-box">A ~7,900× speedup. <strong>SQL/Power BI anchor:</strong> a dict is a lookup table with a primary-key index. <code>d[key]</code> is an indexed seek; scanning a list of pairs is a full table scan. It's also exactly a DAX <code>LOOKUPVALUE</code> — fetch a value by its key.</div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Property 2</span> Keys must be hashable</div>
  <p class="body-p" style="margin:0 0 8px 0">Same rule as set elements (2C) and tuple dict-keys (2B): keys must be immutable-enough to have a stable hash. Strings, numbers, tuples-of-immutables, frozensets — fine. Lists, dicts, sets — not.</p>
  <div class="danger-box"><code>{[1,2]: "x"}</code> raises <code>unhashable type: 'list'</code>. Values, by contrast, can be anything — lists, other dicts, objects, whatever.</div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Property 3</span> Insertion-ordered (Python 3.7+)</div>
  <p class="body-p" style="margin:0 0 8px 0">Unlike sets, dicts preserve the order keys were inserted. Inserting <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">z, a, m, b</code> iterates back as <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">z, a, m, b</code>. This was an implementation detail in 3.6 and became a language guarantee in 3.7. It's why dicts can replace <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">OrderedDict</code> for most uses now, and why <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">dict.fromkeys</code> works as an order-preserving dedup (2A/2B/2C callback).</p>
</div>

<hr class="divider">

<!-- 1.2 -->
<div class="part-header" style="border-left-color:#a6e3a1; margin-top:10px"><h2 style="color:#a6e3a1; font-size:1.1em;">1.2 The complete method surface</h2></div>

<p class="body-p">Dicts have a rich method set. Here's every one that matters, with what it does, what it returns, and complexity. All verified.</p>

<div class="method-block">
  <div class="method-title"><span class="method-badge">Access methods</span></div>
  <div class="method-row"><code>d[key]</code> — get the value for key. O(1). Raises <code>KeyError</code> if missing.</div>
  <div class="method-row"><code>d.get(key, default=None)</code> — get the value, or default if missing. O(1). Never raises — the safe accessor. <code>d.get("z")</code> → <code>None</code>, <code>d.get("z", 0)</code> → <code>0</code>.</div>
  <div class="method-row"><code>d.setdefault(key, default)</code> — return <code>d[key]</code> if present; else set <code>d[key] = default</code> and return it. O(1). Sets <code>c</code> when new, leaves <code>a</code> unchanged when present. The workhorse for "get-or-initialize" (grouping patterns).</div>
</div>

<div class="method-block">
  <div class="method-title"><span class="method-badge">View methods</span> — return dynamic views, not lists</div>
  <div class="method-row"><code>d.keys()</code> — a view of keys.</div>
  <div class="method-row"><code>d.values()</code> — a view of values.</div>
  <div class="method-row"><code>d.items()</code> — a view of <code>(key, value)</code> tuples. All O(1) to create (they're lazy views over the live dict). The <code>(key, value)</code> tuples from <code>.items()</code> are the 2B tuples you unpack in loops.</div>
</div>

<div class="method-block">
  <div class="method-title"><span class="method-badge">Mutation methods</span></div>
  <div class="method-row"><code>d[key] = value</code> — insert or overwrite. O(1) amortized.</div>
  <div class="method-row"><code>d.update(other)</code> — merge another dict/iterable of pairs in; existing keys overwritten. O(k). <code>update({"d":4,"a":10})</code> added <code>d</code>, overwrote <code>a</code>.</div>
  <div class="method-row"><code>d.pop(key, default)</code> — remove and return the value; default avoids <code>KeyError</code>. O(1). <code>pop("d")</code> → <code>4</code>; <code>pop("z", "none")</code> → <code>"none"</code>.</div>
  <div class="method-row"><code>d.popitem()</code> — remove and return the last inserted <code>(key, value)</code> pair (LIFO, 3.7+). O(1). Returns <code>('c', 3)</code>.</div>
  <div class="method-row"><code>d.clear()</code> — empty the dict. O(n).</div>
</div>

<hr class="divider">

<!-- 1.3 -->
<div class="part-header" style="border-left-color:#f38ba8; margin-top:10px"><h2 style="color:#f38ba8; font-size:1.1em;">1.3 The single most important distinction: [] vs .get()</h2></div>

<p class="body-p">This is the dict equivalent of list's None-returning-methods trap — the thing that most separates fluent from beginner code:</p>

<div class="code-block"><pre>d = {<span class="cs">"a"</span>: <span class="cn">1</span>}
d[<span class="cs">"missing"</span>]          <span class="cc"># KeyError: 'missing'   - RAISES</span>
d.get(<span class="cs">"missing"</span>)      <span class="cc"># None                   - safe</span>
d.get(<span class="cs">"missing"</span>, <span class="cn">0</span>)   <span class="cc"># 0                       - safe with default</span></pre></div>

<div class="why-box">Use <code>d[key]</code> when the key must exist (you want to fail loudly if it doesn't). Use <code>.get()</code> when the key might not exist and you have a sensible fallback. Reaching for <code>d[key]</code> on uncertain data is the #1 source of <code>KeyError</code> crashes; <code>.get()</code> with a default is the fix. This choice — fail-loud vs. safe-default — is a real design decision you make constantly.</div>

<hr class="divider">

<!-- 1.4 -->
<div class="part-header" style="border-left-color:#cba6f7; margin-top:10px"><h2 style="color:#cba6f7; font-size:1.1em;">1.4 The setdefault pattern — get-or-initialize</h2></div>

<p class="body-p"><code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">setdefault</code> deserves special mention because it powers grouping — one of the most common real tasks:</p>

<div class="code-block"><pre>groups = {}
<span class="ck">for</span> word <span class="ck">in</span> words:
    groups.setdefault(word[<span class="cn">0</span>], []).append(word)   <span class="cc"># first letter -> list of words</span></pre></div>

<div class="step-box">For each word: <code>setdefault(key, [])</code> returns the existing list if the key's been seen, or creates a new empty list, stores it, and returns it — either way you get a list to <code>.append</code> to. This is the 2C anagram-grouping pattern from 2C, now with its proper tool. (Chunk B shows <code>defaultdict</code>, which makes this even cleaner.)</div>

<div class="part-header" style="border-left-color:#a6e3a1"><h2 style="color:#a6e3a1; font-size:1.1em;">1.5 Dict comprehensions — build, filter, invert</h2></div>

<p class="body-p">Like lists (2A) and sets (2C), dicts have a comprehension form. The syntax adds the key: value pair:</p>

<div class="code-block"><pre>{x: x*x <span class="ck">for</span> x <span class="ck">in</span> <span class="cm">range</span>(<span class="cn">5</span>)}                    <span class="cc"># {0:0, 1:1, 2:4, 3:9, 4:16}</span></pre></div>

<p class="body-p">Three common patterns, all verified:</p>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Build</span> Transform an iterable into a mapping</div>
  <div class="code-block"><pre>{x: x*x <span class="ck">for</span> x <span class="ck">in</span> <span class="cm">range</span>(<span class="cn">5</span>)}    <span class="cc"># number -> its square</span></pre></div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Filter</span> Keep only pairs matching a condition — iterate .items()</div>
  <div class="code-block"><pre>prices = {<span class="cs">"apple"</span>: <span class="cn">3</span>, <span class="cs">"banana"</span>: <span class="cn">1</span>, <span class="cs">"cherry"</span>: <span class="cn">5</span>}
{k: v <span class="ck">for</span> k, v <span class="ck">in</span> prices.items() <span class="ck">if</span> v > <span class="cn">2</span>}     <span class="cc"># {'apple':3, 'cherry':5}</span></pre></div>
  <div class="sql-box">This is <code>SELECT key, value FROM prices WHERE value > 2</code>.</div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Invert</span> Swap keys and values</div>
  <div class="code-block"><pre>{v: k <span class="ck">for</span> k, v <span class="ck">in</span> prices.items()}    <span class="cc"># {3:'apple', 1:'banana', 5:'cherry'}</span></pre></div>
  <div class="why-box">Inverting is common for reverse lookups (value→key), but note the catch: if values aren't unique, later pairs overwrite earlier ones (duplicate values collapse — since keys must be unique). Fine when values are unique (like an index map); lossy otherwise.</div>
</div>

<hr class="divider">

<!-- 1.6 -->
<div class="part-header" style="border-left-color:#a6e3a1"><h2 style="color:#a6e3a1; font-size:1.1em;">1.6 Iteration — keys by default, items for pairs</h2></div>

<p class="body-p">The one thing that trips people up: iterating a dict directly gives keys, not pairs.</p>

<div class="code-block"><pre>d = {<span class="cs">"a"</span>: <span class="cn">1</span>, <span class="cs">"b"</span>: <span class="cn">2</span>}
<span class="ck">for</span> k <span class="ck">in</span> d:              <span class="cc"># iterates KEYS -> 'a', 'b'</span>
    ...
<span class="ck">for</span> k <span class="ck">in</span> d.keys():      <span class="cc"># same thing, explicit</span>
    ...
<span class="ck">for</span> v <span class="ck">in</span> d.values():    <span class="cc"># iterates VALUES -> 1, 2</span>
    ...
<span class="ck">for</span> k, v <span class="ck">in</span> d.items():  <span class="cc"># iterates (key, value) TUPLES -> ('a',1), ('b',2)</span>
    ...</pre></div>

<div class="why-box">The rule: <code>for x in d</code> gives keys. When you want both key and value, use <code>.items()</code> and unpack the tuple in the loop header (<code>for k, v in d.items()</code>) — this is the 2B tuple-unpacking pattern, and it's the most common dict loop in real code. A frequent beginner bug is <code>for k, v in d:</code> (without <code>.items()</code>), which tries to unpack each key and fails.</div>

<hr class="divider">

<!-- 1.7 -->
<div class="part-header" style="border-left-color:#a6e3a1"><h2 style="color:#a6e3a1; font-size:1.1em;">1.7 The collections power tools — Counter and defaultdict</h2></div>

<p class="body-p">These two are specialized dicts that eliminate boilerplate you'd otherwise write constantly. They're heavily used in ML data prep and interviews.</p>

<div class="method-block">
  <div class="method-title"><span class="method-badge">Counter</span> Counting made trivial — a dict subclass that counts hashable items</div>
  <div class="code-block"><pre><span class="ck">from</span> collections <span class="ck">import</span> Counter
c = <span class="cm">Counter</span>(<span class="cs">"mississippi"</span>)        <span class="cc"># Counter({'i':4, 's':4, 'p':2, 'm':1})</span>
c.most_common(<span class="cn">2</span>)                  <span class="cc"># [('i',4), ('s',4)] - top 2 by count</span>
<span class="cm">Counter</span>(<span class="cs">"aab"</span>) + <span class="cm">Counter</span>(<span class="cs">"abb"</span>)   <span class="cc"># Counter({'a':3, 'b':3}) - counters add!</span></pre></div>
  <div class="why-box">Counter gives you frequency counting in one line (vs. a manual <code>setdefault</code>/<code>get</code> loop), <code>most_common(n)</code> for top-n (already sorted by count), and arithmetic (<code>+</code>, <code>-</code>, <code>&</code>, <code>|</code>) between counters. This is the tool for word frequencies, label distributions, vote tallies. Directly relevant to OpsRAG — token/tag frequency analysis is a one-liner.</div>
</div>

<div class="method-block">
  <div class="method-title"><span class="method-badge">defaultdict</span> Grouping without the setdefault dance — auto-creates default values for missing keys</div>
  <div class="code-block"><pre><span class="ck">from</span> collections <span class="ck">import</span> defaultdict

groups = <span class="cm">defaultdict</span>(list)                    <span class="cc"># missing key -> new empty list</span>
<span class="ck">for</span> word <span class="ck">in</span> [<span class="cs">"apple"</span>,<span class="cs">"ant"</span>,<span class="cs">"banana"</span>,<span class="cs">"bear"</span>]:
    groups[word[<span class="cn">0</span>]].append(word)              <span class="cc"># no setdefault needed!</span>
<span class="cc"># {'a':['apple','ant'], 'b':['banana','bear']}</span>

counts = <span class="cm">defaultdict</span>(int)                     <span class="cc"># missing key -> 0</span>
<span class="ck">for</span> ch <span class="ck">in</span> <span class="cs">"hello"</span>:
    counts[ch] += <span class="cn">1</span>                           <span class="cc"># no get(ch,0) needed</span>
<span class="cc"># {'h':1, 'e':1, 'l':2, 'o':1}</span></pre></div>
  <div class="why-box">You give <code>defaultdict</code> a factory (<code>list</code>, <code>int</code>, <code>set</code>, etc.); when you access a missing key, it calls the factory to create the default automatically. <code>defaultdict(list)</code> for grouping, <code>defaultdict(int)</code> for counting (though <code>Counter</code> is cleaner for pure counting). This turns the <code>setdefault</code> grouping pattern from Chunk A into cleaner code — compare <code>groups.setdefault(k, []).append(x)</code> vs <code>groups[k].append(x)</code>.</div>
</div>

<hr class="divider">

<!-- 1.8 -->
<div class="part-header" style="border-left-color:#a6e3a1"><h2 style="color:#a6e3a1; font-size:1.1em;">1.8 Dict merging (Python 3.9+)</h2></div>

<p class="body-p">Modern syntax for combining dicts:</p>

<div class="code-block"><pre>a = {<span class="cs">"x"</span>: <span class="cn">1</span>, <span class="cs">"y"</span>: <span class="cn">2</span>}
b = {<span class="cs">"y"</span>: <span class="cn">3</span>, <span class="cs">"z"</span>: <span class="cn">4</span>}
a | b        <span class="cc"># {'x':1, 'y':3, 'z':4}  - b wins on conflict</span></pre></div>

<div class="why-box"><code>y</code> takes <code>b</code>'s value (<code>3</code>), because the right operand wins on key conflicts. Also <code>a |= b</code> to merge in place, and the older <code>{**a, **b}</code> unpacking form (works pre-3.9). This is the fix for the mutable-default config bug from 2A — merge a user config over defaults: <code>config = DEFAULTS | user_overrides</code>.</div>

<hr class="divider">

<!-- 1.9 -->
<div class="part-header" style="border-left-color:#a6e3a1"><h2 style="color:#a6e3a1; font-size:1.1em;">1.9 Key views are set-like</h2></div>

<p class="body-p">A subtle but useful feature: <code style="background:#313244;padding:2px 6px;border-radius:4px;color:#a6e3a1">d.keys()</code> returns a view that supports set operations (because keys are unique and hashable, just like a set):</p>

<div class="code-block"><pre>d = {<span class="cs">"a"</span>: <span class="cn">1</span>, <span class="cs">"b"</span>: <span class="cn">2</span>, <span class="cs">"c"</span>: <span class="cn">3</span>}
d.keys() & {<span class="cs">"a"</span>, <span class="cs">"c"</span>, <span class="cs">"z"</span>}    <span class="cc"># {'a', 'c'} - intersection with a set!</span></pre></div>

<div class="why-box">You can do <code>&</code>, <code>|</code>, <code>-</code>, <code>^</code> directly on <code>.keys()</code> — "which keys do these two dicts share?" is <code>d1.keys() & d2.keys()</code>. This is 2C's set algebra reaching into dicts. Views are also dynamic — they reflect live changes to the dict, and are O(1) to create since they don't copy.</div>

<div class="ex-header"><h2>2. Example</h2></div>

<div class="ex-block">
  <div class="ex-title"><span class="ex-badge">Example 1</span> Word frequency, three ways (the counting task)</div>
  <div class="code-block"><pre>words = <span class="cs">"the cat the dog the cat"</span>.<span class="cm">split</span>()

<span class="cc"># 1. manual with .get(key, 0) - the fundamental pattern</span>
f1 = {}
<span class="ck">for</span> w <span class="ck">in</span> words:
    f1[w] = f1.<span class="cm">get</span>(w, <span class="cn">0</span>) + <span class="cn">1</span>          <span class="cc"># {'the':3, 'cat':2, 'dog':1}</span>

<span class="cc"># 2. defaultdict(int) - no .get needed</span>
<span class="ck">from</span> collections <span class="ck">import</span> defaultdict, Counter
f2 = <span class="cm">defaultdict</span>(<span class="cm">int</span>)
<span class="ck">for</span> w <span class="ck">in</span> words:
    f2[w] += <span class="cn">1</span>                        <span class="cc"># missing key auto-starts at 0</span>

<span class="cc"># 3. Counter - one line</span>
<span class="cm">Counter</span>(words)                        <span class="cc"># Counter({'the':3, 'cat':2, 'dog':1})</span></pre></div>
  <div class="step-box">
    All three identical. This progression is the dict learning arc: <code>.get(k, 0)</code> teaches the mechanism, <code>defaultdict(int)</code> removes boilerplate, <code>Counter</code> is the production one-liner. Know all three — interviewers sometimes ask you to do it without <code>Counter</code> to see if you understand the mechanism.
  </div>
</div>

<div class="ex-block">
  <div class="ex-title"><span class="ex-badge">Example 2</span> Grouping records (the setdefault/defaultdict task)</div>
  <div class="code-block"><pre>people = [(<span class="cs">"eng"</span>,<span class="cs">"Alice"</span>), (<span class="cs">"sales"</span>,<span class="cs">"Bob"</span>), (<span class="cs">"eng"</span>,<span class="cs">"Cara"</span>), (<span class="cs">"sales"</span>,<span class="cs">"Dan"</span>)]
groups = <span class="cm">defaultdict</span>(<span class="cm">list</span>)
<span class="ck">for</span> dept, name <span class="ck">in</span> people:            <span class="cc"># unpack each tuple (2B)</span>
    groups[dept].<span class="cm">append</span>(name)
<span class="cc"># {'eng': ['Alice','Cara'], 'sales': ['Bob','Dan']}</span></pre></div>
  <div class="step-box">
    Group items by a key into lists — <code>defaultdict(list)</code> auto-creates the empty list. This is the C7 anagram pattern from 2C, now with its natural tool. SQL anchor: <code>GROUP BY dept</code>. One of the most common real data tasks, period.
  </div>
</div>

<div class="ex-block">
  <div class="ex-title"><span class="ex-badge">Example 3</span> Nested dicts (JSON-shaped data)</div>
  <div class="code-block"><pre>users = {
    <span class="cs">"u1"</span>: {<span class="cs">"name"</span>: <span class="cs">"Alice"</span>, <span class="cs">"roles"</span>: [<span class="cs">"admin"</span>, <span class="cs">"dev"</span>]},
    <span class="cs">"u2"</span>: {<span class="cs">"name"</span>: <span class="cs">"Bob"</span>,   <span class="cs">"roles"</span>: [<span class="cs">"dev"</span>]},
}
users[<span class="cs">"u1"</span>][<span class="cs">"name"</span>]                        <span class="cc"># "Alice"  - chained access</span>
users[<span class="cs">"u1"</span>][<span class="cs">"roles"</span>][<span class="cn">0</span>]                    <span class="cc"># "admin"  - dict->list->index</span>
users.<span class="cm">get</span>(<span class="cs">"u3"</span>, {}).<span class="cm">get</span>(<span class="cs">"name"</span>, <span class="cs">"unknown"</span>) <span class="cc"># "unknown" - safe nested access</span></pre></div>
  <div class="step-box">
    This is the shape of essentially all API/JSON data — dicts nested in dicts, with lists inside. The safe-nested pattern <code>d.get("k", {}).get("inner", default)</code> avoids <code>KeyError</code> when a level might be missing — critical for real-world data where fields are optional. Values being "anything" (2D's rule) is what makes this work: a value can be a dict, a list, or a list of dicts.
  </div>
</div>

<div class="ex-block">
  <div class="ex-title"><span class="ex-badge">Example 4</span> Index map and inversion (the ML vocabulary task)</div>
  <div class="code-block"><pre>vocab = [<span class="cs">"cat"</span>, <span class="cs">"dog"</span>, <span class="cs">"bird"</span>]
word_to_idx = {w: i <span class="ck">for</span> i, w <span class="ck">in</span> <span class="cm">enumerate</span>(vocab)}   <span class="cc"># {'cat':0,'dog':1,'bird':2}</span>
idx_to_word = {i: w <span class="ck">for</span> w, i <span class="ck">in</span> word_to_idx.<span class="cm">items</span>()} <span class="cc"># {0:'cat',1:'dog',2:'bird'}</span></pre></div>
  <div class="step-box">
    Building a token→integer map (and its inverse) is step one of every embedding/NLP pipeline — the exact continuation of 2C's vocabulary-building example. <code>enumerate</code> supplies the indices, the comprehension builds the map, and inverting via <code>.items()</code> gives you decode-back capability.
  </div>
</div>

<div class="ex-block">
  <div class="ex-title"><span class="ex-badge">Example 5</span> Merging configs (the reproducibility fix)</div>
  <div class="code-block"><pre>DEFAULTS = {<span class="cs">"lr"</span>: <span class="cn">0.01</span>, <span class="cs">"epochs"</span>: <span class="cn">100</span>, <span class="cs">"batch"</span>: <span class="cn">32</span>}
user     = {<span class="cs">"lr"</span>: <span class="cn">0.001</span>, <span class="cs">"batch"</span>: <span class="cn">64</span>}
config = DEFAULTS | user     <span class="cc"># {'lr':0.001, 'epochs':100, 'batch':64}</span></pre></div>
  <div class="step-box">
    User values override defaults, <code>epochs</code> falls back to the default. This is the clean, immutable-friendly way to layer configuration, and it structurally avoids the mutable-shared-config bug from 2A's ML section.
  </div>

<!-- ══ SECTION 3: EDGE CASES ══ -->
<div class="warn-header"><h2>3. Edge Cases</h2></div>
<p style="color:#cdd6f4;font-family:'Segoe UI',sans-serif;font-size:0.93em;line-height:1.7;margin:0 0 14px 0">Seven cases, all verified.</p>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 1</span> d[key] raises, .get() is safe</div>
  <div class="code-block"><pre>d = {<span class="cs">"a"</span>: <span class="cn">1</span>}
d[<span class="cs">"b"</span>]              <span class="cc"># KeyError: 'b'</span>
d.<span class="cm">get</span>(<span class="cs">"b"</span>, <span class="cn">0</span>)       <span class="cc"># 0</span></pre></div>
  <div class="why-box">
    <strong>Why it matters:</strong> the dict's headline trap. <code>d[key]</code> on a missing key crashes; <code>.get(key, default)</code> returns the default. Use <code>[]</code> when the key must exist (fail loud), <code>.get()</code> when it might not (safe fallback). This single choice prevents most dict-related crashes.
  </div>
</div>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 2</span> for k, v in d: fails without .items()</div>
  <div class="code-block"><pre>d = {<span class="cs">"a"</span>: <span class="cn">1</span>, <span class="cs">"b"</span>: <span class="cn">2</span>}
<span class="ck">for</span> k, v <span class="ck">in</span> d:      <span class="cc"># ValueError: not enough values to unpack (expected 2, got 1)</span>
    ...</pre></div>
  <div class="why-box">
    <strong>Why it matters:</strong> iterating a dict yields keys, not pairs. Trying to unpack each key (a single string) into two variables fails. You must use <code>.items()</code>: <code>for k, v in d.items()</code>. This is a top beginner bug, and the error message ("expected 2, got 1") is the tell.
  </div>
</div>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 3</span> Duplicate keys collapse (last wins)</div>
  <div class="code-block"><pre>{<span class="cs">"a"</span>: <span class="cn">1</span>, <span class="cs">"a"</span>: <span class="cn">2</span>, <span class="cs">"a"</span>: <span class="cn">3</span>}          <span class="cc"># {'a': 3}</span>
<span class="cm">dict</span>([(<span class="cs">"x"</span>, <span class="cn">1</span>), (<span class="cs">"x"</span>, <span class="cn">2</span>)])        <span class="cc"># {'x': 2}</span></pre></div>
  <div class="why-box">
    <strong>Why it matters:</strong> keys are unique — a later assignment to the same key overwrites, silently. This bites when building a dict from pairs that have duplicate keys (you lose data without warning), and it's why inverting a dict with non-unique values is lossy (Chunk B).
  </div>
</div>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 4</span> Mutable default value aliasing (the 2A trap, in dict form)</div>
  <div class="code-block"><pre>d = <span class="cm">dict</span>.<span class="cm">fromkeys</span>([<span class="cs">"a"</span>, <span class="cs">"b"</span>], [])   <span class="cc"># BOTH keys share ONE list!</span>
d[<span class="cs">"a"</span>].<span class="cm">append</span>(<span class="cn">1</span>)
<span class="cm">print</span>(d)                            <span class="cc"># {'a': [1], 'b': [1]}  -- b changed too!</span>

d2 = {k: [] <span class="ck">for</span> k <span class="ck">in</span> [<span class="cs">"a"</span>, <span class="cs">"b"</span>]}    <span class="cc"># comprehension -> distinct lists</span>
d2[<span class="cs">"a"</span>].<span class="cm">append</span>(<span class="cn">1</span>)
<span class="cm">print</span>(d2)                           <span class="cc"># {'a': [1], 'b': []}   -- correct</span></pre></div>
  <div class="why-box">
    <strong>Why it matters:</strong> <code>dict.fromkeys(keys, [])</code> evaluates <code>[]</code> once and gives every key the same list object (Session 1 aliasing, exactly like <code>[[0]*3]*3</code>). Appending to one appends to all. Fix: a comprehension <code>{k: [] for k in keys}</code> creates a fresh list per key. This is a genuinely nasty bug because <code>fromkeys</code> looks like the obvious tool.
  </div>
</div>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 5</span> Modifying a dict during iteration raises</div>
  <div class="code-block"><pre>d = {<span class="cs">"a"</span>: <span class="cn">1</span>, <span class="cs">"b"</span>: <span class="cn">2</span>, <span class="cs">"c"</span>: <span class="cn">3</span>}
<span class="ck">for</span> k <span class="ck">in</span> d:
    <span class="ck">if</span> d[k] == <span class="cn">2</span>:
        <span class="ck">del</span> d[k]          <span class="cc"># RuntimeError: dictionary changed size during iteration</span></pre></div>
  <div class="why-box">
    <strong>Why it matters:</strong> like sets (2C Trap 3), dicts raise <code>RuntimeError</code> if you change their size mid-iteration — stricter than lists (which fail silently). Fix: build a new dict with a comprehension (<code>{k:v for k,v in d.items() if v != 2}</code>) or iterate over a copy of the keys (<code>for k in list(d):</code>).
  </div>
</div>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 6</span> 1 and 1.0 are the same key; membership checks keys</div>
  <div class="code-block"><pre>d = {<span class="cn">1</span>: <span class="cs">"a"</span>, <span class="cs">"1"</span>: <span class="cs">"b"</span>, <span class="cn">1.0</span>: <span class="cs">"c"</span>}
<span class="cc"># -> {1: 'c', '1': 'b'}   -- 1 and 1.0 collapsed (1 == 1.0, same hash)</span>
<span class="cs">"a"</span> <span class="ck">in</span> d                  <span class="cc"># False -- 'in' checks KEYS, not values</span></pre></div>
  <div class="why-box">
    <strong>Why it matters:</strong> two subtleties. First, <code>1</code> and <code>1.0</code> are equal and hash identically, so they're the same key — <code>1.0: "c"</code> overwrote <code>1: "a"</code>. (The string <code>"1"</code> is distinct.) Second, <code>x in d</code> tests keys, not values — checking for a value requires <code>x in d.values()</code> (O(n)).
  </div>
</div>

<div class="edge-block">
  <div class="edge-title"><span class="edge-badge">Edge 7</span> .get() with a default doesn't store; setdefault does</div>
  <div class="code-block"><pre>d = {}
d.<span class="cm">get</span>(<span class="cs">"k"</span>, [])            <span class="cc"># returns [] but d is STILL {}</span>
d.<span class="cm">setdefault</span>(<span class="cs">"k"</span>, []).<span class="cm">append</span>(<span class="cn">1</span>)   <span class="cc"># stores [] AND returns it -> d = {'k': [1]}</span></pre></div>
  <div class="why-box">
    <strong>Why it matters:</strong> a subtle distinction that trips people building nested structures. <code>.get(k, default)</code> returns the default but doesn't insert it — the dict is unchanged. <code>.setdefault(k, default)</code> inserts the default (if missing) and returns it, so you can immediately mutate what it returns. For "get-or-create-then-modify," you need <code>setdefault</code> (or <code>defaultdict</code>), not <code>.get()</code>.
  </div>
</div>

<!-- ══ SECTION 4: GOLDEN RULES ══ -->
<div class="part-header"><h2>4. Golden Rules</h2></div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 1</span> The <code>.get()</code> vs <code>[]</code> decision — fail-loud or safe-default (full depth)</div>
  <div class="rule-body">
    This is the dict decision you make most often, so it deserves proper treatment rather than a one-liner. The choice encodes intent about whether a key is expected to exist, and getting it right is a real signal of fluency.<br><br>
    <strong style="color:#89b4fa">Use <code>d[key]</code> when the key must exist.</strong> If a missing key means something is genuinely wrong — a bug, corrupt data, a broken contract — you want the <code>KeyError</code>. Failing loudly at the point of the problem is better than silently continuing with wrong data. Example: looking up a required config value you know you set. If it's missing, crash now with a clear error, don't paper over it.<br><br>
    <strong style="color:#a6e3a1">Use <code>d.get(key, default)</code> when the key might legitimately be absent.</strong> If absence is a normal, expected case with a sensible fallback, <code>.get()</code> handles it cleanly. Example: reading an optional field from API data, or counting with <code>d.get(k, 0) + 1</code> where a new key naturally starts at zero.<br><br>
    The anti-pattern that reveals inexperience is the <code>try/except KeyError</code> wrapper around <code>d[key]</code> when you just want a default — that's what <code>.get()</code> is for:
  </div>
  <div class="code-block"><pre><span class="cc"># clumsy</span>
<span class="ck">try</span>:
    x = d[key]
<span class="ck">except</span> <span class="cm">KeyError</span>:
    x = <span class="cn">0</span>
<span class="cc"># fluent</span>
x = d.<span class="cm">get</span>(key, <span class="cn">0</span>)</pre></div>
  <div class="rule-body">
    And the deeper rule: don't check-then-access when one call does it. <code>if key in d: x = d[key]</code> is two hash lookups; <code>d.get(key)</code> is one. For get-or-create, <code>setdefault</code>/<code>defaultdict</code> beats <code>if key not in d: d[key] = []</code>.
  </div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 2</span> Reach for <code>Counter</code> to count, <code>defaultdict</code> to group.</div>
  <div class="rule-body">Manual <code>.get(k,0)+1</code> loops and <code>setdefault(k,[]).append</code> work and are worth understanding, but the collections tools are cleaner and signal you know the stdlib. Use <code>Counter(items)</code> for frequencies, <code>defaultdict(list)</code> for grouping.</div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 3</span> Never use a mutable default with <code>dict.fromkeys</code>.</div>
  <div class="rule-body"><code>dict.fromkeys(keys, [])</code> shares one list across all keys (aliasing). Use a comprehension <code>{k: [] for k in keys}</code> for distinct values.</div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 4</span> Use <code>.items()</code> to iterate key+value; plain iteration gives keys only.</div>
  <div class="rule-body"><code>for k, v in d.items()</code>, not <code>for k, v in d</code>.</div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 5</span> Merge with <code>|</code> (3.9+) or <code>{**a, **b}</code>; right side wins.</div>
  <div class="rule-body">Clean way to layer configs over defaults.</div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 6</span> Remember membership and <code>in</code> check keys, not values.</div>
  <div class="rule-body"><code>k in d</code> is O(1) on keys; <code>v in d.values()</code> is O(n). Need fast value lookups? Build an inverted dict.</div>
</div>

<div class="rule-block">
  <div class="rule-title"><span class="rule-badge">Rule 7</span> To modify while iterating, iterate a copy or build a new dict.</div>
  <div class="rule-body"><code>for k in list(d):</code> or a comprehension. Modifying size mid-iteration raises <code>RuntimeError</code>.</div>
</div>

<!-- ══ SECTION 5: COMMON TRAPS ══ -->
<div class="trap-header"><h2>5. Common Traps</h2></div>

<div class="trap-block">
  <div class="trap-title"><span class="trap-badge">Trap 1</span> KeyError on <code>d[key]</code>. (the classic)</div>
  <div class="code-block"><pre>value = d[user_input]     <span class="cc"># crashes if user_input isn't a key</span></pre></div>
  <div class="why-box">The most common dict crash. The tell: <code>KeyError</code> in a traceback. Fix: <code>.get(key, default)</code> when absence is possible, or validate with <code>if key in d</code> first when you need to branch.</div>
</div>

<div class="trap-block">
  <div class="trap-title"><span class="trap-badge">Trap 2</span> <code>for k, v in d:</code> instead of <code>d.items()</code>. (tight)</div>
  <div class="code-block"><pre><span class="ck">for</span> k, v <span class="ck">in</span> d:            <span class="cc"># ValueError: not enough values to unpack</span></pre></div>
  <div class="why-box">Iterating a dict yields keys, not pairs. Fix: <code>for k, v in d.items()</code>. The tell: "not enough values to unpack (expected 2, got 1)."</div>
</div>

<div class="trap-block">
  <div class="trap-title"><span class="trap-badge">Trap 3</span> <code>dict.fromkeys</code> with a mutable default shares one object. (full depth — it's the dict version of a Session 1 trap)</div>
  <div class="code-block"><pre>d = <span class="cm">dict</span>.<span class="cm">fromkeys</span>([<span class="cs">"a"</span>, <span class="cs">"b"</span>, <span class="cs">"c"</span>], [])
d[<span class="cs">"a"</span>].<span class="cm">append</span>(<span class="cn">1</span>)
<span class="cm">print</span>(d)     <span class="cc"># {'a':[1], 'b':[1], 'c':[1]}  -- ALL changed!</span></pre></div>
  <div class="why-box"><code>fromkeys</code> evaluates the default <code>[]</code> once and points every key at that single list object — the exact aliasing from <code>[[0]*3]*3</code> (2A Edge 6) and the mutable-default-argument trap (Session 1). Append through one key, all keys see it, because they're one list. The tell: setting a value on one key mysteriously changes others. Fix: comprehension <code>{k: [] for k in keys}</code> — gives distinct lists. This is nasty precisely because <code>fromkeys</code> looks like the right tool for "initialize these keys to empty lists."</div>
</div>

<div class="trap-block">
  <div class="trap-title"><span class="trap-badge">Trap 4</span> Silent key overwrite from duplicates. (tight)</div>
  <div class="code-block"><pre><span class="cm">dict</span>([(<span class="cs">"x"</span>, <span class="cn">1</span>), (<span class="cs">"x"</span>, <span class="cn">2</span>)])    <span class="cc"># {'x': 2}  -- first value lost silently</span></pre></div>
  <div class="why-box">Building a dict from pairs with duplicate keys keeps only the last — no warning, data lost. The tell: a dict smaller than the input, or missing values you expected. Fix: if you need to keep all values per key, group into lists with <code>defaultdict(list)</code> instead.</div>
</div>

<div class="trap-block">
  <div class="trap-title"><span class="trap-badge">Trap 5</span> Modifying during iteration. (tight)</div>
  <div class="code-block"><pre><span class="ck">for</span> k <span class="ck">in</span> d:
    <span class="ck">if</span> <span class="cm">cond</span>(d[k]): <span class="ck">del</span> d[k]    <span class="cc"># RuntimeError: dictionary changed size</span></pre></div>
  <div class="why-box">Stricter than lists (raises rather than misbehaving silently). Fix: <code>for k in list(d):</code> or build a new dict with a comprehension.</div>
</div>

<div class="trap-block">
  <div class="trap-title"><span class="trap-badge">Trap 6</span> Assuming <code>.get()</code> inserts the default. (tight)</div>
  <div class="code-block"><pre>d.<span class="cm">get</span>(<span class="cs">"k"</span>, []).<span class="cm">append</span>(<span class="cn">1</span>)      <span class="cc"># appends to a throwaway list; d unchanged!</span></pre></div>
  <div class="why-box"><code>.get()</code> returns the default but doesn't store it — the append is lost. For get-or-create-then-mutate, use <code>setdefault</code> or <code>defaultdict</code>. The tell: a nested structure you thought you were building stays empty.</div>
</div>

<div class="ml-header"><h2>7. ML Real-World Connection</h2></div>

<p class="ml-body">The dict is the backbone of ML code — more than any other structure. Features, configs, vocabularies, class distributions, and caches are all dicts.</p>

<div class="ml-block">
  <div class="ml-block-title"><span class="ml-badge">ML 1</span> A feature vector / data row is a dict</div>
  <p class="ml-body">Before data becomes a NumPy array, it's usually a dict of named features:</p>
  <div class="code-block"><pre>sample = {<span class="cs">"age"</span>: <span class="cn">34</span>, <span class="cs">"income"</span>: <span class="cn">55000</span>, <span class="cs">"region"</span>: <span class="cs">"south"</span>, <span class="cs">"score"</span>: <span class="cn">0.87</span>}
sample[<span class="cs">"age"</span>]                    <span class="cc"># 34</span>
sample.<span class="cm">get</span>(<span class="cs">"missing_feat"</span>, <span class="cn">0</span>)    <span class="cc"># 0 - safe default for absent features</span></pre></div>
  <p class="ml-body">This is how a single record looks coming out of a database, an API, or a pandas row (<code>.to_dict()</code>). The <code>.get(feat, default)</code> pattern handles missing features gracefully — critical when different records have different available fields.</p>
</div>

<div class="ml-block">
  <div class="ml-block-title"><span class="ml-badge">ML 2</span> Class distribution with Counter — the first thing you check before training</div>
  <div class="code-block"><pre>labels = [<span class="cs">"spam"</span>,<span class="cs">"ham"</span>,<span class="cs">"spam"</span>,<span class="cs">"spam"</span>,<span class="cs">"ham"</span>,<span class="cs">"spam"</span>]
dist = <span class="cm">Counter</span>(labels)                          <span class="cc"># {'spam':4, 'ham':2}</span>
total = <span class="cm">sum</span>(dist.<span class="cm">values</span>())
balance = {k: <span class="cm">round</span>(v/total, <span class="cn">2</span>) <span class="ck">for</span> k, v <span class="ck">in</span> dist.<span class="cm">items</span>()}   <span class="cc"># {'spam':0.67, 'ham':0.33}</span></pre></div>
  <p class="ml-body">Checking class balance is step zero of any classification task — a 67/33 split tells you whether you need class weighting, resampling, or stratified splitting. <code>Counter</code> + a comprehension gives you the distribution in two lines. Directly relevant to any classifier you build.</p>
</div>

<div class="ml-block">
  <div class="ml-block-title"><span class="ml-badge">ML 3</span> Vocabulary encoding — the tokenizer core</div>
  <div class="code-block"><pre>vocab = {}
<span class="ck">for</span> doc <span class="ck">in</span> corpus:
    <span class="ck">for</span> tok <span class="ck">in</span> doc.<span class="cm">split</span>():
        <span class="ck">if</span> tok <span class="ck">not in</span> vocab:
            vocab[tok] = <span class="cm">len</span>(vocab)      <span class="cc"># next token gets the next integer id</span>
<span class="ck">def</span> <span class="cm">encode</span>(doc):
    <span class="ck">return</span> [vocab[t] <span class="ck">for</span> t <span class="ck">in</span> doc.<span class="cm">split</span>()]   <span class="cc"># "deploy docker" -> [0, 1]</span></pre></div>
  <p class="ml-body">The <code>vocab[tok] = len(vocab)</code> trick assigns sequential integer IDs as new tokens appear — the exact mechanism inside every tokenizer and embedding layer. This is 2C's vocabulary-building and 2D's index-map examples combined into the real thing. Central to OpsRAG's retrieval side.</p>
</div>

<div class="ml-block">
  <div class="ml-block-title"><span class="ml-badge">ML 4</span> Memoization — caching with a dict</div>
  <div class="code-block"><pre>cache = {}
<span class="ck">def</span> <span class="cm">fib</span>(n):
    <span class="ck">if</span> n < <span class="cn">2</span>: <span class="ck">return</span> n
    <span class="ck">if</span> n <span class="ck">in</span> cache: <span class="ck">return</span> cache[n]           <span class="cc"># O(1) cache hit</span>
    cache[n] = <span class="cm">fib</span>(n-<span class="cn">1</span>) + <span class="cm">fib</span>(n-<span class="cn">2</span>)

  <p class="ml-body">A dict cache turns exponential recomputation into linear — the essence of dynamic programming and of <code>functools.lru_cache</code>. In ML this is how you cache expensive embeddings, API responses, or feature computations. The O(1) key lookup is what makes it worthwhile.</p>
</div>

  <div class="ml-block-title"><span class="ml-badge">ML 5</span> Config as a dict, merged over defaults</div>

<div class="interview-header"><h2>8. Interview Questions</h2></div>
<div class="sub-header"><h3>8a — Conceptual Q&A (12 questions)</h3></div>

<div class="qa-block"><div class="qa-q"><span class="q-num">1</span> How is a dict implemented, and why is key access O(1)?</div>
  <div class="qa-a"><span class="ans-label">A</span> As a hash table. <code>d[k]</code> computes <code>hash(k)</code>, jumps to the bucket, returns the value — no scan. Average O(1) regardless of size.</div></div>

<div class="qa-block"><div class="qa-q"><span class="q-num">2</span> What can be a dict key? A value?</div>
  <div class="qa-a"><span class="ans-label">A</span> Keys must be hashable (immutable-enough): strings, numbers, tuples-of-immutables, frozensets. Values can be anything, including lists, dicts, and objects.</div></div>

<div class="qa-block"><div class="qa-q"><span class="q-num">3</span> <code>d[key]</code> vs <code>d.get(key)</code>?</div>
  <div class="qa-a"><span class="ans-label">A</span> <code>d[key]</code> raises <code>KeyError</code> if missing; <code>.get(key, default)</code> returns the default (or <code>None</code>) and never raises. Use <code>[]</code> when the key must exist, <code>.get()</code> when absence is expected.</div></div>

<div class="qa-block"><div class="qa-q"><span class="q-num">4</span> Do dicts preserve order?</div>
  <div class="qa-a"><span class="ans-label">A</span> Yes, insertion order, guaranteed since Python 3.7 (implementation detail in 3.6). Unlike sets, which are unordered.</div></div>

<div class="qa-block"><div class="qa-q"><span class="q-num">5</span> What does iterating a dict give you?</div>
  <div class="qa-a"><span class="ans-label">A</span> Keys by default (<code>for k in d</code>). Use <code>.items()</code> for (key, value) pairs, <code>.values()</code> for values.</div></div>

<div class="qa-block"><div class="qa-q"><span class="q-num">6</span> Counter vs defaultdict?</div>
  <div class="qa-a"><span class="ans-label">A</span> <code>Counter</code> is a dict specialized for counting (<code>most_common</code>, arithmetic). <code>defaultdict(factory)</code> auto-creates a default for missing keys — <code>defaultdict(list)</code> for grouping, <code>defaultdict(int)</code> for counting. Use <code>Counter</code> for pure counts, <code>defaultdict</code> for grouping.</div></div>

<div class="qa-block"><div class="qa-q"><span class="q-num">7</span> What's wrong with <code>dict.fromkeys(keys, [])</code>?</div>
  <div class="qa-a"><span class="ans-label">A</span> All keys share one list object (the default is evaluated once). Mutating one mutates all. Use <code>{k: [] for k in keys}</code> for distinct lists.</div></div>

<div class="qa-block"><div class="qa-q"><span class="q-num">8</span> Does <code>x in d</code> check keys or values? Complexity of each?</div>
  <div class="qa-a"><span class="ans-label">A</span> Keys, O(1). Checking values (<code>x in d.values()</code>) is O(n) — values aren't hash-indexed.</div></div>

<div class="qa-block"><div class="qa-q"><span class="q-num">9</span> How do you invert a dict with non-unique values?</div>
  <div class="qa-a"><span class="ans-label">A</span> Not with a plain comprehension (last-wins loses data). Use <code>defaultdict(list)</code> and append each key to its value's list.</div></div>

<div class="qa-block"><div class="qa-q"><span class="q-num">10</span> What happens with duplicate keys when building a dict?</div>
  <div class="qa-a"><span class="ans-label">A</span> The last value wins, silently. <code>{"a":1, "a":2}</code> is <code>{"a":2}</code>.</div></div>

<div class="qa-block"><div class="qa-q"><span class="q-num">11</span> How do you merge two dicts?</div>
  <div class="qa-a"><span class="ans-label">A</span> <code>a | b</code> (3.9+, right wins), <code>a |= b</code> (in place), or <code>{**a, **b}</code> (older). Or <code>a.update(b)</code> to mutate a.</div></div>

<div class="qa-block"><div class="qa-q"><span class="q-num">12</span> Why can't you modify a dict while iterating it?</div>
  <div class="qa-a"><span class="ans-label">A</span> Changing its size mid-iteration raises <code>RuntimeError</code> (dicts detect the change). Iterate <code>list(d)</code> or build a new dict with a comprehension.</div></div>

<div class="part-header"><h2>1. Theory</h2></div>

<h3 class="sub">🔹 A string is an immutable sequence</h3>

<div class="theory-box">
  <p style="margin:0 0 10px 0">A <code>str</code> is an ordered sequence of characters you can index and slice like a list (2A) — but you <strong style="color:#89b4fa">cannot change it in place</strong>, exactly like a tuple (2B). Any "edit" builds a <strong>new</strong> string object and rebinds your name to it.</p>
  <p style="margin:0">This one fact drives the whole session: methods, formatting, and the big performance trap all fall out of immutability.</p>
</div>

<div class="info-box">💡 <strong>SQL / Power BI anchor:</strong> like a text value in a column — you don't mutate a cell's characters, you compute a new value and overwrite. <code>REPLACE()</code> / DAX <code>SUBSTITUTE()</code> return a new string; they never edit the original in place.</div>

<div class="code-block">
s = <span class="cs">"hello"</span><br>
s[<span class="cn">0</span>] = <span class="cs">"H"</span>        <span class="cc"># attempt to edit in place</span><br>
s = <span class="cs">"H"</span> + s[<span class="cn">1</span>:]    <span class="cc"># rebuild -> a brand-new object</span>
</div>
<div class="output-block">
TypeError: 'str' object does not support item assignment<br>
>>> s<br>
'Hello'
</div>

<div class="mut-grid">
  <div class="mut-card mc-imm">
    <div class="mc-title">🔒 Immutable (str, like tuple)</div>
    <div class="mc-body">
      indexable, sliceable<br>
      hashable → dict key / set member<br>
      every "change" = new object
    </div>
  </div>
  <div class="mut-card mc-mut">
    <div class="mc-title">🔓 Build with list, then join</div>
    <div class="mc-body">
      parts = []<br>
      parts.append(chunk)<br>
      "".join(parts)  <span class="cc"># O(n)</span>
    </div>
  </div>
</div>

<div class="why-box">
  <strong>Why it matters (callback to 2A):</strong> because a string can't grow in place, <code>s += chunk</code> inside a loop rebuilds the entire string every iteration — the O(n²) accumulator trap from Lists. The fix is the same shape: collect in a <code>list</code>, then <code>"".join(parts)</code> once — O(n).
</div>

<div class="note-box">💡 Immutability is also why strings are <strong>hashable</strong> — that's what lets them be dict keys and set members (2C/2D). Same immutability → hashability chain as tuples.</div>
