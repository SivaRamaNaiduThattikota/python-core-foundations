### Exercises — Solutions

```python
# E1 (Easy) — Squares of even numbers 0-20
print([x*x for x in range(21) if x % 2 == 0])
```

```python
# E2 (Easy) — Dict comp n -> n**2 for 1..5
print({n: n**2 for n in range(1, 6)})
```

```python
# E3 (Easy) — Set of unique word lengths
words = ["a", "bb", "cc", "ddd"]
print({len(w) for w in words})
```

```python
# E4 (Easy) — Flatten a 2D list (outer for first)
m = [[1, 2], [3, 4], [5]]
print([x for row in m for x in row])
```

```python
# E5 (Medium) — Filter dict comp: keep scores >= 50
pairs = [("Alice", 88), ("Bob", 40), ("Cara", 50)]
print({name: s for name, s in pairs if s >= 50})
```

```python
# E6 (Medium) — Invert a dict (unique values)
d = {"a": 1, "b": 2}
print({v: k for k, v in d.items()})   # lossy if values repeat
```

```python
# E7 (Medium) — Transpose a matrix (nested comprehension)
m = [[1, 2, 3], [4, 5, 6]]
cols = len(m[0])
print([[row[i] for row in m] for i in range(cols)])
```

```python
# E8 (Medium) — Sum of squares of odd numbers 1-100 (genexpr, no list built)
print(sum(x*x for x in range(1, 101) if x % 2))
```

```python
# E9 (Medium) — Any word longer than 10 chars? (short-circuits)
words = ["short", "a_very_long_word", "ok"]
print(any(len(w) > 10 for w in words))
```

```python
# E10 (Hard) — Fibonacci generator
from itertools import islice
def fib():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

print(list(islice(fib(), 10)))
```

```python
# E11 (Hard) — Streaming pipeline: total length of non-empty lines
lines = ["  hi ", "", "  bye"]
print(sum(len(ln.strip()) for ln in lines if ln.strip()))
```

```python
# E12 (Hard) — Recursively flatten arbitrary nesting (generator)
def flatten(nested):
    for x in nested:
        if isinstance(x, list):
            yield from flatten(x)
        else:
            yield x

print(list(flatten([1, [2, [3, [4, 5]]], 6])))
```

### Code Challenges — Solutions

```python
# C1 (Easy) — List of the first n squares
def squares(n):
    return [x*x for x in range(n)]

print(squares(5))
```

```python
# C2 (Easy) — Sum of even numbers (genexpr)
def sum_evens(nums):
    return sum(x for x in nums if x % 2 == 0)

print(sum_evens([1, 2, 3, 4, 5, 6]))
```

```python
# C3 (Medium) — Invert a dict
def invert(d):
    return {v: k for k, v in d.items()}

print(invert({"a": 1, "b": 2, "c": 3}))
```

```python
# C4 (Medium) — Batch a sequence into chunks of size n (ML batching / DataLoader)
def chunk(seq, n):
    for i in range(0, len(seq), n):
        yield seq[i:i+n]

print(list(chunk([1, 2, 3, 4, 5, 6, 7], 3)))   # [[1,2,3],[4,5,6],[7]]
```

```python
# C5 (Medium) — Fibonacci generator
from itertools import islice
def fib():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

print(list(islice(fib(), 8)))
```

```python
# C6 (Medium) — Cumulative sum generator (itertools.accumulate)
def cumsum(nums):
    total = 0
    for n in nums:
        total += n
        yield total

print(list(cumsum([1, 2, 3, 4])))
```

```python
# C7 (Hard) — Sliding-window generator
def windows(seq, k):
    for i in range(len(seq) - k + 1):
        yield seq[i:i+k]

print(list(windows("abcde", 3)))
```

```python
# C8 (Hard) — Infinite prime generator + first N
from itertools import islice
def primes():
    found, n = [], 2
    while True:
        if all(n % p for p in found):   # short-circuiting genexpr
            found.append(n)
            yield n
        n += 1

print(list(islice(primes(), 6)))
```
